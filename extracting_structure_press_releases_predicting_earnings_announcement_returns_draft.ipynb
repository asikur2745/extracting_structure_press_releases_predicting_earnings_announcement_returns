## Introduction

This project provides a Python implementation of the methodologies presented in the 2025 paper "Extracting the Structure of Press Releases for Predicting Earnings Announcement Returns." The core of this repository is the iPython Notebook `extracting_structure_press_releases_predicting_earnings_announcement_returns_draft.ipynb`, which contains a comprehensive suite of functions to replicate the paper's findings, from initial data validation to the final generation of all analytical tables and figures.

The paper investigates the predictive power of textual "soft information" in corporate earnings press releases relative to numerical "hard information" (earnings surprise). This codebase operationalizes the paper's framework, allowing users to:
-   Rigorously validate and manage the entire experimental configuration.
-   Process and clean raw HTML press releases from SEC filings.
-   Generate five distinct sets of textual features using classical and deep learning NLP models.
-   Train predictive models in a rolling-window framework to prevent look-ahead bias.
-   Perform a full suite of econometric, interpretability, and simulation analyses to replicate the paper's key tables and figures.

## Theoretical Background

The implemented methods are grounded in asset pricing, econometrics, and natural language processing.

**1. Hard vs. Soft Information:**
The study tests the relative importance of two information types:
-   **Hard Information (Earnings Surprise):** The quantitative surprise, defined as the deviation of reported earnings from analyst expectations, scaled by price.
    $$
    \text{Surprise}_{c,t} = \frac{\text{EPS}_{c,\tau} - E_{\tau-1}[\text{EPS}_{c,t}]}{P_{c,\tau-5}}
    $$
-   **Soft Information (Textual Content):** The qualitative narrative content of the press release, captured by various NLP models.

**2. Rolling-Window Lasso Regression:**
To convert high-dimensional text features into a single predictive "soft score" without look-ahead bias, a rolling-window estimation is used. For each year `t`, a Lasso regression is trained on year `t`'s data to learn a mapping from text features to returns. This model is then used to predict out-of-sample scores for year `t+1`.
$$
\hat{\mathbf{w}}_t = \arg\min_{\mathbf{w}} \left\{ \frac{1}{2N_{t}} \|\mathbf{X}_{t}\mathbf{w} - \mathbf{y}_{t}\|_2^2 + \lambda \|\mathbf{w}\|_1 \right\} \quad \implies \quad \text{SoftScore}_{t+1} = \mathbf{X}_{t+1} \cdot \hat{\mathbf{w}}_t
$$

**3. Panel Data Regression with Clustered Standard Errors:**
To assess the explanatory power of the signals, a cross-sectional regression is estimated with standard errors clustered by both firm (`permno`) and time (`ann_trade_date`). This is the standard in financial econometrics for addressing potential correlation in residuals across both dimensions.
$$
\text{Ret}_{c,\tau} = \alpha + \beta_0 \text{Surprise}_{c,t} + \beta_1 \text{Soft}_{c,t} + \epsilon_{c,\tau}
$$

## Features

The provided iPython Notebook (`extracting_structure_press_releases_predicting_earnings_announcement_returns_draft.ipynb`) implements the full research pipeline, including:

-   **Modular, Multi-Task Architecture:** The entire pipeline is broken down into 23 distinct, modular tasks, each with its own orchestrator function.
-   **Configuration-Driven Design:** All study parameters are managed in an external `config.yaml` file, allowing for easy customization and replication.
-   **Multi-Modal NLP Feature Engineering:** Complete pipeline for generating features from five different models: BKMX, Online LDA (oLDA), BERT, FinBERT, and MPNET.
-   **Econometrically Sound Modeling:** Implements rolling-window estimation to prevent look-ahead bias and uses two-way clustered standard errors for robust inference.
-   **Advanced Interpretability Suite:** Includes SHAP analysis for feature importance, and a full LLM-based pipeline (using GPT-4o) for topic labeling, taxonomy creation, and token-level attribution.
-   **Realistic Trading Simulations:** Implements a market efficiency test and a "hacking scenario" analysis with careful handling of transaction costs and market microstructure details.
-   **Automated Validation:** Concludes with a comprehensive validation step that programmatically compares all generated results against the key numerical findings reported in the source paper.

## Methodology Implemented

The core analytical steps directly implement the methodology from the paper:

1.  **Validation & Filtering (Tasks 1-2):** Ingests and validates the `config.yaml` and raw data, then applies the paper's sample selection criteria.
2.  **Text Cleaning (Task 3):** Processes raw HTML into two standardized text formats for different model types.
3.  **Vectorization (Tasks 4-6):** Generates all five sets of textual features (BKMX, oLDA, BERT, FinBERT, MPNET).
4.  **Predictive Modeling (Tasks 7-11):** Runs the rolling Lasso estimation to generate soft scores, prepares final signals, runs baseline and combined regressions, and computes SHAP importance.
5.  **Interpretability (Tasks 12-17):** Executes the full LLM-based pipeline to understand the thematic content of the oLDA and BERT-family models.
6.  **Simulations (Tasks 18-21):** Builds the tools for and executes the market efficiency and hacking scenario simulations.
7.  **Diagnostics & Final Validation (Tasks 22-23):** Computes summary statistics and visualizations, and runs a final, automated check of all results against the paper's benchmarks.

All functions are self-contained, fully documented with type hints and docstrings, and designed for professional-grade execution.

## Key Callable: `run_full_analysis_pipeline`

The project is designed around a single, top-level user-facing interface function:

-   **`run_full_analysis_pipeline`:** This master orchestrator function, located in the final section of the notebook, runs the entire automated research pipeline from end-to-end. A single call to this function reproduces the entire computational portion of the project, from data validation to the final report.

## Usage

The `extracting_structure_press_releases_predicting_earnings_announcement_returns_draft.ipynb` notebook provides a complete, step-by-step guide. The primary workflow is to execute the final cell of the notebook, which calls the top-level `run_full_analysis_pipeline` orchestrator:

```python
# Final cell of the notebook

# Define paths and load data/config
DATA_PATH = Path('path/to/your/data.parquet')
CONFIG_PATH = Path('config.yaml')
ARTIFACTS_PATH = Path('./study_artifacts')

# Load data and configuration
events_df = pd.read_parquet(DATA_PATH)
raw_events_df = events_df.copy()
with open(CONFIG_PATH, 'r') as f:
    study_params = yaml.safe_load(f)

# Run the entire study
final_results = run_full_analysis_pipeline(
    events_df=events_df,
    raw_events_df=raw_events_df,
    study_params=study_params,
    artifacts_path=ARTIFACTS_PATH
)

# The `final_results` dictionary will contain all key outputs.
```

# Paper

Title: "*Extracting the Structure of Press Releases for Predicting Earnings Announcement Returns*"

Authors: Yuntao Wu, Ege Mert Akin, Charles Martineau, Vincent Grégoire, Andreas Veneris

E-Journal Submission Date: 29 September 2025 (V1) and 4 October 2025 (V2)

Conference Affiliation: 6th ACM International Conference on AI in Finance

Link: https://www.arxiv.org/abs/2509.24254

Abstract:

We examine how textual features in earnings press releases predict stock returns on earnings announcement days. Using over 138,000 press releases from 2005 to 2023, we compare traditional bag-of-words and BERT-based embeddings. We find that press release content (soft information) is as informative as earnings surprise (hard information), with FinBERT yielding the highest predictive power. Combining models enhances explanatory strength and interpretability of the content of press releases. Stock prices fully reflect the content of press releases at market open. If press releases are leaked, it offers predictive advantage. Topic analysis reveals self-serving bias in managerial narratives. Our framework supports real-time return prediction through the integration of online learning, provides interpretability and reveals the nuanced role of language in price formation.


# Summary

### **The Core Research Question and Contribution**

The fundamental question the authors address is a classic one in finance, viewed through a modern lens: **How much of a stock's price movement on an earnings announcement day is driven by the numerical surprise, and how much is driven by the narrative text accompanying it?**

In the parlance of the field, they are quantifying the relative importance of "hard" information (the earnings surprise, i.e., the deviation from analyst consensus) versus "soft" information (the qualitative content of the press release).

Their primary contributions are threefold:
1.  **A Rigorous Model Comparison:** They systematically compare traditional bag-of-words models (LDA) with modern transformer-based models (the BERT family, including the domain-specific FinBERT) on a very large and recent dataset (over 138,000 releases from 2005-2023).
2.  **Methodological Soundness:** They employ a rolling-window estimation framework to avoid look-ahead bias, which is critical for any study claiming predictive power. They also use SHAP (Shapley Additive exPlanations) values to provide a theoretically grounded measure of feature importance, moving beyond simple regression coefficients.
3.  **Direct Economic Tests:** They don't just stop at statistical significance. They test the economic implications of their findings by examining market efficiency and simulating a "hacking" scenario to quantify the value of pre-released information.

### **Data Assembly and Preprocessing**

This is the foundational, and often most challenging, part of any such study.
*   **Data Sources:** They merge data from standard academic sources:
    *   **I/B/E/S:** For analyst forecasts to compute the earnings surprise.
    *   **CRSP/TAQ:** For stock returns (daily and intraday).
    *   **SEC EDGAR:** For the raw HTML text of the 8-K filings containing the press releases.
*   **Defining the Variables:**
    *   **Hard Information (Surprise):** Defined conventionally as the actual reported Earnings Per Share (EPS) minus the median analyst forecast, scaled by the stock price five days prior (Equation 1). This is a standard specification.
    *   **Soft Information (Text):** The core of their work. They perform extensive cleaning on the raw HTML, removing boilerplate sections like "forward-looking statements," tables, and contact information. This is a non-trivial engineering task crucial for reducing noise.
*   **Sample:** The final dataset is large, covering nearly two decades and thousands of unique firms, which lends significant statistical power to their conclusions.

### **The Methodological Framework: From Text to Prediction**

This is where the computer science and econometrics truly merge. The authors construct a pipeline to convert unstructured text into a predictive signal.

1.  **Text Vectorization (Feature Extraction):** They test two families of techniques:
    *   **Bag-of-Words (BoW):** Instead of simple word counts, they use Latent Dirichlet Allocation (LDA). This is a generative probabilistic model that represents each document as a mixture of "topics," where each topic is a distribution over words. They cleverly use an *online LDA (oLDA)*, which allows the model to adapt to new vocabulary and themes over time—a significant improvement over a static model.
    *   **Contextual Embeddings (BERT):** They use pre-trained transformer models (BERT-base, MPNET, and FinBERT) to convert each press release into a 768-dimensional vector. This is done by taking the mean of the hidden states of the first 512 tokens. The key advantage here is that BERT captures context and semantic relationships, unlike BoW. FinBERT, being fine-tuned on financial text, is hypothesized to be the superior model.

2.  **Predictive Modeling:** They use a two-stage approach:
    *   **Stage 1 (Signal Construction):** For each NLP model, they train a Lasso regression on a rolling annual basis to predict announcement returns (`Ret_c,t`) from the text vectors (`X_c,t`). The fitted prediction from this model becomes their "Soft" information score for the following year. This rolling design is excellent practice, as it ensures the model is only using past information to make predictions.
    *   **Stage 2 (Horse Race):** They run a simple cross-sectional regression (Table 1) to see how well the `Surprise` (hard info) and the `Soft` score (soft info) explain returns.

3.  **Interpretability:**
    *   **SHAP Values:** They use SHAP to decompose the prediction of the Stage 2 model, attributing the final return prediction to the hard and soft components. This provides a more robust measure of importance than R-squared contributions.
    *   **Topic Analysis:** For the LDA models, interpretability is built-in via the topics. For the BERT models, they perform a clever mapping: they use the learned Lasso weights to identify which of the 768 embedding dimensions are important, and then trace these back to the original tokens. They use GPT-4 to label and categorize these influential tokens, revealing the underlying themes driving the predictions.

### **Key Empirical Findings**

The results are clear and compelling.

*   **Soft Information is as Important as Hard Information:** The headline result from **Table 1** is that the `Soft` score derived from FinBERT has a SHAP value of 52.12%, slightly *exceeding* the 47.88% for the numerical earnings surprise. This is a powerful statement: the market reacts to the narrative just as much as, if not more than, the headline number. The combined model achieves an R² of 4.4%, which is highly respectable for return prediction at this frequency.
*   **FinBERT is the Superior Model:** Across all specifications, FinBERT provides the most potent `Soft` signal. This validates the hypothesis that domain-specific language models are superior for tasks in specialized fields like finance.
*   **Managers Exhibit Self-Serving Bias:** The topic analysis (**Figure 4, Table 4**) provides a fascinating behavioral insight. When results are good (positive returns), the influential text focuses on internal factors like "Financial Performance" and "Financial Metrics." When results are bad (negative returns), the narrative shifts to blame external factors like "Market and Economic Factors" and "Sector-Specific News." This is a data-driven confirmation of a well-known psychological bias.
*   **Information is Not Leaked (On Average):** In **Table 3**, they show that while their `Soft` score predicts returns on the announcement day, it has no predictive power for returns on the *prior* day. This suggests that, in general, the content of press releases is not systematically leaking into the market before the official release.

### **Economic Applications and Implications**

The authors translate their statistical findings into two powerful economic tests.

1.  **Market Efficiency Test:** They construct a long-short trading strategy based on agreement between the hard (surprise) and soft (SoftMean) signals. They enter trades at 9:45 AM on the announcement day and exit at the close. The result: the strategy produces a statistically significant *negative* alpha after accounting for transaction costs (bid-ask spread). **This is strong evidence for market efficiency.** The information, both hard and soft, is fully incorporated into the price by the time the market is orderly (i.e., after the opening auction), leaving no profit for the taking.

2.  **The "Hacking" Simulation:** This is a clever counterfactual. They ask: "What if you had this information *before* the release?" Using Precision@k as their metric, they show that an informed trader could identify the top 10 best-performing stocks with ~52% accuracy when hard and soft signals agree (**Table 6**). This is a massive improvement over the ~35% accuracy when using only the surprise. This elegantly quantifies the immense economic value of this "insider" information and highlights the critical importance of cybersecurity for newswire services.

### **Overall Assessment**

*   **Strengths:**
    *   **Comprehensive and Rigorous:** The use of a large, modern dataset, careful methodology, and robust economic tests makes the conclusions highly credible.
    *   **High Relevance:** It addresses a central question in finance with state-of-the-art tools.
    *   **Excellent Interpretation:** The combination of SHAP values and GPT-4-assisted topic analysis provides a deep, intuitive understanding of *why* the models work, moving beyond a "black box" approach.

*   **Potential Extensions and Minor Critiques:**
    *   The mapping of BERT weights back to tokens is an intelligent heuristic, but it's still an approximation. More direct methods for transformer interpretability, like analyzing attention heads or using integrated gradients, could provide further insight.
    *   The paper focuses on the press release in isolation. The information discovery process is dynamic. A natural next step, which the authors acknowledge, is to integrate the text and audio of the subsequent conference call to see how the market narrative evolves and how price discovery continues.
    *   While they test for efficiency at the market open, there could be a more granular story in the first few minutes of trading. An analysis using millisecond-level data could explore the microstructure of how this information is impounded.

In conclusion, the authors provide compelling, robust evidence that the narrative content of earnings press releases is a first-order determinant of stock returns, with predictive power on par with the headline earnings number. Their work underscores the increasing efficiency of financial markets while simultaneously quantifying the enormous value of non-public information. A very solid contribution.

## Draft 1


### **Documentation of Final Implemented Callables**
#### **Task 1: `validate_inputs_and_schema` (Orchestrator)**

*   **Inputs:** A raw `pandas.DataFrame` (`events_df`) and a `dict` (`study_params`) containing the project configuration.
*   **Processes:**
    1.  Calls `_validate_configuration` to parse and validate the `study_params` dictionary against a comprehensive Pydantic schema, ensuring all parameters are present, correctly typed, and logically consistent.
    2.  Calls `_validate_dataframe_schema` to exhaustively check the input DataFrame's index structure and the presence and `dtype` of all required columns against the project's data dictionary.
    3.  Calls `_validate_dataframe_consistency` to perform vectorized checks on the mathematical and logical integrity of the data (e.g., verifying that `ret_announce_cc` is correctly calculated from prices).
*   **Outputs:** A tuple containing the validated (but unmodified) `DataFrame` and a validated, type-coerced `StudyConfig` object.
*   **Transformation:** This function is a pure validation gateway. It does not transform data but rather inspects it and will raise a `ValueError` if any check fails, halting the pipeline before corrupt or invalid data can be processed.
*   **Role in Research Pipeline:** This function serves as the rigorous, mandatory first step of the entire pipeline. It ensures the reproducibility and integrity of the analysis by guaranteeing that the inputs—both data and parameters—are in a known, valid state before any computation begins. It is the foundational guardrail for the entire project.



#### **Task 2: `apply_sample_filters` (Orchestrator)**

*   **Inputs:** A `pandas.DataFrame` and a validated `StudyConfig` object.
*   **Processes:**
    1.  Calls `_filter_by_release_session` to remove events that do not occur after market hours, as per the study's focus on overnight news.
    2.  Calls `_filter_by_coverage_and_length` to remove events without analyst coverage and press releases that are too short or too long.
    3.  Calls the remediated `_filter_by_language_duplicates_and_date` to restrict the sample to English-language documents, the specified date range (2005-2023), and to remove duplicate filings for the same event using the `groupby().apply()` method.
*   **Outputs:** A tuple containing a new, smaller `pandas.DataFrame` representing the final analytical sample, and a `dict` (`filter_report`) detailing the number of observations removed at each stage.
*   **Transformation:** This function transforms the raw dataset into the final analytical sample by applying a sequence of row-filtering operations. The number of rows is reduced from the initial count to the final sample size of approximately 138,000 events.
*   **Role in Research Pipeline:** This function directly implements the sample selection criteria described in **Section 3 (Data)** of the paper. It ensures that the analysis is performed on the exact subset of data that meets the study's methodological requirements, such as having analyst coverage to compute the earnings surprise and being an after-hours announcement.



#### **Task 3: `cleanse_and_standardize_text_data` (Remediated Orchestrator)**

*   **Inputs:** A `pandas.DataFrame` containing the `press_release_html_raw` column and a validated `StudyConfig` object.
*   **Processes:**
    1.  Calls `_extract_text_from_html` to parse the raw HTML, strip all `<table>` elements, and extract the raw text content.
    2.  Calls `_remove_boilerplate_and_patterns` to apply a series of regular expressions that remove uninformative content like forward-looking statements, contact information, and URLs.
    3.  Calls `_normalize_text` to perform Unicode normalization, lowercasing, and whitespace collapsing, producing the `press_release_text_clean` column.
    4.  Calls the new helper `_process_text_for_bow` on the `press_release_text_clean` column. This helper tokenizes, removes stop words, and lemmatizes the text, producing the `press_release_text_bow` column.
*   **Outputs:** The input `DataFrame` augmented with four new columns: `press_release_text_clean`, `doc_length_clean`, `press_release_text_bow`, and `doc_length_bow`.
*   **Transformation:** This function transforms the unstructured `press_release_html_raw` column into two distinct, structured text representations suitable for different types of NLP models.
*   **Role in Research Pipeline:** This function implements the "text preprocessing steps" mentioned in **Section 3 (Data)**. It is a critical ETL (Extract, Transform, Load) step that prepares the raw textual data for all subsequent vectorization and modeling tasks.



#### **Task 4: `vectorize_text_with_bkmx` (Orchestrator)**

*   **Inputs:** A `pandas.DataFrame` with the `press_release_text_clean` column and a validated `StudyConfig` object.
*   **Processes:**
    1.  Calls `_load_and_process_bkmx_taxonomy` to load the predefined financial topic taxonomy from a JSON file and convert it into efficient lookup structures.
    2.  Calls the remediated `_compute_bkmx_topic_intensities` to efficiently count the occurrences of all BKMX phrases in each document using a vectorized approach, producing a 180-dimensional topic intensity vector for each document.
    3.  Calls `_aggregate_to_metatopics` to sum the intensities of the 180 topics into the 11 selected metatopics.
*   **Outputs:** The input `DataFrame` augmented with 11 new columns, each representing the intensity of a specific BKMX metatopic (e.g., `bkmx_Corporate_Earnings`).
*   **Transformation:** This function transforms the cleaned text strings into a low-dimensional (11-D) numerical representation based on a fixed, external dictionary of financial themes.
*   **Role in Research Pipeline:** This function implements one of the baseline "bag-of-words" approaches mentioned in **Section 4.1 (Bag-of-Words)**. It provides a simple, interpretable feature set based on the established BKMX taxonomy.



#### **Task 5: `vectorize_text_with_olda_rolling` (Remediated Orchestrator)**

*   **Inputs:** A `pandas.DataFrame` with the `press_release_text_bow` column and a validated `StudyConfig` object.
*   **Processes:**
    1.  Iterates through a rolling window of years (2006-2023).
    2.  In each iteration `t`, it trains a `CountVectorizer` and an `OnlineLDA` model on all data up to year `t-1`.
    3.  It then uses the fitted model to `.transform()` the documents from year `t`, generating a 50-dimensional document-topic probability distribution for each. This distribution is a direct implementation of the topic attention vector concept from Equation (2): $f_{i,a} = \frac{\sum_{j=1}^{N_a} \mathbf{1}(z_{j,a} = i)}{\sum_{q=1}^{K} \sum_{j=1}^{N_a} \mathbf{1}(z_{j,a} = q)}$.
*   **Outputs:** A tuple containing: (1) the input `DataFrame` augmented with 50 new columns (`olda_topic_0` to `olda_topic_49`), (2) the final, most comprehensively trained `LatentDirichletAllocation` model object, and (3) its corresponding `CountVectorizer` object.
*   **Transformation:** This function transforms the preprocessed BoW text strings into a 50-dimensional numerical representation based on topics learned *dynamically* from the corpus itself, while rigorously avoiding look-ahead bias.
*   **Role in Research Pipeline:** This function implements the more advanced, adaptive topic modeling approach (oLDA) described in **Section 4.1 (Bag-of-Words)**. The rolling protocol is essential for ensuring the out-of-sample validity of the generated features.



#### **Task 6: `extract_transformer_embeddings` (Remediated Orchestrator)**

*   **Inputs:** A `pandas.DataFrame` with the `press_release_text_clean` column, a `StudyConfig` object, and an `artifacts_path`.
*   **Processes:**
    1.  Iterates through the configured Transformer models (BERT, FinBERT, MPNET).
    2.  For each model, it loads the pre-trained model and tokenizer.
    3.  It processes the text in batches, performing a forward pass to get the `last_hidden_state` tensor for each document.
    4.  It computes the document-level embedding by mean-pooling the token embeddings: $\mathbf{e}_{\text{doc}} = \frac{1}{T} \sum_{t=1}^{T} \mathbf{h}_t$.
    5.  It saves the full `(512, 768)` `last_hidden_state` tensor for each document to disk for later use.
*   **Outputs:** The input `DataFrame` augmented with 768 new columns for each of the Transformer models (e.g., `finbert_0` to `finbert_767`).
*   **Transformation:** This function transforms the cleaned text strings into very high-dimensional (768-D), dense numerical representations that capture deep contextual and semantic meaning.
*   **Role in Research Pipeline:** This function implements the modern NLP approach described in **Section 4.2 (BERT Models)**. It generates the most powerful, but least interpretable, features used in the study.



#### **Task 7: `estimate_rolling_soft_scores` (Remediated Orchestrator)**

*   **Inputs:** A `pandas.DataFrame` containing all previously generated features, a `StudyConfig` object, and an `artifacts_path`.
*   **Processes:**
    1.  Iterates through each of the five vectorization methods (BKMX, oLDA, BERT, etc.).
    2.  For each method, it runs a rolling window loop from 2005 to 2022.
    3.  In each iteration `t`, it trains a `Lasso` regression model to predict announcement returns (`ret_announce_cc`) using the features from year `t`. This is a direct implementation of the Lasso objective function from Equation (3): $\arg \min_{\mathbf{w}} \frac{1}{2N} \|\mathbf{X}\mathbf{w} - \mathbf{y}\|_2^2 + \lambda \|\mathbf{w}\|_1$.
    4.  It saves the fitted Lasso model to disk.
    5.  It uses the fitted model to predict returns on the data from year `t+1`, creating the out-of-sample "soft score".
*   **Outputs:** A tuple containing: (1) the input `DataFrame` augmented with five new `soft_*` score columns, and (2) a nested dictionary of the fitted Lasso coefficient vectors for each model and year.
*   **Transformation:** This function transforms the high-dimensional feature sets for each method into a single, univariate predictive score (`Soft_{c,t}`), while rigorously avoiding look-ahead bias.
*   **Role in Research Pipeline:** This is the heart of the predictive modeling, described in **Section 5.1 (Return Score Construction)**. It is the crucial step that converts the raw text features into a single, powerful predictor of stock returns.



#### **Task 8: `prepare_final_signals` (Remediated Orchestrator)**

*   **Inputs:** A `pandas.DataFrame` containing the five `soft_*` scores and the raw `surprise` column, and a `StudyConfig` object.
*   **Processes:**
    1.  Calls `_aggregate_soft_signals` to compute the `soft_mean` column by averaging the five individual soft scores.
    2.  Calls `_winsorize_earnings_surprise` to cap the raw `surprise` column at the 1st and 99th percentiles, creating the `surprise_winsorized` column.
    3.  Calls `_validate_signal_distributions` to log descriptive statistics and correlations of the final signals as a sanity check.
*   **Outputs:** The input `DataFrame` augmented with the `soft_mean` and `surprise_winsorized` columns.
*   **Transformation:** This function performs the final data preparation steps on the key predictor variables before they are used in the main econometric analysis.
*   **Role in Research Pipeline:** This function prepares the final variables needed for the regressions in **Section 5.1**, specifically for Equation (4): $\text{Ret}_{c,\tau} = \alpha + \beta_0 \text{Surprise}_{c,t} + \beta_1 \text{Soft}_{c,t} + \epsilon$.



#### **Task 9: `estimate_baseline_regressions` (Remediated Orchestrator)**

*   **Inputs:** The fully prepared `pandas.DataFrame` and a `StudyConfig` object.
*   **Processes:**
    1.  Iterates through each of the five soft signal models.
    2.  For each model, it prepares the regression sample and estimates the baseline OLS regression using `linearmodels.PanelOLS` with two-way clustered standard errors (by firm and time). This is a direct implementation of Equation (4): $\text{Ret}_{c,\tau} = \alpha + \beta_0 \text{Surprise}_{c,t}^{\text{wins}} + \beta_1 \text{Soft}_{c,t}^{I} + \epsilon$.
*   **Outputs:** A tuple containing: (1) a dictionary of formatted regression summary statistics for each model, and (2) a dictionary of the fitted `PanelResults` objects themselves.
*   **Transformation:** This function transforms the prepared data into a set of statistical estimates (coefficients, standard errors, R-squared), quantifying the relationship between the signals and returns.
*   **Role in Research Pipeline:** This function generates the primary empirical results of the paper, corresponding to the analysis presented in **Table 1**.



#### **Task 10: `estimate_combined_regressions` (Orchestrator)**

*   **Inputs:** The fully prepared `pandas.DataFrame` and a `StudyConfig` object.
*   **Processes:**
    1.  Estimates the "joint" regression model, which includes the surprise and all five soft signals as predictors, again using `PanelOLS` with two-way clustering. This corresponds to Equation (6) in the paper's notation.
    2.  Estimates the "composite" model, which uses the `soft_mean` signal.
    3.  Computes Variance Inflation Factors (VIF) for the joint model to diagnose multicollinearity.
*   **Outputs:** A dictionary containing the detailed regression results for both the "joint" and "composite" models.
*   **Transformation:** This function transforms the data into statistical estimates for the "horse race" regressions.
*   **Role in Research Pipeline:** This function generates the results for the comparative analysis shown in **Table 2**, allowing for an assessment of the marginal contribution of each soft signal.



#### **Task 11: `compute_shap_importance_for_baselines` (Remediated Orchestrator)**

*   **Inputs:** The `pandas.DataFrame`, the dictionary of `fitted_baseline_models` from Task 9, and a `StudyConfig` object.
*   **Processes:**
    1.  Iterates through the pre-fitted baseline regression models.
    2.  For each model, it instantiates a `shap.LinearExplainer` using the model's coefficients.
    3.  It computes the exact Shapley values for each feature for every observation.
    4.  It aggregates these to find the mean absolute SHAP value for each feature, implementing Equation (5): $I_j = \frac{1}{N} \sum_{i=1}^{N} |\phi_{i,j}|$.
    5.  It normalizes these values to percentages: $\hat{I}_j = \frac{I_j}{\sum_{k} I_k} \times 100$.
*   **Outputs:** A dictionary mapping each model name to a dictionary of its features' normalized importance percentages.
*   **Transformation:** This function transforms the fitted linear models into a robust, theoretically-grounded measure of global feature importance.
*   **Role in Research Pipeline:** This function generates the crucial feature importance results discussed in **Section 5.1** and reported in **Table 1**, providing the basis for the claim that soft information is as important as hard information.



#### **Task 12-17 (Interpretability Pipeline)**

*   **`label_lda_topics_with_llm`:** Transforms the numerical topic-word distributions from the final oLDA model into human-readable labels and definitions using GPT-4o.
*   **`condense_topics_into_metatopics`:** Transforms the 50 labeled topics into a smaller, coherent, and mutually exclusive taxonomy of 9 metatopics, again using GPT-4o.
*   **`extract_influential_tokens` (Remediated):** Transforms the opaque Lasso weights and token embeddings into a concrete list of the top 5 positive and top 5 negative words for each document by implementing the equation $\mathbf{IS} = \mathbf{E} \cdot \hat{\mathbf{w}}$.
*   **`classify_influential_tokens`:** Transforms the thousands of unique influential tokens into a simple mapping from each token to one of the 9 metatopic IDs, using GPT-4o as a classifier.
*   **`compute_olda_metatopic_analytics`:** Transforms the time series of oLDA Lasso weights and topic attentions into two summary DataFrames showing the evolution of metatopic polarity ($w_M$) and explained variance ($h(M)$) over time.
*   **`analyze_token_polarity_by_metatopic` (Remediated):** Transforms the document-level influential token data and the token-metatopic map into a final summary table that cross-tabulates token counts by metatopic and polarity.
*   **Role in Research Pipeline:** This entire suite of functions implements the deep interpretability analyses described in **Section 5.2 (Topic Analysis)**. They provide the evidence for the "self-serving bias" hypothesis and generate the results shown in **Figure 4** and **Table 4**.



#### **Task 18-21 (Simulation Pipeline)**

*   **`create_trading_simulation_orchestrator` & `create_hacking_scenario_orchestrator`:** These are factory functions that produce the tools for the simulations. They do not transform data themselves.
*   **`execute_market_efficiency_simulation`:** Transforms the full event DataFrame into a daily time series of long-short portfolio returns by simulating a realistic trading strategy. It then transforms this return series into a set of alpha estimates by regressing it on an intercept and Fama-French factors, implementing Equations (9) and (10): $\text{LS}_t = \alpha$ and $\text{LS}_t = \alpha + \beta_0 (\text{Mkt} - \text{rf})_t + \dots$.
*   **`execute_hacking_scenario_evaluation` (Remediated):** Transforms a filtered subset of the event data into a summary table of Precision@k scores. It partitions the data into winners and losers and then, for various predictors, calculates the hit rate of identifying the top performers using the equation $P@k = \frac{C}{k}$.
*   **Role in Research Pipeline:** These functions implement the two main applications of the framework. Task 19 executes the market efficiency test described in **Section 6.1**, while Task 21 executes the "hacking scenario" analysis from **Section 6.2**, generating the results for **Tables 5 and 6**.



#### **Task 22-23 (Diagnostics and Validation)**

*   **`compute_diagnostic_statistics`:** Transforms the full DataFrame into three key diagnostic outputs: a yearly summary table (for Figure 1), a dictionary of cosine similarities between embeddings, and a 5x5 pivot table of mean returns (for Figure 2).
*   **`validate_all_study_findings`:** This is a meta-analytical function. It takes all the final result artifacts from the entire pipeline as input and transforms them into a single, human-readable validation report string, comparing each computed metric against its benchmark from the paper.
*   **Role in Research Pipeline:** These functions provide the essential sanity checks and final verification for the entire project, ensuring the results are both internally consistent and externally valid with respect to the source paper.

```python
for flag in ['cleaning_removed_tables', 'cleaning_removed_forward_looking', 'cleaning_removed_non_gaap',

             'cleaning_lemmatized', 'cleaning_stopwords_removed']:
    events_df[flag] = False
    
# Add placeholder columns for text that will be generated.
events_df['press_release_text_clean'] = ''
events_df['doc_length_raw'] = events_df['press_release_html_raw'].str.len()
events_df['doc_length_clean'] = 0

# Create the pristine raw copy *before* any filtering.
raw_events_df = events_df.copy()

print(f"Successfully created synthetic 'events_df' and 'raw_events_df' with {len(events_df)} events.")
```

````python
# Task 1: Validate Configuration Parameters and Input Schema Integrity

# ==============================================================================
# Task 1, Step 1: Configuration Schema and Validation
# ==============================================================================
# ==============================================================================
# Section 1.1: Data Acquisition and Filtering Parameters
# ==============================================================================
class DataParams(BaseModel):
    """
    Defines the schema for data acquisition and filtering parameters used in the
    initial sample construction phase of the earnings press release analysis.

    This model validates parameters that control:
    - Temporal scope of the study (start_date, end_date)
    - Document filtering criteria (length, language, form types)
    - After-hours release filtering
    - Analyst coverage requirements
    - Duplicate resolution strategies

    All parameters are validated for type consistency and logical constraints
    to ensure reproducible sample construction following the methodology
    described in the research paper.

    Attributes:
        start_date: ISO 8601 formatted start date for study period
        end_date: ISO 8601 formatted end date for study period
        timezone: IANA timezone identifier for timestamp operations
        after_hours_filter: Flag to retain only after-hours releases
        accepted_form_types: List of SEC form types to include
        accepted_exhibit_prefixes: List of exhibit prefixes to include
        forecast_window_days: Days for analyst forecast window (typically 90)
        min_document_chars: Minimum character count for document inclusion
        max_document_chars: Maximum character count for document inclusion
        price_adjustments: Convention for stock price adjustments
        duplicate_event_resolution: Strategy for handling duplicate events
        require_analyst_coverage: Flag to require analyst forecast coverage
        allow_regular_hours: Flag to include regular trading hours releases
        keep_language: List of language codes to retain
        missing_timestamp_policy: Strategy for missing timestamp handling
    """

    # The inclusive start date for the study period in ISO 8601 format
    # Used to filter pr_timestamp_et >= start_date
    start_date: str

    # The inclusive end date for the study period in ISO 8601 format
    # Used to filter pr_timestamp_et <= end_date
    end_date: str

    # The IANA timezone identifier for all timestamp operations
    # Ensures consistent timezone handling across all temporal operations
    timezone: str

    # A flag indicating whether to apply the after-hours filter
    # When True, retains only pr_is_after_hours = True events
    after_hours_filter: bool

    # A list of accepted SEC form types (e.g., '8-K', '8-K/A')
    # Used to filter edgar_form_type column
    accepted_form_types: List[str]

    # A list of accepted exhibit prefixes (e.g., '99.')
    # Used to filter pr_exhibit_id column for press release exhibits
    accepted_exhibit_prefixes: List[str]

    # The number of days for the analyst forecast window. Must be positive
    # Defines the window for collecting analyst forecasts before announcement
    forecast_window_days: int = Field(..., gt=0)

    # The minimum character count for a document. Must be positive
    # Documents with doc_length_clean < min_document_chars are excluded
    min_document_chars: int = Field(..., gt=0)

    # The maximum character count for a document. Must be positive
    # Documents with doc_length_clean > max_document_chars are excluded
    max_document_chars: int = Field(..., gt=0)

    # The convention used for price adjustments (e.g., 'CRSP_adjusted')
    # Specifies the adjustment methodology for stock prices
    price_adjustments: str

    # The rule for resolving duplicate events
    # Defines strategy when multiple filings exist for same event
    duplicate_event_resolution: str

    # A flag to enforce the requirement of analyst forecast coverage
    # When True, excludes events with analyst_coverage_count = 0
    require_analyst_coverage: bool

    # A flag to control the inclusion of regular-hours announcements
    # When False, excludes pr_session = 'regular_hours' events
    allow_regular_hours: bool

    # A list of accepted language codes for documents
    # Used to filter doc_language column (e.g., ['en', 'en-US'])
    keep_language: List[str]

    # The policy for handling events with missing press release timestamps
    # Defines fallback strategy when pr_timestamp_et is missing
    missing_timestamp_policy: str

    @root_validator
    def validate_date_range(cls, values: Dict[str, Any]) -> Dict[str, Any]:
        """
        Validates that start_date is not chronologically after end_date.

        This validator ensures temporal consistency in the study period
        specification, preventing logical errors in date range filtering.

        Args:
            values: Dictionary of field values from the model

        Returns:
            Dictionary of validated values

        Raises:
            ValueError: If start_date > end_date chronologically
        """
        # Retrieve start and end date strings from the model's values
        start_str, end_str = values.get('start_date'), values.get('end_date')

        # Proceed only if both values are present
        if start_str and end_str:
            try:
                # Parse date strings to datetime objects for comparison
                start_dt = datetime.datetime.fromisoformat(start_str)
                end_dt = datetime.datetime.fromisoformat(end_str)

                # Compare the parsed datetime objects
                if start_dt > end_dt:
                    # Raise an error if the date range is invalid
                    raise ValueError("'start_date' cannot be after 'end_date'.")
            except ValueError as e:
                # Re-raise with more context if ISO format is invalid
                raise ValueError(f"Invalid date format in date range: {e}")

        # Return the validated values
        return values

    @validator('start_date', 'end_date')
    def validate_iso_date_format(cls, v: str) -> str:
        """
        Validates that date strings conform to ISO 8601 format.

        Args:
            v: Date string to validate

        Returns:
            Validated date string

        Raises:
            ValueError: If date string is not valid ISO 8601 format
        """
        try:
            # Attempt to parse as ISO 8601 format
            datetime.datetime.fromisoformat(v)
        except ValueError:
            # Raise error with specific format requirement
            raise ValueError(f"Date '{v}' must be in ISO 8601 format (YYYY-MM-DD)")

        return v


# ==============================================================================
# Section 1.2: Linking Parameters (Identifiers and Bridges)
# ==============================================================================
class LinkingParams(BaseModel):
    """
    Defines the schema for parameters related to linking different financial
    datasets using various entity identifiers such as PERMNO, CIK, CUSIP, and ticker.

    This model validates parameters that control:
    - Identifier precedence order for linking operations
    - IBES-CRSP link table version specification
    - Historical ticker symbol handling policies
    - Dual-class share resolution strategies
    - Corporate actions data source specification

    These parameters ensure consistent and reproducible linking across
    CRSP, Compustat, I/B/E/S, and other financial databases used in the study.

    Attributes:
        id_precedence: Ordered list of identifier types for linking priority
        ibes_link_table_version: Version of IBES-CRSP link table to use
        ticker_history_handling: Policy for historical ticker changes
        dual_class_resolution: Strategy for resolving dual-class share issues
        corporate_actions_source: Database source for corporate actions data
    """

    # The order of precedence for identifiers used in linking operations
    # Determines priority when multiple identifiers are available
    id_precedence: List[str]

    # The version of the IBES-CRSP link table to be used
    # Specifies which version of linking table for consistent results
    ibes_link_table_version: str

    # The policy for handling historical ticker symbols
    # Defines strategy for ticker changes over time
    ticker_history_handling: str

    # The method for resolving issues with dual-class shares
    # Specifies approach for companies with multiple share classes
    dual_class_resolution: str

    # The source database for corporate actions data
    # Identifies data provider for corporate actions (splits, dividends, etc.)
    corporate_actions_source: str

    @validator('id_precedence')
    def validate_id_precedence_list(cls, v: List[str]) -> List[str]:
        """
        Validates that identifier precedence list contains valid identifier types.

        Args:
            v: List of identifier type strings

        Returns:
            Validated identifier precedence list

        Raises:
            ValueError: If list is empty or contains invalid identifier types
        """
        # Check that list is not empty
        if not v:
            raise ValueError("id_precedence cannot be empty")

        # Define valid identifier types based on financial database standards
        valid_ids = {'permno', 'cik', 'cusip8', 'cusip', 'ticker', 'ibes_ticker'}

        # Check that all identifiers are valid
        invalid_ids = set(v) - valid_ids
        if invalid_ids:
            raise ValueError(f"Invalid identifier types: {invalid_ids}. "
                           f"Valid types: {valid_ids}")

        return v


# ==============================================================================
# Section 1.3: Text Cleaning and Provenance Parameters
# ==============================================================================
class StripBoilerplateParams(BaseModel):
    """
    Defines flags for removing specific boilerplate sections commonly found
    in earnings press releases that do not contain material information.

    This model controls the removal of standardized legal and administrative
    text that appears consistently across press releases but does not contribute
    to the analysis of firm-specific content.

    Attributes:
        forward_looking_statements: Remove forward-looking statement disclaimers
        non_gaap_disclosures: Remove non-GAAP reconciliation boilerplate
        contacts_and_ir_headers: Remove contact information and IR headers
        exhibit_headers_numbers: Remove exhibit headers and numbering
    """

    # Flag to remove forward-looking statement disclaimers
    # Removes legal boilerplate about forward-looking statements
    forward_looking_statements: bool

    # Flag to remove non-GAAP reconciliation boilerplate
    # Removes standardized non-GAAP disclosure language
    non_gaap_disclosures: bool

    # Flag to remove contact information and investor relations headers
    # Removes standardized contact blocks and IR information
    contacts_and_ir_headers: bool

    # Flag to remove exhibit headers and numbering
    # Removes administrative headers like "Exhibit 99.1"
    exhibit_headers_numbers: bool


class RemovePatternsParams(BaseModel):
    """
    Defines flags for removing common noisy text patterns that do not
    contribute meaningful information to the textual analysis.

    This model controls the removal of technical artifacts and
    administrative content that can interfere with natural language
    processing and topic modeling.

    Attributes:
        urls: Remove HTTP/HTTPS URLs from text
        emails: Remove email addresses from text
        phone_numbers: Remove phone number patterns from text
        standalone_numbers: Remove isolated numeric tokens
        repeated_headers: Remove repeated header patterns
    """

    # Flag to remove HTTP/HTTPS URLs from text
    # Removes web addresses that don't contribute to content analysis
    urls: bool

    # Flag to remove email addresses from text
    # Removes email patterns using regex matching
    emails: bool

    # Flag to remove phone number patterns from text
    # Removes various phone number formats
    phone_numbers: bool

    # Flag to remove isolated numeric tokens
    # Removes standalone numbers that aren't part of financial metrics
    standalone_numbers: bool

    # Flag to remove repeated header patterns
    # Removes repetitive administrative headers
    repeated_headers: bool


class NormalizationParams(BaseModel):
    """
    Defines parameters for text normalization operations applied to
    press release content to ensure consistent text representation.

    This model controls Unicode normalization, whitespace handling,
    and case normalization to create standardized text for analysis.

    Attributes:
        whitespace_collapse: Collapse multiple whitespace to single space
        unicode_normalize: Unicode normalization form (e.g., 'NFKC')
        lowercase: Convert all text to lowercase
    """

    # Flag to collapse multiple whitespace characters to single space
    # Standardizes whitespace formatting across documents
    whitespace_collapse: bool

    # Unicode normalization form to apply (e.g., 'NFKC')
    # Ensures consistent Unicode character representation
    unicode_normalize: str

    # Flag to convert all text to lowercase
    # Standardizes case for consistent text processing
    lowercase: bool

    @validator('unicode_normalize')
    def validate_unicode_form(cls, v: str) -> str:
        """
        Validates that Unicode normalization form is supported.

        Args:
            v: Unicode normalization form string

        Returns:
            Validated normalization form

        Raises:
            ValueError: If normalization form is not supported
        """
        # Define valid Unicode normalization forms
        valid_forms = {'NFC', 'NFKC', 'NFD', 'NFKD'}

        # Check that form is valid
        if v not in valid_forms:
            raise ValueError(f"Invalid Unicode normalization form: {v}. "
                           f"Valid forms: {valid_forms}")

        return v


class TokenProcessingParams(BaseModel):
    """
    Defines parameters for token-level processing specifically for
    bag-of-words models, including lemmatization and stopword removal.

    This model controls advanced text processing operations that are
    applied only to bag-of-words vectorization pipelines, not to
    BERT-family transformer models.

    Attributes:
        lemmatize: Apply lemmatization to reduce words to root forms
        stopwords_list: Specification of stopword list to use
        remove_repeated_patterns: Remove repeated token patterns
    """

    # Flag to apply lemmatization to reduce words to root forms
    # Uses WordNet lemmatizer to normalize word variations
    lemmatize: bool

    # Specification of stopword list to use (e.g., 'english_default')
    # Defines which stopword list for filtering common words
    stopwords_list: str

    # Flag to remove repeated token patterns
    # Removes patterns like "year year" -> "year"
    remove_repeated_patterns: bool


class LanguageDetectionParams(BaseModel):
    """
    Defines parameters for language detection and filtering to ensure
    analysis is performed only on documents in specified languages.

    This model controls automatic language detection and filtering
    based on confidence thresholds to maintain text quality.

    Attributes:
        enforce_language: Flag to enforce language filtering
        threshold: Confidence threshold for language detection (0.0-1.0)
    """

    # Flag to enforce language filtering based on detection
    # When True, filters documents based on detected language
    enforce_language: bool

    # Confidence threshold for language detection (0.0-1.0)
    # Minimum confidence required for language classification
    threshold: float = Field(..., ge=0.0, le=1.0)


class TextCleaningParams(BaseModel):
    """
    Defines the complete schema for all text cleaning and preprocessing steps
    applied to earnings press releases before vectorization.

    This model orchestrates all cleaning operations including HTML processing,
    boilerplate removal, pattern filtering, normalization, and token processing.
    The cleaning pipeline ensures consistent, high-quality text input for
    subsequent natural language processing operations.

    Attributes:
        remove_html_tables: Flag to remove HTML table elements
        strip_boilerplate_sections: Parameters for boilerplate removal
        remove_patterns: Parameters for pattern-based filtering
        normalization: Parameters for text normalization
        token_processing_bow_only: Parameters for token-level processing
        language_detection: Parameters for language filtering
        cleaning_audit_log: Flag to maintain detailed cleaning logs
    """

    # Flag to remove HTML table elements from press releases
    # Removes <table> tags and contents which often contain financial data
    remove_html_tables: bool

    # Parameters for removing specific boilerplate sections
    # Controls removal of standardized legal and administrative text
    strip_boilerplate_sections: StripBoilerplateParams

    # Parameters for removing common noisy text patterns
    # Controls removal of URLs, emails, phone numbers, etc.
    remove_patterns: RemovePatternsParams

    # Parameters for text normalization operations
    # Controls Unicode, whitespace, and case normalization
    normalization: NormalizationParams

    # Parameters for token-level processing (bag-of-words only)
    # Controls lemmatization and stopword removal for BoW models
    token_processing_bow_only: TokenProcessingParams

    # Parameters for language detection and filtering
    # Controls automatic language detection and confidence thresholds
    language_detection: LanguageDetectionParams

    # Flag to maintain detailed cleaning audit logs
    # When True, logs all cleaning operations for reproducibility
    cleaning_audit_log: bool


# ==============================================================================
# Section 1.4: Vectorization Parameters (BKMX, oLDA, Transformers)
# ==============================================================================
class BKMXParams(BaseModel):
    """
    Defines parameters for the BKMX taxonomy vectorization method, which uses
    a pre-defined topic taxonomy to extract thematic content from press releases.

    This model controls the application of the BKMX (Bossaerts, Kahana,
    Moldovanu, and Xie) taxonomy for mapping press release content to
    predefined economic and political topics.

    Attributes:
        taxonomy_resource_path: File path to BKMX taxonomy resource
        use_bkxmx: Flag to enable BKMX vectorization
        metatopics_selected: List of metatopics to include in analysis
    """

    # File path to the BKMX taxonomy resource file
    # Points to JSON file containing 180 topics and phrase mappings
    taxonomy_resource_path: str

    # Flag to enable BKMX vectorization in the pipeline
    # When True, applies BKMX taxonomy matching to documents
    use_bkxmx: bool

    # List of BKMX metatopics to include in the analysis
    # Subset of 11 metatopics from full BKMX taxonomy
    metatopics_selected: List[str]

    @validator('metatopics_selected')
    def validate_metatopics(cls, v: List[str]) -> List[str]:
        """
        Validates that selected metatopics are from the BKMX taxonomy.

        Args:
            v: List of metatopic names

        Returns:
            Validated metatopic list

        Raises:
            ValueError: If metatopics are not valid BKMX categories
        """
        # Define valid BKMX metatopics as specified in the study
        valid_metatopics = {
            "Announcements", "Corporate Earnings", "Financial Intermediaries",
            "National Politics", "Science / Arts", "Economic Growth",
            "Financial Markets", "Industry", "International Relations",
            "Oil / Mining", "Technology"
        }

        # Check that all selected metatopics are valid
        invalid_topics = set(v) - valid_metatopics
        if invalid_topics:
            raise ValueError(f"Invalid BKMX metatopics: {invalid_topics}. "
                           f"Valid topics: {valid_metatopics}")

        return v


class BoWVectorizerParams(BaseModel):
    """
    Defines parameters for the Bag-of-Words CountVectorizer used in the
    online LDA (oLDA) pipeline for document-term matrix construction.

    This model specifies the exact configuration for scikit-learn's
    CountVectorizer to ensure reproducible vocabulary construction
    and document representation.

    Attributes:
        analyzer: Type of analysis ('word' for word-level analysis)
        ngram_range: Range of n-gram sizes (1,1 for unigrams only)
        min_df: Minimum document frequency for vocabulary inclusion
        max_df: Maximum document frequency for vocabulary inclusion
        binary: Flag for binary vs. count vectorization
        lowercase: Flag to convert to lowercase before tokenization
        vocabulary_policy: Policy for vocabulary fitting across years
        random_seed: Random seed for reproducible results
    """

    # Type of analysis - must be 'word' for word-level tokenization
    # Specifies tokenization at word level rather than character level
    analyzer: constr(regex=r'^word$')

    # Range of n-gram sizes as (min_n, max_n) tuple
    # Set to (1,1) for unigrams only as specified in methodology
    ngram_range: Tuple[int, int]

    # Minimum document frequency for vocabulary inclusion
    # Words appearing in fewer than min_df documents are excluded
    min_df: int = Field(..., gt=0)

    # Maximum document frequency for vocabulary inclusion
    # Words appearing in more than max_df fraction of documents are excluded
    max_df: float = Field(..., gt=0.0, lt=1.0)

    # Flag for binary vs. count vectorization
    # When True, uses binary indicators; when False, uses term counts
    binary: bool

    # Flag to convert text to lowercase before tokenization
    # Ensures case-insensitive vocabulary construction
    lowercase: bool

    # Policy for vocabulary fitting across rolling windows
    # Controls whether vocabulary is fit on training years only
    vocabulary_policy: str

    # Random seed for reproducible vectorization results
    # Ensures consistent vocabulary ordering across runs
    random_seed: int

    @validator('ngram_range')
    def validate_ngram_range(cls, v: Tuple[int, int]) -> Tuple[int, int]:
        """
        Validates that n-gram range is logically consistent.

        Args:
            v: Tuple of (min_n, max_n) for n-gram range

        Returns:
            Validated n-gram range tuple

        Raises:
            ValueError: If n-gram range is invalid
        """
        min_n, max_n = v

        # Check that both values are positive
        if min_n <= 0 or max_n <= 0:
            raise ValueError("n-gram range values must be positive")

        # Check that min_n <= max_n
        if min_n > max_n:
            raise ValueError("min_n cannot be greater than max_n in ngram_range")

        return v


class RollingProtocolParams(BaseModel):
    """
    Defines the schema for the rolling window protocol used in online LDA
    to ensure proper temporal partitioning and avoid look-ahead bias.

    This model specifies how data is partitioned across years for training
    and scoring to maintain the temporal structure required for realistic
    out-of-sample evaluation.

    Attributes:
        initial_train_year: First year used for model training
        score_years: List of years for which predictions are generated
        train_on_year: Template for training year specification
        score_on_year: Template for scoring year specification
        fit_vectorizer_on_train_only: Flag to fit vocabulary on training only
    """

    # First year used for initial model training
    # Defines the starting point for the rolling window protocol
    initial_train_year: int

    # List of years for which out-of-sample predictions are generated
    # Typically ranges from 2006 to 2023 for the main analysis
    score_years: List[int]

    # Template string for training year specification
    # Describes how training years are determined (e.g., "t")
    train_on_year: str

    # Template string for scoring year specification
    # Describes how scoring years are determined (e.g., "t+1")
    score_on_year: str

    # Flag to fit vectorizer vocabulary on training years only
    # Prevents look-ahead bias in vocabulary construction
    fit_vectorizer_on_train_only: bool

    @validator('score_years')
    def validate_score_years(cls, v: List[int]) -> List[int]:
        """
        Validates that score years are consecutive and reasonable.

        Args:
            v: List of years for scoring

        Returns:
            Validated score years list

        Raises:
            ValueError: If score years are not consecutive or reasonable
        """
        # Check that list is not empty
        if not v:
            raise ValueError("score_years cannot be empty")

        # Check that years are reasonable (between 2000 and 2030)
        if min(v) < 2000 or max(v) > 2030:
            raise ValueError("Score years must be between 2000 and 2030")

        # Check that years are consecutive
        sorted_years = sorted(v)
        for i in range(1, len(sorted_years)):
            if sorted_years[i] - sorted_years[i-1] != 1:
                raise ValueError("Score years must be consecutive")

        return v


class OLDAParams(BaseModel):
    """
    Defines hyperparameters for the Online Latent Dirichlet Allocation (oLDA)
    model used for topic modeling of earnings press releases.

    This model specifies the exact configuration for scikit-learn's
    LatentDirichletAllocation with online learning to ensure reproducible
    topic extraction and document representation.

    Attributes:
        num_topics: Number of topics to extract (K=50 in the study)
        learning_method: Learning algorithm ('online' for online LDA)
        learning_offset: Learning rate offset parameter (τ₀)
        decay: Learning rate decay parameter (κ)
        doc_topic_prior: Document-topic Dirichlet prior (α)
        topic_word_prior: Topic-word Dirichlet prior (η)
        max_iter: Maximum number of iterations for convergence
        evaluate_every: Frequency of likelihood evaluation (-1 for none)
        random_seed: Random seed for reproducible topic extraction
        rolling_protocol: Parameters for rolling window application
    """

    # Number of topics to extract from the corpus
    # Set to 50 topics as specified in the study methodology
    num_topics: int = Field(..., gt=1)

    # Learning algorithm - must be 'online' for online LDA
    # Enables incremental learning suitable for rolling windows
    learning_method: constr(regex=r'^online$')

    # Learning rate offset parameter (τ₀ in LDA literature)
    # Controls the learning rate schedule for online updates
    learning_offset: float = Field(..., gt=0)

    # Learning rate decay parameter (κ in LDA literature)
    # Controls how quickly the learning rate decreases
    decay: float = Field(..., gt=0.5, le=1.0)

    # Document-topic Dirichlet prior parameter (α)
    # Higher values encourage documents to contain multiple topics
    doc_topic_prior: float = Field(..., gt=0)

    # Topic-word Dirichlet prior parameter (η)
    # Higher values encourage topics to contain more words
    topic_word_prior: float = Field(..., gt=0)

    # Maximum number of iterations for convergence
    # Sets upper bound on optimization iterations
    max_iter: int = Field(..., gt=0)

    # Frequency of likelihood evaluation (-1 for no evaluation)
    # Controls computational overhead of convergence monitoring
    evaluate_every: int

    # Random seed for reproducible topic extraction
    # Ensures consistent topic assignments across runs
    random_seed: int

    # Parameters for rolling window protocol application
    # Controls how oLDA is applied across temporal partitions
    rolling_protocol: RollingProtocolParams


class TransformerModelParams(BaseModel):
    """
    Defines parameters for Transformer model embedding extraction using
    BERT-family models for document representation.

    This model specifies the configuration for loading and applying
    pre-trained transformer models (BERT, FinBERT, MPNET) to extract
    contextual embeddings from press release text.

    Attributes:
        model_names: Dictionary mapping model keys to HuggingFace identifiers
        model_revision: Git revision or tag for model versions
        device: Device specification for model inference ('auto', 'cpu', 'cuda')
        batch_size: Batch size for efficient inference
        max_token_length: Maximum sequence length for tokenization
        truncation_policy: Strategy for handling long sequences
        pooling: Pooling strategy for sequence-level representation
        use_last_hidden_state: Flag to use final layer representations
        normalize_embeddings: Flag to normalize embedding vectors
        fine_tuning: Flag to enable fine-tuning (False for frozen models)
        tokenizer_fast: Flag to use fast tokenizer implementation
        random_seed: Random seed for reproducible inference
    """

    # Dictionary mapping model keys to HuggingFace model identifiers
    # Specifies exact model versions for reproducible results
    model_names: Dict[str, str]

    # Git revision or tag for model versions
    # Ensures exact model version specification
    model_revision: str

    # Device specification for model inference
    # 'auto' selects GPU if available, otherwise CPU
    device: str

    # Batch size for efficient transformer inference
    # Balances memory usage and computational efficiency
    batch_size: int = Field(..., gt=0)

    # Maximum sequence length for tokenization
    # Set to 512 tokens as specified in methodology
    max_token_length: int = Field(..., gt=0)

    # Strategy for handling sequences longer than max_token_length
    # 'head' truncation retains first tokens as specified
    truncation_policy: str

    # Pooling strategy for converting token embeddings to document embeddings
    # 'mean_last' uses mean pooling of last hidden states
    pooling: str

    # Flag to use final layer hidden state representations
    # When True, extracts embeddings from last transformer layer
    use_last_hidden_state: bool

    # Flag to normalize embedding vectors to unit length
    # When True, applies L2 normalization to embeddings
    normalize_embeddings: bool

    # Flag to enable fine-tuning (False for frozen pre-trained models)
    # When False, uses models as frozen feature extractors
    fine_tuning: bool

    # Flag to use fast tokenizer implementation
    # When True, uses optimized Rust-based tokenizers
    tokenizer_fast: bool

    # Random seed for reproducible transformer inference
    # Ensures consistent results across runs
    random_seed: int

    @validator('model_names')
    def validate_model_names(cls, v: Dict[str, str]) -> Dict[str, str]:
        """
        Validates that model names dictionary contains required models.

        Args:
            v: Dictionary of model key to identifier mappings

        Returns:
            Validated model names dictionary

        Raises:
            ValueError: If required models are missing
        """
        # Define required model keys as specified in the study
        required_models = {'bert', 'finbert', 'mpnet'}

        # Check that all required models are present
        missing_models = required_models - set(v.keys())
        if missing_models:
            raise ValueError(f"Missing required models: {missing_models}")

        # Check that model identifiers are non-empty strings
        for key, model_id in v.items():
            if not isinstance(model_id, str) or not model_id.strip():
                raise ValueError(f"Model identifier for '{key}' must be a non-empty string")

        return v

    @validator('device')
    def validate_device(cls, v: str) -> str:
        """
        Validates that device specification is supported.

        Args:
            v: Device specification string

        Returns:
            Validated device specification

        Raises:
            ValueError: If device specification is invalid
        """
        # Define valid device specifications
        valid_devices = {'auto', 'cpu', 'cuda', 'cuda:0', 'cuda:1', 'mps'}

        # Check for exact matches or cuda:N pattern
        if v in valid_devices or (v.startswith('cuda:') and v[5:].isdigit()):
            return v

        raise ValueError(f"Invalid device specification: {v}. "
                        f"Valid options: {valid_devices} or 'cuda:N'")


class VectorizationParams(BaseModel):
    """
    Defines the complete schema for all text vectorization methods used
    in the study, including BKMX taxonomy, online LDA, and transformer models.

    This model orchestrates all vectorization approaches to ensure consistent
    and reproducible document representation across different methodologies.
    Each vectorization method produces features for the Lasso regression
    mapping text content to stock returns.

    Attributes:
        bkxmx: Parameters for BKMX taxonomy vectorization
        bag_of_words_vectorizer: Parameters for CountVectorizer used with oLDA
        olda: Parameters for online Latent Dirichlet Allocation
        transformers: Parameters for BERT-family transformer models
    """

    # Parameters for BKMX taxonomy-based vectorization
    # Controls application of pre-defined economic topic taxonomy
    bkxmx: BKMXParams

    # Parameters for bag-of-words CountVectorizer used with oLDA
    # Controls document-term matrix construction for topic modeling
    bag_of_words_vectorizer: BoWVectorizerParams

    # Parameters for online Latent Dirichlet Allocation topic modeling
    # Controls unsupervised topic extraction from press releases
    olda: OLDAParams

    # Parameters for BERT-family transformer model embedding extraction
    # Controls contextual embedding generation using pre-trained models
    transformers: TransformerModelParams


# ==============================================================================
# Section 1.5: Modeling and Rolling Estimation Parameters
# ==============================================================================
class LassoParams(BaseModel):
    """
    Defines hyperparameters for the Lasso regression model used to map
    text features to stock returns in the rolling estimation framework.

    This model implements the regularized regression specified in the study:
    ŵ = argmin_w {(1/2N)||X_std w - y||²₂ + λ||w||₁}

    The Lasso provides feature selection and prevents overfitting in the
    high-dimensional text-to-return mapping problem.

    Attributes:
        lambda_val: L1 regularization parameter (λ=1e-5 in study)
        fit_intercept: Flag to include intercept term in regression
        standardize_X: Flag to standardize feature matrix
        standardize_y: Flag to standardize target variable
        solver: Optimization algorithm ('cd' for coordinate descent)
        max_iter: Maximum iterations for convergence
        tol: Convergence tolerance for optimization
        random_seed: Random seed for reproducible results
    """

    # L1 regularization parameter (λ in the Lasso objective)
    # Set to 1e-5 as specified in the study methodology
    lambda_val: float = Field(..., alias='lambda', ge=0)

    # Flag to include intercept term in the regression
    # When True, fits α + βX; when False, fits βX only
    fit_intercept: bool

    # Flag to standardize the feature matrix X before fitting
    # When True, applies z-score normalization to features
    standardize_X: bool

    # Flag to standardize the target variable y before fitting
    # When False, uses raw returns as specified in methodology
    standardize_y: bool

    # Optimization algorithm for Lasso regression
    # 'cd' specifies coordinate descent as used in scikit-learn
    solver: str

    # Maximum number of iterations for convergence
    # Set to 10000 to ensure convergence for high-dimensional problems
    max_iter: int = Field(..., gt=0)

    # Convergence tolerance for optimization
    # Set to 1e-6 for precise convergence as specified
    tol: float = Field(..., gt=0)

    # Random seed for reproducible regression results
    # Ensures consistent feature selection across runs
    random_seed: int

    @validator('solver')
    def validate_solver(cls, v: str) -> str:
        """
        Validates that Lasso solver is supported.

        Args:
            v: Solver specification string

        Returns:
            Validated solver specification

        Raises:
            ValueError: If solver is not supported
        """
        # Define valid solvers for Lasso regression
        valid_solvers = {'cd', 'saga', 'liblinear'}

        # Check that solver is valid
        if v not in valid_solvers:
            raise ValueError(f"Invalid Lasso solver: {v}. "
                           f"Valid solvers: {valid_solvers}")

        return v


class RollingEstimationParams(BaseModel):
    """
    Defines the schema for the rolling estimation framework used to generate
    out-of-sample soft scores while avoiding look-ahead bias.

    This model specifies how the temporal partitioning is implemented to
    train models on year t and score year t+1, ensuring realistic
    investment strategy evaluation.

    Attributes:
        initial_train_year: First year used for model training (2005)
        score_years: Years for which predictions are generated (2006-2023)
        partition_key: Method for temporal partitioning
        guard_lookahead: Flag to enforce strict temporal separation
        recompute_per_model: Flag to refit models for each vectorization method
    """

    # First year used for initial model training
    # Set to 2005 to provide initial training data
    initial_train_year: int

    # List of years for which out-of-sample predictions are generated
    # Typically 2006-2023 for the main analysis period
    score_years: List[int]

    # Method for temporal partitioning of the data
    # Specifies how calendar years are extracted from timestamps
    partition_key: str

    # Flag to enforce strict temporal separation and avoid look-ahead bias
    # When True, ensures training data precedes scoring data
    guard_lookahead: bool

    # Flag to refit models separately for each vectorization method
    # When True, fits separate Lasso models for each text representation
    recompute_per_model: bool

    @validator('initial_train_year')
    def validate_initial_train_year(cls, v: int) -> int:
        """
        Validates that initial training year is reasonable.

        Args:
            v: Initial training year

        Returns:
            Validated initial training year

        Raises:
            ValueError: If year is not reasonable
        """
        # Check that year is in reasonable range
        if v < 2000 or v > 2020:
            raise ValueError(f"Initial train year {v} must be between 2000 and 2020")

        return v

    @root_validator
    def validate_year_consistency(cls, values: Dict[str, Any]) -> Dict[str, Any]:
        """
        Validates consistency between initial training year and score years.

        Args:
            values: Dictionary of field values from the model

        Returns:
            Dictionary of validated values

        Raises:
            ValueError: If years are inconsistent
        """
        # Retrieve initial training year and score years
        initial_year = values.get('initial_train_year')
        score_years = values.get('score_years')

        # Check consistency if both are present
        if initial_year is not None and score_years is not None:
            min_score_year = min(score_years)

            # Ensure initial training year precedes scoring
            if initial_year >= min_score_year:
                raise ValueError(f"Initial train year ({initial_year}) must be "
                               f"before minimum score year ({min_score_year})")

        return values


class SoftSignalAggregationParams(BaseModel):
    """
    Defines parameters for aggregating individual soft scores from different
    vectorization methods into a composite signal.

    This model specifies how to combine the soft scores from BKMX, oLDA,
    BERT, MPNET, and FinBERT into a single aggregated signal for analysis.
    The aggregation creates Soft^Mean as specified in the study.

    Attributes:
        models_included: List of vectorization methods to include
        aggregation: Aggregation method ('simple_mean' for arithmetic mean)
        output_field: Name of the aggregated signal column
    """

    # List of vectorization methods to include in aggregation
    # Typically includes all five methods: BKMX, oLDA, BERT, MPNET, FinBERT
    models_included: List[str] = Field(..., min_items=1)

    # Aggregation method for combining individual soft scores
    # 'simple_mean' computes arithmetic mean across methods
    aggregation: str

    # Name of the output field for the aggregated signal
    # Creates column with this name containing the composite score
    output_field: str = Field(..., min_length=1)

    @validator('models_included')
    def validate_models_included(cls, v: List[str]) -> List[str]:
        """
        Validates that included models are valid vectorization methods.

        Args:
            v: List of model names to include

        Returns:
            Validated model list

        Raises:
            ValueError: If model names are invalid
        """
        # Define valid vectorization methods
        valid_models = {'bkxmx', 'olda', 'bert', 'mpnet', 'finbert'}

        # Check that all models are valid
        invalid_models = set(v) - valid_models
        if invalid_models:
            raise ValueError(f"Invalid models: {invalid_models}. "
                           f"Valid models: {valid_models}")

        return v

    @validator('aggregation')
    def validate_aggregation_method(cls, v: str) -> str:
        """
        Validates that aggregation method is supported.

        Args:
            v: Aggregation method name

        Returns:
            Validated aggregation method

        Raises:
            ValueError: If aggregation method is not supported
        """
        # Define valid aggregation methods
        valid_methods = {'simple_mean', 'weighted_mean', 'median'}

        # Check that method is valid
        if v not in valid_methods:
            raise ValueError(f"Invalid aggregation method: {v}. "
                           f"Valid methods: {valid_methods}")

        return v


class ModelingParams(BaseModel):
    """
    Defines the complete schema for the predictive modeling stage, including
    Lasso regression, rolling estimation, and soft signal aggregation.

    This model orchestrates the machine learning pipeline that maps
    text features to stock returns using regularized regression with
    proper temporal validation to avoid look-ahead bias.

    Attributes:
        lasso: Parameters for Lasso regression model
        rolling_estimation: Parameters for rolling window validation
        soft_signal_aggregation: Parameters for signal aggregation
    """

    # Parameters for Lasso regression model fitting
    # Controls the regularized regression mapping text to returns
    lasso: LassoParams

    # Parameters for rolling window estimation framework
    # Controls temporal partitioning to avoid look-ahead bias
    rolling_estimation: RollingEstimationParams

    # Parameters for aggregating individual soft scores
    # Controls creation of composite signal from multiple methods
    soft_signal_aggregation: SoftSignalAggregationParams


# ==============================================================================
# Section 1.6: Regression and Inference Parameters
# ==============================================================================
class ClusterParams(BaseModel):
    """
    Defines parameters for computing clustered standard errors in the
    cross-sectional regression analysis.

    This model specifies the two-way clustering by firm (permno) and
    time (ann_trade_date) as standard in asset pricing research to
    account for within-firm and within-time correlation in residuals.

    Attributes:
        two_way: Flag to enable two-way clustering
        cluster_by: List of variables to cluster by
    """

    # Flag to enable two-way clustering of standard errors
    # When True, clusters by both firm and time dimensions
    two_way: bool

    # List of variables to use for clustering
    # Typically ['permno', 'ann_trade_date'] for firm and time clustering
    cluster_by: List[str]

    @validator('cluster_by')
    def validate_cluster_variables(cls, v: List[str]) -> List[str]:
        """
        Validates that clustering variables are appropriate.

        Args:
            v: List of clustering variable names

        Returns:
            Validated clustering variables

        Raises:
            ValueError: If clustering variables are invalid
        """
        # Check that list is not empty
        if not v:
            raise ValueError("cluster_by cannot be empty when clustering is enabled")

        # Define valid clustering variables for the study
        valid_vars = {'permno', 'ann_trade_date', 'cik', 'ticker'}

        # Check that all variables are valid
        invalid_vars = set(v) - valid_vars
        if invalid_vars:
            raise ValueError(f"Invalid clustering variables: {invalid_vars}. "
                           f"Valid variables: {valid_vars}")

        return v


class RegressionParams(BaseModel):
    """
    Defines the complete schema for the final econometric analysis including
    regression specifications, inference parameters, and sample restrictions.

    This model controls the cross-sectional regressions that test the
    relationship between text-based soft information and stock returns:
    Ret_{c,τ} = α + β₀ Surprise_{c,τ} + β₁ Soft_{c,τ} + ε_{c,τ}

    Attributes:
        baseline_spec: Regression formula for baseline specification
        combined_spec: Regression formula for combined specification
        winsorize_surprise_quantiles: Quantiles for surprise winsorization
        cluster_standard_errors: Parameters for clustered standard errors
        intercept: Flag to include intercept in regressions
        robust_covariance: Type of robust covariance estimator
        software_engine: Statistical software for regression estimation
        sample_period: Date range for regression analysis
    """

    # Regression formula for baseline specification
    # Specifies the basic model with surprise and one soft signal
    baseline_spec: str

    # Regression formula for combined specification
    # Specifies the model with surprise and all soft signals
    combined_spec: str

    # Quantiles for winsorizing earnings surprise variable
    # Tuple of (lower, upper) quantiles for outlier treatment
    winsorize_surprise_quantiles: Tuple[float, float]

    # Parameters for clustered standard error computation
    # Controls two-way clustering by firm and time
    cluster_standard_errors: ClusterParams

    # Flag to include intercept term in regressions
    # When True, estimates α + βX; when False, estimates βX
    intercept: bool

    # Type of robust covariance estimator to use
    # 'HC1' specifies Huber-White heteroskedasticity-consistent estimator
    robust_covariance: str

    # Statistical software engine for regression estimation
    # 'statsmodels_OLS' specifies using statsmodels OLS implementation
    software_engine: str

    # Date range for regression analysis as (start, end) tuple
    # Typically ('2006-01-01', '2023-12-31') for main analysis
    sample_period: Tuple[str, str]

    @validator('winsorize_surprise_quantiles')
    def validate_quantiles(cls, v: Tuple[float, float]) -> Tuple[float, float]:
        """
        Validates the format, bounds, and order of winsorization quantiles.

        This validator ensures that quantiles are properly specified for
        outlier treatment of the earnings surprise variable, following
        standard practices in financial research.

        Args:
            v: Tuple of (lower, upper) quantiles for winsorization

        Returns:
            Validated quantiles tuple

        Raises:
            ValueError: If quantiles are improperly specified
        """
        # Check that input is a 2-element tuple
        if not (isinstance(v, tuple) and len(v) == 2):
            raise ValueError("Quantiles must be a 2-element tuple")

        lower, upper = v

        # Check that both are valid probabilities
        if not (0.0 <= lower <= 1.0 and 0.0 <= upper <= 1.0):
            raise ValueError("Quantiles must be between 0 and 1")

        # Check that lower < upper
        if lower >= upper:
            raise ValueError("Lower quantile must be less than upper quantile")

        return v

    @validator('sample_period')
    def validate_sample_period(cls, v: Tuple[str, str]) -> Tuple[str, str]:
        """
        Validates that sample period dates are properly formatted and ordered.

        Args:
            v: Tuple of (start_date, end_date) strings

        Returns:
            Validated sample period tuple

        Raises:
            ValueError: If dates are invalid or misordered
        """
        # Check that input is a 2-element tuple
        if not (isinstance(v, tuple) and len(v) == 2):
            raise ValueError("Sample period must be a 2-element tuple")

        start_str, end_str = v

        try:
            # Parse dates to validate format
            start_dt = datetime.datetime.fromisoformat(start_str)
            end_dt = datetime.datetime.fromisoformat(end_str)

            # Check that start is before end
            if start_dt >= end_dt:
                raise ValueError("Start date must be before end date")

        except ValueError as e:
            # Re-raise with more context if date parsing fails
            raise ValueError(f"Invalid date format in sample period: {e}")

        return v

    @validator('robust_covariance')
    def validate_robust_covariance(cls, v: str) -> str:
        """
        Validates that robust covariance estimator is supported.

        Args:
            v: Robust covariance estimator name

        Returns:
            Validated estimator name

        Raises:
            ValueError: If estimator is not supported
        """
        # Define valid robust covariance estimators
        valid_estimators = {'HC0', 'HC1', 'HC2', 'HC3', 'HAC', 'cluster'}

        # Check that estimator is valid
        if v not in valid_estimators:
            raise ValueError(f"Invalid robust covariance estimator: {v}. "
                           f"Valid estimators: {valid_estimators}")

        return v


# ==============================================================================
# Section 1.7: SHAP (Attribution) Parameters
# ==============================================================================
class SHAPParams(BaseModel):
    """
    Defines the schema for SHAP (Shapley Additive exPlanations) feature
    attribution analysis used to interpret the relative importance of
    earnings surprise vs. soft information in predicting returns.

    This model implements the exact Shapley value computation:
    I_j = (1/N) Σᵢ |E_{S⊂X\{j}}[y_{S∪{j}} - y_S]|

    The SHAP analysis provides model-agnostic interpretability for the
    linear regression models used in the study.

    Attributes:
        method: SHAP computation method ('exact' for linear models)
        background_set: Dataset used as SHAP background reference
        aggregation: Method for aggregating SHAP values across samples
        report_normalized_percent: Flag to report normalized percentages
    """

    # SHAP computation method to use
    # 'exact' provides exact Shapley values for linear models
    method: str

    # Dataset used as background reference for SHAP computation
    # 'training_year_samples' uses training data as background
    background_set: str

    # Method for aggregating SHAP values across samples
    # 'mean_absolute' computes mean of absolute SHAP values
    aggregation: str

    # Flag to report SHAP importance as normalized percentages
    # When True, normalizes importance to sum to 100%
    report_normalized_percent: bool

    @validator('method')
    def validate_shap_method(cls, v: str) -> str:
        """
        Validates that SHAP computation method is supported.

        Args:
            v: SHAP method name

        Returns:
            Validated SHAP method

        Raises:
            ValueError: If method is not supported
        """
        # Define valid SHAP methods
        valid_methods = {'exact', 'kernel', 'tree', 'linear', 'deep'}

        # Check that method is valid
        if v not in valid_methods:
            raise ValueError(f"Invalid SHAP method: {v}. "
                           f"Valid methods: {valid_methods}")

        return v

    @validator('aggregation')
    def validate_aggregation(cls, v: str) -> str:
        """
        Validates that SHAP aggregation method is supported.

        Args:
            v: Aggregation method name

        Returns:
            Validated aggregation method

        Raises:
            ValueError: If aggregation method is not supported
        """
        # Define valid aggregation methods
        valid_aggregations = {'mean_absolute', 'mean', 'median', 'max'}

        # Check that aggregation is valid
        if v not in valid_aggregations:
            raise ValueError(f"Invalid SHAP aggregation: {v}. "
                           f"Valid aggregations: {valid_aggregations}")

        return v


# ==============================================================================
# Section 1.8: Interpretability (LLM) Parameters and Prompts
# ==============================================================================
class LLMParams(BaseModel):
    """
    Defines parameters for interacting with Large Language Model APIs
    for interpretability tasks including topic labeling and token classification.

    This model specifies the exact configuration for GPT-4o API calls
    used to provide human-interpretable labels for oLDA topics and
    to classify influential tokens into meaningful categories.

    Attributes:
        model_identifier: Specific LLM model version identifier
        temperature: Sampling temperature for response generation
        top_p: Nucleus sampling parameter for response diversity
        frequency_penalty: Penalty for repeated token frequency
        presence_penalty: Penalty for repeated token presence
        response_format: Required response format specification
        request_timeout_seconds: Timeout for API requests
        max_retries: Maximum number of retry attempts
        rate_limit_rps: Rate limit in requests per second
        caching: Flag to enable response caching
        audit_logging: Flag to enable detailed API logging
    """

    # Specific LLM model version identifier
    # Set to 'gpt-4o-2024-08-06' for reproducible results
    model_identifier: str

    # Sampling temperature for response generation (0.0-2.0)
    # Set to 0.0 for deterministic, reproducible responses
    temperature: float = Field(..., ge=0.0, le=2.0)

    # Nucleus sampling parameter for response diversity (0.0-1.0)
    # Set to 1.0 to consider all probability mass
    top_p: float = Field(..., ge=0.0, le=1.0)

    # Penalty for repeated token frequency (-2.0 to 2.0)
    # Set to 0.0 for no frequency-based penalty
    frequency_penalty: float

    # Penalty for repeated token presence (-2.0 to 2.0)
    # Set to 0.0 for no presence-based penalty
    presence_penalty: float

    # Required response format specification
    # Set to {"type": "json_object"} to enforce JSON responses
    response_format: Dict[str, str]

    # Timeout for API requests in seconds
    # Set to 60 seconds for reasonable response time
    request_timeout_seconds: int = Field(..., gt=0)

    # Maximum number of retry attempts for failed requests
    # Set to 3 for robust error handling
    max_retries: int = Field(..., ge=0)

    # Rate limit in requests per second to comply with API limits
    # Set to 1 RPS to avoid rate limiting
    rate_limit_rps: int = Field(..., gt=0)

    # Flag to enable response caching for efficiency
    # When True, caches responses to avoid redundant API calls
    caching: bool

    # Flag to enable detailed API logging for reproducibility
    # When True, logs all API interactions for audit trail
    audit_logging: bool

    @validator('response_format')
    def validate_response_format(cls, v: Dict[str, str]) -> Dict[str, str]:
        """
        Validates that response format is properly specified.

        Args:
            v: Response format dictionary

        Returns:
            Validated response format

        Raises:
            ValueError: If response format is invalid
        """
        # Check that format specifies JSON output
        if v.get('type') != 'json_object':
            raise ValueError("Response format must specify 'json_object' type")

        return v


class ChunkingParams(BaseModel):
    """
    Defines parameters for chunking data for LLM requests to stay within
    token limits and optimize API usage efficiency.

    This model specifies how topics and tokens are batched for processing
    by the LLM to ensure requests stay within context window limits
    while maintaining processing efficiency.

    Attributes:
        token_chunk_size_for_categorization: Maximum tokens per classification chunk
        topic_batch_size: Number of topics per labeling batch
    """

    # Maximum number of tokens per chunk for token classification
    # Set to 500 to stay well within LLM context limits
    token_chunk_size_for_categorization: int

    # Number of topics to include per labeling batch
    # Set to 10 topics per batch for efficient processing
    topic_batch_size: int

    @validator('token_chunk_size_for_categorization')
    def validate_chunk_size(cls, v: int) -> int:
        """
        Validates that chunk size is reasonable for LLM processing.

        Args:
            v: Token chunk size

        Returns:
            Validated chunk size

        Raises:
            ValueError: If chunk size is invalid
        """
        # Check that chunk size is positive and reasonable
        if v <= 0:
            raise ValueError("Token chunk size must be positive")

        if v > 2000:
            raise ValueError("Token chunk size should not exceed 2000 for efficiency")

        return v


class Prompts(BaseModel):
    """
    Stores the templates for system and user prompts used in LLM interactions
    for topic labeling, metatopic condensation, and token classification.

    This model contains the exact prompt templates that ensure consistent
    and reproducible LLM responses for interpretability tasks.

    Attributes:
        topic_labeler_system_prompt: System prompt for topic labeling
        topic_labeler_user_prompt: User prompt template for topic labeling
        metatopic_condenser_system_prompt: System prompt for metatopic condensation
        metatopic_condenser_user_prompt: User prompt template for metatopic condensation
        token_classifier_system_prompt: System prompt for token classification
        token_classifier_user_prompt: User prompt template for token classification
    """

    # System prompt for oLDA topic labeling task
    # Defines the role and requirements for topic annotation
    topic_labeler_system_prompt: str

    # User prompt template for topic labeling requests
    # Contains placeholders for topic data and output format specification
    topic_labeler_user_prompt: str

    # System prompt for metatopic condensation task
    # Defines the role and requirements for topic taxonomy creation
    metatopic_condenser_system_prompt: str

    # User prompt template for metatopic condensation requests
    # Contains placeholders for labeled topics and metatopic count
    metatopic_condenser_user_prompt: str

    # System prompt for token classification task
    # Defines the role and requirements for token categorization
    token_classifier_system_prompt: str

    # User prompt template for token classification requests
    # Contains placeholders for metatopic taxonomy and token chunks
    token_classifier_user_prompt: str

    @validator('topic_labeler_system_prompt', 'topic_labeler_user_prompt',
              'metatopic_condenser_system_prompt', 'metatopic_condenser_user_prompt',
              'token_classifier_system_prompt', 'token_classifier_user_prompt')
    def validate_prompt_content(cls, v: str) -> str:
        """
        Validates that prompt content is non-empty and properly formatted.

        Args:
            v: Prompt string content

        Returns:
            Validated prompt content

        Raises:
            ValueError: If prompt content is invalid
        """
        # Check that prompt is not empty
        if not v or not v.strip():
            raise ValueError("Prompt content cannot be empty")

        # Check that prompt contains basic instructions
        if len(v.strip()) < 50:
            raise ValueError("Prompt content is too short to be meaningful")

        return v


class TaxonomyPolicy(BaseModel):
    """
    Defines the policy for creating and managing the metatopic taxonomy
    used for interpretability analysis.

    This model specifies how the hierarchical topic structure is created
    and maintained to ensure mutual exclusivity and comprehensive coverage
    of extracted topics.

    Attributes:
        taxonomy_source: Source of the taxonomy ('oLDA_derived')
        metatopic_count: Number of metatopics to create
        enforce_mutual_exclusivity: Flag to enforce mutual exclusivity
        allow_other_bucket: Flag to allow 'Other' category for unclassified items
    """

    # Source of the taxonomy structure
    # 'oLDA_derived' indicates taxonomy derived from oLDA topics
    taxonomy_source: str

    # Number of metatopics to create in the taxonomy
    # Set to 9 as specified in the study methodology
    metatopic_count: int

    # Flag to enforce mutual exclusivity of topic assignments
    # When True, ensures each topic appears in exactly one metatopic
    enforce_mutual_exclusivity: bool

    # Flag to allow 'Other' category for unclassified items
    # When True, creates bucket for tokens that don't fit metatopics
    allow_other_bucket: bool

    @validator('metatopic_count')
    def validate_metatopic_count(cls, v: int) -> int:
        """
        Validates that metatopic count is reasonable.

        Args:
            v: Number of metatopics

        Returns:
            Validated metatopic count

        Raises:
            ValueError: If count is invalid
        """
        # Check that count is positive and reasonable
        if v <= 0:
            raise ValueError("Metatopic count must be positive")

        if v > 20:
            raise ValueError("Metatopic count should not exceed 20 for interpretability")

        return v


class InterpretabilityParams(BaseModel):
    """
    Defines the complete schema for all model interpretability steps including
    LLM-based topic labeling, metatopic creation, and token classification.

    This model orchestrates the interpretability pipeline that provides
    human-readable explanations for the topic models and token attributions
    used in the text-to-return mapping.

    Attributes:
        llm: Parameters for LLM API interactions
        chunking: Parameters for data chunking strategies
        prompts: Templates for LLM prompt engineering
        taxonomy_policy: Policy for taxonomy creation and management
    """

    # Parameters for Large Language Model API interactions
    # Controls GPT-4o configuration for interpretability tasks
    llm: LLMParams

    # Parameters for chunking data for LLM processing
    # Controls how topics and tokens are batched for API calls
    chunking: ChunkingParams

    # Templates for LLM prompt engineering
    # Contains all prompt templates for consistent LLM interactions
    prompts: Prompts

    # Policy for taxonomy creation and management
    # Controls how metatopic hierarchy is created and maintained
    taxonomy_policy: TaxonomyPolicy


# ==============================================================================
# Section 1.9: Trading Simulation Parameters (Market Efficiency Test)
# ==============================================================================
class QuoteValidationParams(BaseModel):
    """
    Defines rules for validating NBBO (National Best Bid and Offer) quotes
    used in the market efficiency trading simulation.

    This model specifies the quality filters applied to intraday quotes
    to ensure realistic trading simulation based on valid market data.

    Attributes:
        exclude_locked: Flag to exclude locked markets (bid >= ask)
        exclude_crossed: Flag to exclude crossed markets (bid > ask)
        valid_conditions_only: Flag to use only quotes with valid conditions
    """

    # Flag to exclude locked markets where bid price equals ask price
    # When True, filters out quotes where bid >= ask
    exclude_locked: bool

    # Flag to exclude crossed markets where bid price exceeds ask price
    # When True, filters out quotes where bid > ask
    exclude_crossed: bool

    # Flag to use only quotes with valid market condition codes
    # When True, excludes quotes with irregular condition flags
    valid_conditions_only: bool


class TradingSimulationParams(BaseModel):
    """
    Defines the complete schema for the market efficiency trading simulation
    that tests whether soft information can generate abnormal returns.

    This model implements the trading strategies described in the study:
    - Strategy 1: Enter at quotes, exit at close prices
    - Strategy 2: Enter and exit at quotes (full transaction costs)

    The simulation tests market efficiency by examining post-open profitability.

    Attributes:
        entry_time_et: Time for trade entry (09:45:00 ET)
        exit_time_et: Time for trade exit (16:00:00 ET)
        use_nbbo: Flag to use NBBO quotes for entry/exit
        quote_validation_rules: Rules for validating quote quality
        max_relative_spread_entry: Maximum relative spread for entry
        require_signal_agreement: Flag to require signal agreement
        ranking_method: Method for ranking stocks ('surprise' or 'soft')
        portfolio_weighting: Weighting scheme for portfolio construction
        position_count_policy: Policy for balancing long/short positions
        allow_shorting: Flag to allow short positions
        strategies: List of trading strategies to evaluate
        min_names_per_side: Minimum stocks required per portfolio side
        taq_timezone: Timezone for TAQ timestamp interpretation
        taq_timestamp_tolerance_seconds: Tolerance for timestamp matching
        slippage_model: Model for price impact and slippage
    """

    # Time for trade entry in HH:MM:SS format (09:45:00 ET)
    # Set after market open to avoid microstructure noise
    entry_time_et: str

    # Time for trade exit in HH:MM:SS format (16:00:00 ET)
    # Set at market close for end-of-day liquidation
    exit_time_et: str

    # Flag to use NBBO quotes for entry and exit pricing
    # When True, uses best bid/offer prices from TAQ data
    use_nbbo: bool

    # Rules for validating quote quality before trading
    # Filters for locked, crossed, and invalid condition codes
    quote_validation_rules: QuoteValidationParams

    # Maximum relative spread allowed for trade entry
    # Set to 0.20 (20%) to exclude illiquid stocks
    max_relative_spread_entry: float = Field(..., ge=0.0)

    # Flag to require agreement between surprise and soft signals
    # When True, trades only when both signals have same sign
    require_signal_agreement: bool

    # Method for ranking stocks within portfolio sides
    # 'surprise' ranks by earnings surprise; 'soft' ranks by soft signal
    ranking_method: str

    # Weighting scheme for portfolio construction
    # 'market_cap_t_minus_1' uses market capitalization weights
    portfolio_weighting: str

    # Policy for balancing long and short positions
    # 'min(|L|,|S|)' uses minimum of available long/short candidates
    position_count_policy: str

    # Flag to allow short selling in the simulation
    # When True, enables short positions for negative signals
    allow_shorting: bool

    # List of trading strategies to evaluate
    # Includes 'OC1_close_price_exit' and 'OC2_quote_to_quote'
    strategies: List[str]

    # Minimum number of stocks required per portfolio side
    # Set to 1 to allow single-stock portfolios when necessary
    min_names_per_side: int

    # Timezone for interpreting TAQ timestamps
    # Set to 'America/New_York' for Eastern Time
    taq_timezone: str

    # Tolerance in seconds for matching timestamps in TAQ data
    # Set to 5 seconds for reasonable timestamp matching
    taq_timestamp_tolerance_seconds: int

    # Model for incorporating price impact and slippage
    # 'none' assumes no market impact for simulation simplicity
    slippage_model: str

    @validator('entry_time_et', 'exit_time_et')
    def validate_time_format(cls, v: str) -> str:
        """
        Validates that time strings are in proper HH:MM:SS format.

        Args:
            v: Time string to validate

        Returns:
            Validated time string

        Raises:
            ValueError: If time format is invalid
        """
        try:
            # Parse time to validate format
            datetime.datetime.strptime(v, '%H:%M:%S')
        except ValueError:
            raise ValueError(f"Time '{v}' must be in HH:MM:SS format")

        return v

    @validator('ranking_method')
    def validate_ranking_method(cls, v: str) -> str:
        """
        Validates that ranking method is supported.

        Args:
            v: Ranking method name

        Returns:
            Validated ranking method

        Raises:
            ValueError: If ranking method is invalid
        """
        # Define valid ranking methods
        valid_methods = {'surprise', 'soft', 'combined'}

        # Check that method is valid
        if v not in valid_methods:
            raise ValueError(f"Invalid ranking method: {v}. "
                           f"Valid methods: {valid_methods}")

        return v

    @validator('strategies')
    def validate_strategies(cls, v: List[str]) -> List[str]:
        """
        Validates that trading strategies are supported.

        Args:
            v: List of strategy names

        Returns:
            Validated strategy list

        Raises:
            ValueError: If strategies are invalid
        """
        # Define valid trading strategies
        valid_strategies = {'OC1_close_price_exit', 'OC2_quote_to_quote'}

        # Check that all strategies are valid
        invalid_strategies = set(v) - valid_strategies
        if invalid_strategies:
            raise ValueError(f"Invalid strategies: {invalid_strategies}. "
                           f"Valid strategies: {valid_strategies}")

        return v


# ==============================================================================
# Section 1.10: Hacking Scenario Parameters (Precision@k)
# ==============================================================================
class HackingSimulationParams(BaseModel):
    """
    Defines the schema for the 'hacking scenario' Precision@k analysis
    that evaluates the value of early access to press release information.

    This model implements the evaluation framework that measures how well
    combined hard and soft information can identify top-performing stocks
    when available before public release, as documented in historical
    press release hacking incidents.

    Attributes:
        min_daily_announcements: Minimum announcements per day for analysis
        precision_at_k_values: List of k values for Precision@k evaluation
        evaluate_sides: Sides to evaluate ('winners', 'losers', or both)
        agreement_required: Flag to require signal agreement
        tie_breaker: Variables for tie-breaking in rankings
    """

    # Minimum number of announcements required per day for analysis
    # Set to 20 to ensure sufficient sample size per day
    min_daily_announcements: int = Field(..., gt=0)

    # List of k values for Precision@k evaluation
    # Typically [1, 2, 3, 5, 10] to assess ranking quality
    precision_at_k_values: List[int]

    # Sides of return distribution to evaluate
    # ['winners', 'losers'] evaluates both positive and negative tails
    evaluate_sides: List[str]

    # Flag to require agreement between surprise and soft signals
    # When True, analyzes only events with aligned signals
    agreement_required: bool

    # Variables used for tie-breaking in rankings
    # Typically ['market_cap_t_minus_1', 'permno'] for stable ordering
    tie_breaker: List[str]

    @validator('precision_at_k_values')
    def validate_precision_k_values(cls, v: List[int]) -> List[int]:
        """
        Validates that Precision@k values are reasonable.

        Args:
            v: List of k values for evaluation

        Returns:
            Validated k values list

        Raises:
            ValueError: If k values are invalid
        """
        # Check that list is not empty
        if not v:
            raise ValueError("precision_at_k_values cannot be empty")

        # Check that all values are positive
        if any(k <= 0 for k in v):
            raise ValueError("All Precision@k values must be positive")

        # Check that values are reasonable (not too large)
        if any(k > 100 for k in v):
            raise ValueError("Precision@k values should not exceed 100")

        return v

    @validator('evaluate_sides')
    def validate_evaluate_sides(cls, v: List[str]) -> List[str]:
        """
        Validates that evaluation sides are valid.

        Args:
            v: List of sides to evaluate

        Returns:
            Validated sides list

        Raises:
            ValueError: If sides are invalid
        """
        # Define valid evaluation sides
        valid_sides = {'winners', 'losers'}

        # Check that all sides are valid
        invalid_sides = set(v) - valid_sides
        if invalid_sides:
            raise ValueError(f"Invalid evaluation sides: {invalid_sides}. "
                           f"Valid sides: {valid_sides}")

        return v


# ==============================================================================
# Section 1.11: Factor Data Parameters
# ==============================================================================
class FactorsParams(BaseModel):
    """
    Defines the schema for loading and merging Fama-French factor data
    used in alpha regressions for the trading simulation analysis.

    This model specifies how the three-factor model data (Mkt-rf, HML, SMB)
    is incorporated into the analysis for risk-adjusted return evaluation.

    Attributes:
        ff3_source: Source identifier for Fama-French factor data
        merge_key: Column name used for merging factor data
        fill_policy: Policy for handling missing factor data
    """

    # Source identifier for Fama-French three-factor model data
    # 'Ken_French_Data_Library (daily)' specifies the data provider
    ff3_source: str

    # Column name used for merging factor data with event data
    # 'ann_trade_date' aligns factors with announcement dates
    merge_key: str

    # Policy for handling missing factor data during merge
    # 'strict' requires exact date matches without forward/backward filling
    fill_policy: str

    @validator('fill_policy')
    def validate_fill_policy(cls, v: str) -> str:
        """
        Validates that fill policy for factor data is supported.

        Args:
            v: Fill policy name

        Returns:
            Validated fill policy

        Raises:
            ValueError: If fill policy is invalid
        """
        # Define valid fill policies
        valid_policies = {'strict', 'forward_fill', 'backward_fill', 'interpolate'}

        # Check that policy is valid
        if v not in valid_policies:
            raise ValueError(f"Invalid fill policy: {v}. "
                           f"Valid policies: {valid_policies}")

        return v


# ==============================================================================
# Section 1.12: Diagnostics and Reproducibility
# ==============================================================================
class DiagnosticsParams(BaseModel):
    """
    Defines the schema for diagnostic and reproducibility checks performed
    throughout the analysis pipeline.

    This model specifies which diagnostic analyses to perform, including
    embedding similarity analysis, dataset count reporting, and visualization
    generation for validation against study benchmarks.

    Attributes:
        compute_embedding_cosine_stats: Flag to compute embedding similarities
        report_dataset_counts: Flag to generate dataset count reports
        double_sort_heatmap: Flag to create double-sort return heatmap
        random_state_global: Global random seed for reproducibility
    """

    # Flag to compute cosine similarities between BERT-family embeddings
    # When True, validates embedding relationships (BERT-FinBERT ~0.72, BERT-MPNET ~0.01)
    compute_embedding_cosine_stats: bool

    # Flag to generate comprehensive dataset count reports by year
    # When True, produces sample evolution statistics for validation
    report_dataset_counts: bool

    # Flag to create double-sort heatmap of returns by surprise and soft signals
    # When True, generates 5x5 quintile heatmap for visualization
    double_sort_heatmap: bool

    # Global random seed for ensuring reproducibility across all components
    # Set to 42 for consistent results across all random operations
    random_state_global: int

    @validator('random_state_global')
    def validate_random_state(cls, v: int) -> int:
        """
        Validates that random state is a reasonable integer.

        Args:
            v: Random state value

        Returns:
            Validated random state

        Raises:
            ValueError: If random state is invalid
        """
        # Check that random state is non-negative
        if v < 0:
            raise ValueError("Random state must be non-negative")

        # Check that random state is reasonable (not excessively large)
        if v > 2**31 - 1:
            raise ValueError("Random state must fit in 32-bit signed integer")

        return v


# ==============================================================================
# Section 1.13: Compliance and Model Cards
# ==============================================================================
class ModelCardParams(BaseModel):
    """
    Stores version identifiers for all pre-trained models used in the analysis
    to ensure reproducibility and compliance with model governance requirements.

    This model maintains a registry of exact model versions for all
    external dependencies including transformer models and language models.

    Attributes:
        bert: Version identifier for BERT model
        finbert: Version identifier for FinBERT model
        mpnet: Version identifier for MPNET model
        gpt_4o: Version identifier for GPT-4o model
    """

    # Version identifier for BERT-base-uncased model
    # Specifies exact Hugging Face model version for reproducibility
    bert: str

    # Version identifier for FinBERT financial domain model
    # Specifies exact ProsusAI/finbert model version
    finbert: str

    # Version identifier for MPNET model
    # Specifies exact microsoft/mpnet-base model version
    mpnet: str

    # Version identifier for GPT-4o language model
    # Specifies exact OpenAI model version (gpt-4o-2024-08-06)
    gpt_4o: str

    @validator('bert', 'finbert', 'mpnet', 'gpt_4o')
    def validate_model_version(cls, v: str) -> str:
        """
        Validates that model version identifier is non-empty.

        Args:
            v: Model version string

        Returns:
            Validated model version

        Raises:
            ValueError: If version is empty
        """
        # Check that version is not empty
        if not v or not v.strip():
            raise ValueError("Model version identifier cannot be empty")

        return v


class ComplianceParams(BaseModel):
    """
    Defines the schema for compliance and model governance parameters
    ensuring adherence to data use agreements and ethical AI practices.

    This model tracks compliance status and governance requirements
    for the research pipeline including data usage rights, privacy
    considerations, and model version tracking.

    Attributes:
        data_use_agreements_checked: Flag confirming data use compliance
        pii_policy: Policy statement for personally identifiable information
        model_card_versions: Registry of all model versions used
        pretraining_cutoff_consideration: Policy for model training data cutoffs
    """

    # Flag confirming that all data use agreements have been verified
    # When True, indicates compliance with data provider requirements
    data_use_agreements_checked: bool

    # Policy statement regarding personally identifiable information handling
    # 'no_PII_expected' indicates press releases should not contain PII
    pii_policy: str

    # Registry of version identifiers for all pre-trained models
    # Maintains complete model lineage for governance and reproducibility
    model_card_versions: ModelCardParams

    # Policy regarding pre-training data cutoffs for temporal validity
    # 'prefer_pre-2020_encoders' avoids look-ahead bias in model training
    pretraining_cutoff_consideration: str

    @validator('pii_policy')
    def validate_pii_policy(cls, v: str) -> str:
        """
        Validates that PII policy is appropriately specified.

        Args:
            v: PII policy string

        Returns:
            Validated PII policy

        Raises:
            ValueError: If PII policy is invalid
        """
        # Define valid PII policies
        valid_policies = {'no_PII_expected', 'PII_anonymized', 'PII_redacted', 'PII_present'}

        # Check that policy is valid
        if v not in valid_policies:
            raise ValueError(f"Invalid PII policy: {v}. "
                           f"Valid policies: {valid_policies}")

        return v


# ==============================================================================
# Final Assembled StudyConfig Model
# ==============================================================================
class StudyConfig(BaseModel):
    """
    The complete, fully-specified Pydantic model for the entire study
    configuration encompassing all parameters used throughout the research pipeline.

    This model provides comprehensive validation for the earnings press release
    analysis framework, ensuring type safety, logical consistency, and
    reproducibility across all components:

    - Data acquisition and sample construction
    - Text cleaning and preprocessing
    - Multi-method vectorization (BKMX, oLDA, BERT-family)
    - Rolling Lasso regression for text-to-return mapping
    - Cross-sectional regression analysis with clustered standard errors
    - SHAP-based interpretability analysis
    - LLM-powered topic labeling and classification
    - Market efficiency testing via trading simulation
    - Precision@k evaluation of early information access
    - Comprehensive diagnostics and validation

    The configuration implements the methodology described in:
    "Extracting the Structure of Press Releases for Predicting Earnings Announcement Returns"

    All parameters are validated for consistency with the paper's specifications,
    including equation implementations, statistical methods, and evaluation metrics.

    Attributes:
        data_params: Parameters for data acquisition and filtering
        linking_params: Parameters for dataset linking operations
        text_cleaning_params: Parameters for text preprocessing pipeline
        vectorization_params: Parameters for all text vectorization methods
        modeling_params: Parameters for predictive modeling with Lasso
        regression_params: Parameters for econometric analysis
        shap_params: Parameters for SHAP interpretability analysis
        interpretability_params: Parameters for LLM-based interpretability
        trading_simulation_params: Parameters for market efficiency testing
        hacking_simulation_params: Parameters for Precision@k evaluation
        factors_params: Parameters for factor data integration
        diagnostics_params: Parameters for diagnostic and validation checks
        compliance_params: Parameters for governance and compliance
    """

    # Parameters controlling data acquisition and initial sample construction
    # Validates temporal scope, document filters, and analyst coverage requirements
    data_params: DataParams

    # Parameters controlling entity linking across financial databases
    # Ensures consistent identifier usage and linking strategies
    linking_params: LinkingParams

    # Parameters controlling comprehensive text cleaning and preprocessing
    # Validates HTML processing, boilerplate removal, and normalization steps
    text_cleaning_params: TextCleaningParams

    # Parameters controlling all text vectorization methodologies
    # Orchestrates BKMX taxonomy, oLDA topic modeling, and BERT-family embeddings
    vectorization_params: VectorizationParams

    # Parameters controlling predictive modeling with rolling Lasso regression
    # Implements text-to-return mapping with proper temporal validation
    modeling_params: ModelingParams

    # Parameters controlling cross-sectional regression analysis
    # Implements two-way clustered inference as standard in asset pricing
    regression_params: RegressionParams

    # Parameters controlling SHAP feature attribution analysis
    # Provides model-agnostic interpretability for relative importance assessment
    shap_params: SHAPParams

    # Parameters controlling LLM-based interpretability pipeline
    # Orchestrates topic labeling, metatopic creation, and token classification
    interpretability_params: InterpretabilityParams

    # Parameters controlling market efficiency trading simulation
    # Tests post-open profitability of soft information signals
    trading_simulation_params: TradingSimulationParams

    # Parameters controlling Precision@k evaluation of early access scenarios
    # Evaluates value of pre-release information access as in hacking incidents
    hacking_simulation_params: HackingSimulationParams

    # Parameters controlling Fama-French factor data integration
    # Enables risk-adjusted return evaluation in trading simulations
    factors_params: FactorsParams

    # Parameters controlling diagnostic analyses and reproducibility checks
    # Validates results against study benchmarks and ensures reproducibility
    diagnostics_params: DiagnosticsParams

    # Parameters ensuring compliance with governance and ethical requirements
    # Tracks model versions, data usage rights, and privacy considerations
    compliance_params: ComplianceParams

    class Config:
        """
        Pydantic configuration for the StudyConfig model.

        This configuration ensures that the model behaves appropriately
        for research reproducibility and validation requirements.
        """
        # Allow population by field name for flexibility
        allow_population_by_field_name = True

        # Validate all fields on assignment to catch errors early
        validate_assignment = True

        # Use enum values for better serialization
        use_enum_values = True

        # Generate schema with field descriptions for documentation
        schema_extra = {
            "title": "Earnings Press Release Analysis Study Configuration",
            "description": "Complete configuration schema for reproducible analysis of earnings press releases and stock return prediction"
        }

    @root_validator
    def validate_global_consistency(cls, values: Dict[str, Any]) -> Dict[str, Any]:
        """
        Validates global consistency across all parameter groups to ensure
        the configuration forms a coherent and executable research pipeline.

        This validator performs cross-parameter consistency checks that cannot
        be validated within individual parameter groups, ensuring temporal
        alignment, method compatibility, and logical coherence.

        Args:
            values: Dictionary containing all validated parameter groups

        Returns:
            Dictionary of validated parameter values

        Raises:
            ValueError: If cross-parameter inconsistencies are detected
        """
        # Extract parameter groups for validation
        data_params = values.get('data_params')
        modeling_params = values.get('modeling_params')
        regression_params = values.get('regression_params')
        diagnostics_params = values.get('diagnostics_params')

        # Validate temporal consistency across parameter groups
        if data_params and modeling_params and regression_params:
            # Check that study period encompasses modeling and regression periods
            study_start = data_params.start_date
            study_end = data_params.end_date

            # Extract rolling estimation years
            rolling_years = modeling_params.rolling_estimation.score_years
            if rolling_years:
                earliest_score_year = min(rolling_years)
                latest_score_year = max(rolling_years)

                # Check that score years fall within study period
                study_start_year = int(study_start.split('-')[0])
                study_end_year = int(study_end.split('-')[0])

                if earliest_score_year < study_start_year:
                    raise ValueError(f"Earliest score year ({earliest_score_year}) "
                                   f"precedes study start year ({study_start_year})")

                if latest_score_year > study_end_year:
                    raise ValueError(f"Latest score year ({latest_score_year}) "
                                   f"exceeds study end year ({study_end_year})")

            # Check regression sample period consistency
            reg_start, reg_end = regression_params.sample_period
            if reg_start < study_start or reg_end > study_end:
                raise ValueError("Regression sample period must fall within study period")

        # Validate random seed consistency across components
        if modeling_params and diagnostics_params:
            # Ensure global random seed is used consistently
            global_seed = diagnostics_params.random_state_global
            lasso_seed = modeling_params.lasso.random_seed

            # Check that component seeds align with global seed
            if lasso_seed != global_seed:
                raise ValueError(f"Lasso random seed ({lasso_seed}) should match "
                               f"global random seed ({global_seed})")

        # Validate vectorization method consistency with aggregation
        vectorization_params = values.get('vectorization_params')
        if vectorization_params and modeling_params:
            # Check that aggregation includes only enabled vectorization methods
            agg_models = modeling_params.soft_signal_aggregation.models_included

            # Validate that BKMX is included only if enabled
            if 'bkxmx' in agg_models and not vectorization_params.bkxmx.use_bkxmx:
                raise ValueError("BKMX included in aggregation but not enabled in vectorization")

        return values



# ==============================================================================
# Task 1, Step 2: DataFrame Schema Validation
# ==============================================================================

def _validate_dataframe_schema(
    df: pd.DataFrame,
    config: StudyConfig
) -> List[str]:
    """
    Validates the DataFrame's index, columns, and dtypes against the full
    project specification.

    This function performs a comprehensive and rigorous check of the input
    DataFrame's structure to ensure it conforms to the required data dictionary.
    It validates the MultiIndex levels, names, and types, and then checks every
    required column for its presence and correct data type, including special
    handling for timezone-aware datetimes and categorical variables.

    Args:
        df: The input DataFrame of earnings announcement events.
        config: The validated study configuration object, used here to get the
                expected timezone.

    Returns:
        A list of string error messages. An empty list indicates that the
        DataFrame schema is fully valid.
    """
    # Initialize a list to collect all validation error messages.
    errors: List[str] = []

    # ==========================================================================
    # Define the Complete Expected Schema
    # ==========================================================================

    # Define the expected structure and dtypes for the MultiIndex.
    expected_index = {
        'n_levels': 3,
        'names': ['ann_trade_date', 'permno', 'event_id'],
        'dtypes': {
            'ann_trade_date': f"datetime64[ns, {config.data_params.timezone}]",
            'permno': 'int64',
            'event_id': 'int64'
        }
    }

    # Define the expected columns and their dtypes. This is the complete list.
    expected_columns: Dict[str, str] = {
        'ann_calendar_date': f"datetime64[ns, {config.data_params.timezone}]",
        'pr_timestamp_et': f"datetime64[ns, {config.data_params.timezone}]",
        'filing_timestamp_et': f"datetime64[ns, {config.data_params.timezone}]",
        'pr_session': 'category',
        'pr_is_after_hours': 'bool',
        'ticker': 'object',
        'cusip8': 'object',
        'cik': 'int64',
        'company_name': 'object',
        'ibes_ticker': 'object',
        'edgar_accession_number': 'object',
        'edgar_form_type': 'object',
        'pr_exhibit_id': 'object',
        'edgar_filing_url': 'object',
        'pr_attachment_url': 'object',
        'doc_mime_type': 'object',
        'doc_language': 'object',
        'actual_eps': 'float64',
        'consensus_forecast': 'float64',
        'analyst_coverage_count': 'int64',
        'forecast_window_days': 'int64',
        'forecast_window_start_ts_et': f"datetime64[ns, {config.data_params.timezone}]",
        'forecast_cutoff_ts_et': f"datetime64[ns, {config.data_params.timezone}]",
        'consensus_method': 'object',
        'surprise': 'float64',
        'price_t_minus_5_adj': 'float64',
        'price_prior_close_adj': 'float64',
        'price_ann_close_adj': 'float64',
        'ret_announce_cc': 'float64',
        'market_cap_t_minus_1': 'float64',
        'nbbo_945_bid': 'float64',
        'nbbo_945_ask': 'float64',
        'nbbo_945_midquote': 'float64',
        'nbbo_945_rel_spread': 'float64',
        'nbbo_945_is_valid': 'bool',
        'nbbo_945_condition_codes': 'object',
        'nbbo_close_bid': 'float64',
        'nbbo_close_ask': 'float64',
        'nbbo_close_is_valid': 'bool',
        'ff_mkt_rf': 'float64',
        'ff_hml': 'float64',
        'ff_smb': 'float64',
        'rf': 'float64',
        'press_release_html_raw': 'object',
        'press_release_text_clean': 'object',
        'doc_length_raw': 'int64',
        'doc_length_clean': 'int64',
        'cleaning_removed_tables': 'bool',
        'cleaning_removed_forward_looking': 'bool',
        'cleaning_removed_non_gaap': 'bool',
        'cleaning_removed_contacts': 'bool',
        'cleaning_removed_exhibit_headers': 'bool',
        'cleaning_removed_urls_emails_numbers': 'bool',
        'cleaning_lemmatized': 'bool',
        'cleaning_stopwords_removed': 'bool',
        'cleaning_notes': 'object',
        'timezone': 'object',
        'data_version_stamp': 'object'
    }

    # ==========================================================================
    # Perform Validation
    # ==========================================================================

    # --- Index Validation ---
    # Check the number of index levels.
    if df.index.nlevels != expected_index['n_levels']:
        errors.append(f"Index validation failed: Expected {expected_index['n_levels']} levels, found {df.index.nlevels}.")

    # Check the names of the index levels.
    if list(df.index.names) != expected_index['names']:
        errors.append(f"Index validation failed: Expected names {expected_index['names']}, found {list(df.index.names)}.")

    # Check the data type of each index level.
    for i, name in enumerate(expected_index['names']):
        level_series = df.index.get_level_values(i)
        expected_dtype = expected_index['dtypes'][name]

        if 'datetime' in expected_dtype:
            # For datetimes, check both the base type and the timezone.
            if not pd.api.types.is_datetime64_any_dtype(level_series):
                errors.append(f"Index level '{name}' dtype mismatch: Expected datetime, found {level_series.dtype}.")
            elif str(level_series.tz) != config.data_params.timezone:
                errors.append(f"Index level '{name}' timezone mismatch: Expected '{config.data_params.timezone}', found '{level_series.tz}'.")
        elif str(level_series.dtype) != expected_dtype:
            errors.append(f"Index level '{name}' dtype mismatch: Expected '{expected_dtype}', found '{level_series.dtype}'.")

    # --- Column Validation ---
    # Check for the presence of all required columns.
    missing_cols = set(expected_columns.keys()) - set(df.columns)
    if missing_cols:
        errors.append(f"Column validation failed: Missing required columns: {sorted(list(missing_cols))}.")

    # Check the data type of each present required column.
    for col, expected_dtype in expected_columns.items():
        if col in df.columns:
            actual_dtype = df[col].dtype

            if 'datetime' in expected_dtype:
                if not pd.api.types.is_datetime64_any_dtype(actual_dtype):
                    errors.append(f"Column '{col}' dtype mismatch: Expected datetime, found '{actual_dtype}'.")
                elif str(df[col].dt.tz) != config.data_params.timezone:
                    errors.append(f"Column '{col}' timezone mismatch: Expected '{config.data_params.timezone}', found '{df[col].dt.tz}'.")

            elif expected_dtype == 'category':
                if not pd.api.types.is_categorical_dtype(actual_dtype):
                    errors.append(f"Column '{col}' dtype mismatch: Expected 'category', found '{actual_dtype}'.")
                # Special validation for 'pr_session' categories.
                elif col == 'pr_session':
                    expected_cats = {'after_close', 'pre_open', 'regular_hours', 'unknown'}
                    actual_cats = set(df[col].cat.categories)
                    if actual_cats != expected_cats:
                        errors.append(f"Column '{col}' categories mismatch: Expected {expected_cats}, found {actual_cats}.")

            elif str(actual_dtype) != expected_dtype:
                errors.append(f"Column '{col}' dtype mismatch: Expected '{expected_dtype}', found '{actual_dtype}'.")

    # --- Data Integrity Check ---
    # Check for columns that are entirely null, which indicates a data loading issue.
    for col in df.columns:
        if df[col].isnull().all():
            errors.append(f"Data integrity warning: Column '{col}' contains only NULL values.")

    return errors

# ==============================================================================
# Task 1, Step 3: Cross-Field Consistency Validation
# ==============================================================================

def _validate_dataframe_consistency(
    df: pd.DataFrame,
    config: StudyConfig
) -> List[str]:
    """
    Performs cross-field consistency checks on the DataFrame.

    Args:
        df: The input DataFrame of earnings announcement events.
        config: The validated study configuration object.

    Returns:
        A list of error messages. An empty list indicates success.
    """
    # Initialize a list to collect consistency error messages.
    errors = []
    # Define a small tolerance for floating-point comparisons.
    tolerance = 1e-9

    # --- Equation Consistency Checks ---
    # Check 1: ret_announce_cc = price_ann_close_adj / price_prior_close_adj - 1
    # Perform check only on rows where denominators are non-zero and non-null.
    valid_ret_mask = df['price_prior_close_adj'].notna() & (df['price_prior_close_adj'] != 0)
    if not np.allclose(
        df.loc[valid_ret_mask, 'ret_announce_cc'],
        df.loc[valid_ret_mask, 'price_ann_close_adj'] / df.loc[valid_ret_mask, 'price_prior_close_adj'] - 1,
        atol=tolerance,
        equal_nan=True # Treat corresponding NaNs as equal
    ):
        errors.append("Consistency check failed for 'ret_announce_cc' calculation.")

    # Check 2: surprise = (actual_eps - consensus_forecast) / price_t_minus_5_adj
    # Perform check only on rows where denominators are non-zero and non-null.
    valid_surprise_mask = df['price_t_minus_5_adj'].notna() & (df['price_t_minus_5_adj'] > 0)
    if not np.allclose(
        df.loc[valid_surprise_mask, 'surprise'],
        (df.loc[valid_surprise_mask, 'actual_eps'] - df.loc[valid_surprise_mask, 'consensus_forecast']) / df.loc[valid_surprise_mask, 'price_t_minus_5_adj'],
        atol=tolerance,
        equal_nan=True
    ):
        errors.append("Consistency check failed for 'surprise' calculation.")

    # --- Logical Consistency Checks ---
    # Check 3: pr_is_after_hours should align with pr_session.
    # Find rows where pr_is_after_hours is True but pr_session is not 'after_close' or 'pre_open'.
    after_hours_violation = df['pr_is_after_hours'] & ~df['pr_session'].isin(['after_close', 'pre_open'])
    if after_hours_violation.any():
        errors.append(
            f"Logical consistency failed: {after_hours_violation.sum()} rows have "
            f"'pr_is_after_hours'=True but 'pr_session' is not 'after_close' or 'pre_open'."
        )

    # Check 4: Analyst coverage requirement.
    if config.data_params.require_analyst_coverage:
        # Find rows where coverage is zero or missing.
        no_coverage = (df['analyst_coverage_count'] == 0) | df['analyst_coverage_count'].isna()
        if no_coverage.any():
            errors.append(
                f"Data requirement failed: {no_coverage.sum()} events have zero or missing "
                f"analyst coverage, but 'require_analyst_coverage' is True."
            )

    return errors


# ==============================================================================
# Task 1: Orchestrator Function
# ==============================================================================

def validate_inputs_and_schema(
    events_df: pd.DataFrame,
    study_params: Dict[str, Any]
) -> Tuple[pd.DataFrame, StudyConfig]:
    """
    Orchestrates the validation of configuration and DataFrame schema/integrity.

    This function serves as the main entry point for Task 1. It validates the
    study configuration, the DataFrame schema, and the cross-field consistency
    of the data. It is a critical first step to ensure the integrity of the
    entire research pipeline.

    Args:
        events_df: A pandas DataFrame containing the raw event data.
        study_params: A dictionary containing the study's configuration
                      parameters, expected to have a top-level key 'study_config'.

    Returns:
        A tuple containing:
        - The original DataFrame (if all validations pass).
        - A validated Pydantic StudyConfig object.

    Raises:
        ValueError: If any validation step fails, containing a detailed
                    and aggregated list of all detected errors.
        KeyError: If the 'study_config' key is missing from study_params.
    """
    # --- Step 0: Initial Input Checks ---
    # Ensure the main configuration key exists.
    if "study_config" not in study_params:
        raise KeyError("The 'study_config' key is missing from the study_params dictionary.")
    # Ensure the input is a pandas DataFrame.
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("The 'events_df' argument must be a pandas DataFrame.")

    # Initialize a list to aggregate all errors from all validation steps.
    all_errors = []

    # --- Step 1: Validate Configuration Dictionary ---
    # This step is critical. If the config is invalid, subsequent steps may fail unpredictably.
    # It returns a validated config object or raises a ValueError.
    try:
        validated_config = _validate_configuration(study_params["study_config"])
    except ValueError as e:
        # If config validation fails, we cannot proceed. Raise immediately.
        raise ValueError(f"Critical configuration error:\n{e}")

    # --- Step 2: Validate DataFrame Schema ---
    # This step checks the structure (index, columns, dtypes) of the DataFrame.
    schema_errors = _validate_dataframe_schema(events_df, validated_config)
    if schema_errors:
        all_errors.extend(schema_errors)

    # --- Step 3: Validate Cross-Field Consistency ---
    # This step checks the logical and mathematical integrity of the data.
    consistency_errors = _validate_dataframe_consistency(events_df, validated_config)
    if consistency_errors:
        all_errors.extend(consistency_errors)

    # --- Final Error Aggregation and Reporting ---
    # If any errors were collected from the DataFrame validation steps, raise a single, comprehensive error.
    if all_errors:
        error_report = "\n- ".join(all_errors)
        raise ValueError(f"Input data validation failed with the following errors:\n- {error_report}")

    # If all validations pass, return the original DataFrame and the validated config object.
    # No modifications are made to the DataFrame in this function.
    return events_df, validated_config
````

````python
# Task 2: Apply After-Hours Filter and Sample Restrictions

# Define a type alias for the statistics dictionary for clarity.
FilterStats = Dict[str, Dict[str, int]]

# ==============================================================================
# Task 2, Step 1: After-Hours Filter
# ==============================================================================

def _filter_by_release_session(
    df: pd.DataFrame,
    config: StudyConfig
) -> Tuple[pd.DataFrame, Dict[str, int]]:
    """
    Filters the DataFrame to include only after-hours press releases.

    This function implements the study's primary focus on news released
    outside of regular trading hours (4:00 p.m. to 9:30 a.m. ET). The
    decision to filter is controlled by the 'allow_regular_hours' parameter
    in the study configuration.

    Args:
        df: The input DataFrame of announcement events.
        config: The validated study configuration object.

    Returns:
        A tuple containing:
        - A new DataFrame with only after-hours events.
        - A dictionary with statistics on the number of events removed.
    """
    # Record the number of observations before applying the filter.
    initial_count = len(df)

    # If the configuration allows regular hours, skip this filter.
    if config.data_params.allow_regular_hours:
        # Prepare statistics indicating no change.
        stats = {
            'removed_count': 0,
            'remaining_count': initial_count
        }
        # Return the original DataFrame and the stats.
        return df.copy(), stats

    # Define the filter mask to keep only rows where 'pr_is_after_hours' is True.
    # This correctly handles NaN values, as they will not satisfy the condition.
    after_hours_mask = df['pr_is_after_hours'] == True

    # Apply the mask to create a new, filtered DataFrame.
    filtered_df = df[after_hours_mask].copy()

    # Record the number of observations remaining after the filter.
    final_count = len(filtered_df)

    # Compile statistics for this filtering step.
    stats = {
        'removed_count': initial_count - final_count,
        'remaining_count': final_count
    }

    # Return the filtered DataFrame and the associated statistics.
    return filtered_df, stats

# ==============================================================================
# Task 2, Step 2: Analyst Coverage and Document Length Filters
# ==============================================================================

def _filter_by_coverage_and_length(
    df: pd.DataFrame,
    config: StudyConfig
) -> Tuple[pd.DataFrame, Dict[str, int]]:
    """
    Applies filters for analyst coverage and document character length.

    This function enforces two key data quality requirements:
    1. Analyst Coverage: If required by the config, it removes events that
       do not have at least one analyst forecast.
    2. Document Length: It removes events where the cleaned press release
       text is either too short or too long, based on configured thresholds.

    Args:
        df: The input DataFrame of announcement events.
        config: The validated study configuration object.

    Returns:
        A tuple containing:
        - A new DataFrame that satisfies the coverage and length criteria.
        - A dictionary with statistics on the number of events removed.
    """
    # Record the number of observations before applying filters.
    initial_count = len(df)

    # Initialize the DataFrame to be filtered.
    filtered_df = df

    # --- Analyst Coverage Filter ---
    # Apply this filter only if specified in the configuration.
    if config.data_params.require_analyst_coverage:
        # Create a mask to keep rows with one or more analysts.
        # This correctly handles NaNs in 'analyst_coverage_count' by excluding them.
        coverage_mask = filtered_df['analyst_coverage_count'] > 0
        # Apply the filter.
        filtered_df = filtered_df[coverage_mask]

    # --- Document Length Filter ---
    # Retrieve min/max character counts from the configuration.
    min_chars = config.data_params.min_document_chars
    max_chars = config.data_params.max_document_chars

    # Create a mask to keep rows where the cleaned document length is within bounds.
    # The .between() method is efficient and correctly handles NaN values.
    length_mask = filtered_df['doc_length_clean'].between(min_chars, max_chars, inclusive='both')

    # Apply the filter.
    filtered_df = filtered_df[length_mask].copy()

    # Record the number of observations remaining after all filters.
    final_count = len(filtered_df)

    # Compile statistics for this combined filtering step.
    stats = {
        'removed_count': initial_count - final_count,
        'remaining_count': final_count
    }

    # Return the filtered DataFrame and the statistics.
    return filtered_df, stats

# ==============================================================================
# Task 2, Step 3: Language, Duplicate, and Date Range Filters
# ==============================================================================

def _filter_by_language_duplicates_and_date(
    df: pd.DataFrame,
    config: StudyConfig
) -> Tuple[pd.DataFrame, Dict[str, int]]:
    """
    Applies final filters for language, duplicates, and study date range.

    This function performs three critical cleaning steps in a specific order to
    efficiently reduce the sample to the final set of unique, relevant events.

    1.  **Language Filter**: Restricts the sample to specified languages (e.g., English).
    2.  **Date Range Filter**: Confines the sample to the official study period.
    3.  **Duplicate Filter**: Removes duplicate press releases for the same event,
        keeping the one associated with the latest SEC filing timestamp. This step
        is implemented using a `groupby().apply()` method to adhere strictly to
        the specified project blueprint.

    Args:
        df: The input DataFrame of announcement events.
        config: The validated study configuration object.

    Returns:
        A tuple containing:
        - The final, cleaned DataFrame, sorted by its original index.
        - A dictionary with statistics on the number of events removed by this
          combined filtering stage.
    """
    # --- Input Validation ---
    # Ensure the input is a pandas DataFrame.
    if not isinstance(df, pd.DataFrame):
        raise TypeError("Input 'df' must be a pandas DataFrame.")
    # Ensure the config is a valid StudyConfig object.
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # Record the number of observations before applying any filters in this function.
    initial_count = len(df)

    # Create a working copy to apply filters to.
    filtered_df = df.copy()

    # --- Filter 1: Language ---
    # Create a boolean mask to keep rows with the specified document languages.
    # This correctly handles NaNs in the 'doc_language' column by excluding them.
    language_mask = filtered_df['doc_language'].isin(config.data_params.keep_language)
    # Apply the language filter.
    filtered_df = filtered_df[language_mask]

    # --- Filter 2: Date Range ---
    # Parse the start and end dates from the configuration string into date objects.
    start_date = datetime.datetime.fromisoformat(config.data_params.start_date).date()
    end_date = datetime.datetime.fromisoformat(config.data_params.end_date).date()

    # Create a boolean mask to keep events within the specified date range (inclusive).
    # Comparing the .dt.date attribute is robust to time-of-day and timezone details.
    date_mask = filtered_df['pr_timestamp_et'].dt.date.between(start_date, end_date, inclusive='both')
    # Apply the date range filter.
    filtered_df = filtered_df[date_mask]

    # --- Filter 3: Duplicate Removal ---
    # Define the columns that uniquely identify a specific press release event.
    deduplication_keys = ['permno', 'pr_timestamp_et', 'pr_exhibit_id']

    # This section implements the deduplication logic as specified by the blueprint,
    # using a groupby-apply method for absolute fidelity to the instructions.

    # Group the DataFrame by the unique event identifiers.
    # `group_keys=False` prevents the group keys from being added to the index.
    # `dropna=False` ensures that events with NaN in keys are treated as a group.
    grouped = filtered_df.groupby(deduplication_keys, group_keys=False, dropna=False)

    # For each group of duplicates, apply a function that finds the index of the
    # row with the maximum (i.e., latest) 'filing_timestamp_et' and selects it.
    # This lambda function is applied to each sub-DataFrame (group) `g`.
    deduplicated_df = grouped.apply(lambda g: g.loc[g['filing_timestamp_et'].idxmax()])

    # Restore the original sort order of the DataFrame by sorting the index.
    final_df = deduplicated_df.sort_index()

    # Record the number of observations remaining after all filters.
    final_count = len(final_df)

    # Compile statistics for this combined filtering step.
    stats = {
        'removed_count': initial_count - final_count,
        'remaining_count': final_count
    }

    # Return the final filtered DataFrame and the associated statistics.
    return final_df, stats

# ==============================================================================
# Task 2: Orchestrator Function
# ==============================================================================

def apply_sample_filters(
    events_df: pd.DataFrame,
    config: StudyConfig
) -> Tuple[pd.DataFrame, FilterStats]:
    """
    Orchestrates the sequential application of all sample selection filters.

    This function provides a complete, end-to-end pipeline for cleaning the
    raw event data and producing the final analytical sample. It calls a
    series of modular filtering functions in a specific order, logging the
    impact of each step to ensure transparency and reproducibility.

    Args:
        events_df: The initial, unfiltered DataFrame of announcement events.
        config: The validated study configuration object.

    Returns:
        A tuple containing:
        - The final, filtered pandas DataFrame ready for analysis.
        - A dictionary containing a detailed report of the filtering process,
          tracking the number of observations removed at each stage.
    """
    # --- Input Validation ---
    # Ensure the input is a pandas DataFrame.
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    # Ensure the config is a valid StudyConfig object (or a compatible Pydantic model).
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # Initialize a dictionary to store the comprehensive filtering report.
    filter_report: FilterStats = {}

    # Make a copy to avoid modifying the original DataFrame in place.
    df_in_progress = events_df.copy()

    # Record the initial state before any filters are applied.
    filter_report['initial'] = {
        'removed_count': 0,
        'remaining_count': len(df_in_progress)
    }

    # --- Step 1: Apply After-Hours Filter ---
    # Filter for events occurring outside of regular trading hours.
    df_in_progress, stats = _filter_by_release_session(df_in_progress, config)
    filter_report['after_hours_filter'] = stats

    # --- Step 2: Apply Coverage and Length Filters ---
    # Filter based on analyst coverage and document length constraints.
    df_in_progress, stats = _filter_by_coverage_and_length(df_in_progress, config)
    filter_report['coverage_and_length_filter'] = stats

    # --- Step 3: Apply Language, Duplicate, and Date Filters ---
    # Apply the final set of filters for language, duplicates, and date range.
    df_in_progress, stats = _filter_by_language_duplicates_and_date(df_in_progress, config)
    filter_report['language_duplicate_date_filter'] = stats

    # The final DataFrame after all filtering steps.
    final_df = df_in_progress

    # Return the final analytical sample and the detailed filtering report.
    return final_df, filter_report
````

````python
# Task 3: Cleanse and Standardize Text Data

# ==============================================================================
# Task 3, Step 1: HTML Parsing and Table Removal
# ==============================================================================

def _extract_text_from_html(html_series: pd.Series) -> pd.Series:
    """
    Parses a Series of HTML strings, removes table elements, and extracts text.

    This function serves the first stage of text cleaning by converting raw
    HTML content into a preliminary clean text format. It is designed to be
    robust against malformed HTML.

    Args:
        html_series: A pandas Series where each element is a raw HTML string.

    Returns:
        A pandas Series of the same index containing the extracted text.
    """
    # Define a helper function to process a single HTML document.
    def parse_single_html(html_content: str) -> str:
        # Perform input validation for the content of the series.
        if not isinstance(html_content, str) or not html_content:
            return ""
        try:
            # Initialize BeautifulSoup with the lxml parser for speed and robustness.
            soup = BeautifulSoup(html_content, 'lxml')

            # Find all table elements within the parsed document.
            tables = soup.find_all('table')

            # Decompose each table, removing it and all its contents from the parse tree.
            for table in tables:
                table.decompose()

            # Extract all remaining text. Use a space separator to prevent words
            # from different tags from merging. Strip removes excess whitespace.
            return soup.get_text(separator=' ', strip=True)

        except Exception:
            # If any parsing error occurs, return an empty string for this document.
            # This ensures the entire process does not fail on one bad file.
            return ""

    # Apply the parsing function to each element of the input Series.
    return html_series.apply(parse_single_html)

# ==============================================================================
# Task 3, Step 2: Boilerplate and Pattern Removal
# ==============================================================================

def _remove_boilerplate_and_patterns(
    text_series: pd.Series,
    config: StudyConfig
) -> pd.Series:
    """
    Removes common boilerplate sections and noisy patterns from a text Series.

    This function uses a series of pre-compiled regular expressions to strip
    uninformative content such as forward-looking statements, non-GAAP
    disclosures, contact information, URLs, and emails.

    Args:
        text_series: A pandas Series of text data.
        config: The validated study configuration object.

    Returns:
        A pandas Series with the specified patterns removed.
    """
    # Initialize the text to be cleaned.
    cleaned_text = text_series.copy()

    # Create a list to hold all regex patterns to be applied.
    patterns_to_remove = []

    # --- Compile Boilerplate Patterns ---
    # Conditionally add patterns based on the configuration.
    if config.text_cleaning_params.strip_boilerplate_sections.get("forward_looking_statements", False):
        patterns_to_remove.append(re.compile(r'forward-looking statements.*?safe harbor', re.IGNORECASE | re.DOTALL))
    if config.text_cleaning_params.strip_boilerplate_sections.get("non_gaap_disclosures", False):
        patterns_to_remove.append(re.compile(r'reconciliation of non-gaap.*?gaap', re.IGNORECASE | re.DOTALL))
    if config.text_cleaning_params.strip_boilerplate_sections.get("contacts_and_ir_headers", False):
        patterns_to_remove.append(re.compile(r'(investor relations|media contact|for more information).*', re.IGNORECASE))
    if config.text_cleaning_params.strip_boilerplate_sections.get("exhibit_headers_numbers", False):
        patterns_to_remove.append(re.compile(r'exhibit \d{1,2}\.\d{1,2}', re.IGNORECASE))

    # --- Compile Common Noise Patterns ---
    if config.text_cleaning_params.remove_patterns.get("urls", False):
        patterns_to_remove.append(re.compile(r'https?://\S+|www\.\S+'))
    if config.text_cleaning_params.remove_patterns.get("emails", False):
        patterns_to_remove.append(re.compile(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'))
    if config.text_cleaning_params.remove_patterns.get("phone_numbers", False):
        patterns_to_remove.append(re.compile(r'\(?\d{3}\)?[-.\s]?\d{3}[-.\s]?\d{4}'))

    # Sequentially apply each compiled regex pattern to the text Series.
    for pattern in patterns_to_remove:
        cleaned_text = cleaned_text.str.replace(pattern, ' ', regex=True)

    return cleaned_text

# ==============================================================================
# Task 3, Step 3: Text Normalization
# ==============================================================================

def _normalize_text(text_series: pd.Series, config: StudyConfig) -> pd.Series:
    """
    Applies final normalization steps to the text data.

    This function performs Unicode normalization, lowercasing, and whitespace
    collapsing to produce the final, standardized text for analysis.

    Args:
        text_series: A pandas Series of text data.
        config: The validated study configuration object.

    Returns:
        A pandas Series of fully normalized text.
    """
    # Initialize the text to be normalized.
    normalized_text = text_series.copy()

    # --- Unicode Normalization ---
    # Apply NFKC normalization to handle compatibility characters and standardize the text.
    if config.text_cleaning_params.normalization.get("unicode_normalize", "NFKC") == "NFKC":
        normalized_text = normalized_text.apply(
            lambda x: unicodedata.normalize('NFKC', x) if isinstance(x, str) else x
        )

    # --- Lowercasing ---
    # Convert all text to lowercase for case-insensitive analysis.
    if config.text_cleaning_params.normalization.get("lowercase", False):
        normalized_text = normalized_text.str.lower()

    # --- Whitespace Collapsing ---
    # Replace all sequences of whitespace characters with a single space and strip ends.
    if config.text_cleaning_params.normalization.get("whitespace_collapse", False):
        normalized_text = normalized_text.str.replace(r'\s+', ' ', regex=True).str.strip()

    return normalized_text

# ==============================================================================
# Task 3: Orchestrator Function
# ==============================================================================

def _process_text_for_bow(text_series: pd.Series) -> pd.Series:
    """
    Applies heavy NLP preprocessing for Bag-of-Words models.

    This function takes a series of cleaned text and applies tokenization,
    stop-word removal, and lemmatization. The resulting tokens are joined
    back into a string, creating a text representation suitable for models
    like LDA that rely on word counts.

    Args:
        text_series: A pandas Series of cleaned text (e.g., after HTML parsing
                     and boilerplate removal).

    Returns:
        A pandas Series of the same index containing the heavily processed text.
    """
    # Initialize the lemmatizer and the set of English stop words for efficiency.
    lemmatizer = WordNetLemmatizer()
    stop_words: Set[str] = set(stopwords.words('english'))

    # Define a helper function to process a single document.
    def process_single_document(text: str) -> str:
        # Handle empty or non-string inputs gracefully.
        if not isinstance(text, str) or not text:
            return ""

        # 1. Tokenize the document into words.
        tokens = word_tokenize(text)

        # 2. Filter out stop words and non-alphabetic tokens.
        filtered_tokens = [
            word for word in tokens
            if word.isalpha() and word not in stop_words
        ]

        # 3. Lemmatize the remaining tokens to their root form.
        lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]

        # 4. Join the processed tokens back into a single string.
        return " ".join(lemmatized_tokens)

    # Apply the processing function to each document in the series.
    # Note: This .apply() can be slow and is a candidate for parallelization
    # with libraries like pandarallel or Dask in a production setting.
    return text_series.apply(process_single_document)


def cleanse_and_standardize_text_data(
    events_df: pd.DataFrame,
    config: StudyConfig
) -> pd.DataFrame:
    """
    Orchestrates the end-to-end text cleaning and standardization pipeline,
    producing two distinct text representations for different model types.

    This function takes a DataFrame with raw HTML and generates two cleaned
    text columns:
    1. 'press_release_text_clean': Lightly processed text suitable for
       context-aware models like Transformers. This involves HTML parsing,
       boilerplate removal, and basic normalization.
    2. 'press_release_text_bow': Heavily processed text suitable for
       Bag-of-Words models like LDA. This version undergoes additional
       stop-word removal and lemmatization.

    Args:
        events_df: The DataFrame containing the column 'press_release_html_raw'.
        config: The validated study configuration object.

    Returns:
        The DataFrame augmented with four new columns:
        - 'press_release_text_clean': Text for Transformer models.
        - 'doc_length_clean': Character count of the 'clean' text.
        - 'press_release_text_bow': Text for Bag-of-Words models.
        - 'doc_length_bow': Character count of the 'bow' text.
    """
    # --- Input Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    if 'press_release_html_raw' not in events_df.columns:
        raise KeyError("DataFrame is missing required column 'press_release_html_raw'.")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # Create a working copy to avoid modifying the original DataFrame.
    df_processed = events_df.copy()

    # --- Pipeline for 'press_release_text_clean' (for Transformers) ---

    # Step 1: Extract text from HTML, removing tables.
    text_after_html = _extract_text_from_html(df_processed['press_release_html_raw'])

    # Step 2: Remove boilerplate sections and common noise patterns.
    text_after_boilerplate = _remove_boilerplate_and_patterns(text_after_html, config)

    # Step 3: Apply final normalization (Unicode, case, whitespace).
    final_clean_text = _normalize_text(text_after_boilerplate, config)

    # Add the 'clean' text column and its length to the DataFrame.
    df_processed['press_release_text_clean'] = final_clean_text
    df_processed['doc_length_clean'] = final_clean_text.str.len().fillna(0).astype(int)

    # --- Pipeline for 'press_release_text_bow' (for BoW Models) ---

    # Step 4: Apply additional BoW-specific processing to the 'clean' text.
    # This includes tokenization, stop-word removal, and lemmatization.
    final_bow_text = _process_text_for_bow(df_processed['press_release_text_clean'])

    # Add the 'bow' text column and its length to the DataFrame.
    df_processed['press_release_text_bow'] = final_bow_text
    df_processed['doc_length_bow'] = final_bow_text.str.len().fillna(0).astype(int)

    # Return the DataFrame augmented with both text representations.
    return df_processed
````

````python
# Task 4: Vectorize Documents via BKMX Taxonomy

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

class BKMXParams(BaseModel):
    """
    A Pydantic model for validating the 'bkxmx' section of the study
    configuration.

    This model defines the schema for parameters related to the BKMX
    (Bybee, Kelly, Manela, Xiu) taxonomy vectorization. It ensures that the
    path to the taxonomy definition and the list of selected metatopics are
    correctly specified.

    Attributes:
        taxonomy_resource_path (str): The file system path to the JSON file
                                      containing the BKMX taxonomy definition.
                                      The path must be a non-empty string.
        metatopics_selected (List[str]): A list of the names of the metatopics
                                         to be used as features. The list must
                                         contain at least one metatopic.
    """
    # The file path to the BKMX taxonomy JSON file.
    # Field validation ensures the path is a non-empty string.
    taxonomy_resource_path: str = Field(..., min_length=1)

    # A list of metatopic names to be extracted as features.
    # Field validation ensures the list is not empty.
    metatopics_selected: List[str] = Field(..., min_items=1)


class VectorizationParams(BaseModel):
    """
    A Pydantic model for validating the 'vectorization_params' section of the
    study configuration.

    This model serves as a container for nested parameter structures related
    to different text vectorization methods.

    Attributes:
        bkxmx (BKMXParams): A nested model containing the validated parameters
                            for the BKMX taxonomy vectorization method.
    """
    # A nested model for the BKMX-specific parameters.
    bkxmx: BKMXParams


class StudyConfig(BaseModel):
    """
    A Pydantic model representing a subsection of the main 'study_config'
    dictionary relevant to Task 4.

    This model ensures that the vectorization parameters, specifically those
    for the BKMX method, are present and correctly structured within the
    overall study configuration.

    Attributes:
        vectorization_params (VectorizationParams): The top-level container for
            all vectorization-related settings.
    """
    # The main container for all vectorization parameters.
    vectorization_params: VectorizationParams


# Define custom type aliases for clarity and maintainability.
TopicPhraseSet = Dict[int, Set[str]]
TopicToMetatopicMap = Dict[int, str]
ProcessedTaxonomy = Tuple[TopicPhraseSet, TopicToMetatopicMap]

# ==============================================================================
# Task 4, Step 1: Load and Validate BKMX Taxonomy
# ==============================================================================

def _load_and_process_bkmx_taxonomy(config: StudyConfig) -> ProcessedTaxonomy:
    """
    Loads the BKMX taxonomy from a JSON file, validates its structure, and
    processes it into efficient lookup data structures.

    This function is the first step in the BKMX vectorization pipeline. It
    ensures that the taxonomy definition is valid and prepares it for the
    computationally intensive task of document mapping.

    Args:
        config: The validated study configuration object, containing the path
                to the taxonomy file and the list of selected metatopics.

    Returns:
        A tuple containing two processed data structures:
        - topic_phrases: A dictionary mapping each of the 180 topic IDs (int)
                         to a set of its associated phrases (str).
        - topic_to_metatopic: A dictionary mapping each topic ID (int) to its
                              parent metatopic name (str).

    Raises:
        FileNotFoundError: If the taxonomy file specified in the config does not exist.
        ValueError: If the JSON is malformed, the taxonomy structure is invalid,
                    or configured metatopics are not found.
    """
    # Retrieve the path to the taxonomy file from the configuration.
    taxonomy_path = Path(config.vectorization_params.bkxmx.taxonomy_resource_path)

    # --- File Loading and Parsing ---
    try:
        # Open and load the JSON file containing the taxonomy definition.
        with taxonomy_path.open('r', encoding='utf-8') as f:
            taxonomy_data = json.load(f)
    except FileNotFoundError:
        # Raise a specific error if the file does not exist.
        raise FileNotFoundError(f"BKMX taxonomy file not found at: {taxonomy_path}")
    except json.JSONDecodeError as e:
        # Raise an error if the file is not valid JSON.
        raise ValueError(f"Failed to parse BKMX taxonomy JSON file: {e}")

    # --- Taxonomy Structure Validation ---
    # Check for the presence of essential top-level keys.
    if "topics" not in taxonomy_data or "metatopics" not in taxonomy_data:
        raise ValueError("Invalid BKMX taxonomy format: missing 'topics' or 'metatopics' key.")

    # Validate the list of selected metatopics from the config against the taxonomy.
    defined_metatopics = set(taxonomy_data["metatopics"].keys())
    selected_metatopics = set(config.vectorization_params.bkxmx.metatopics_selected)
    if not selected_metatopics.issubset(defined_metatopics):
        missing = selected_metatopics - defined_metatopics
        raise ValueError(f"Configured metatopics not found in taxonomy: {missing}")

    # --- Process into Efficient Lookup Structures ---
    # Initialize the dictionaries to be returned.
    topic_phrases: TopicPhraseSet = {}
    topic_to_metatopic: TopicToMetatopicMap = {}

    # Create a mapping from topic names to their integer IDs for robust lookup.
    topic_name_to_id = {details['name']: int(tid) for tid, details in taxonomy_data["topics"].items()}

    # Populate the topic_phrases dictionary.
    for topic_id_str, details in taxonomy_data["topics"].items():
        topic_id = int(topic_id_str)
        # Convert the list of phrases to a set for O(1) average lookup time.
        topic_phrases[topic_id] = set(details.get('phrases', []))

    # Populate the topic_to_metatopic mapping.
    for metatopic_name, metatopic_details in taxonomy_data["metatopics"].items():
        for topic_name in metatopic_details.get('topics', []):
            if topic_name in topic_name_to_id:
                topic_id = topic_name_to_id[topic_name]
                topic_to_metatopic[topic_id] = metatopic_name

    # Final validation to ensure all 180 topics are mapped.
    if len(topic_phrases) != 180 or len(topic_to_metatopic) != 180:
        raise ValueError(f"Taxonomy processing error: Expected 180 topics, but found "
                         f"{len(topic_phrases)} with phrases and {len(topic_to_metatopic)} mapped to metatopics.")

    return topic_phrases, topic_to_metatopic

# ==============================================================================
# Task 4, Step 2: Compute Document-Topic Intensities
# ==============================================================================

def _compute_bkmx_topic_intensities(
    text_series: pd.Series,
    topic_phrases: TopicPhraseSet
) -> pd.DataFrame:
    """
    Computes the intensity of 180 BKMX topics for each document using a
    highly efficient, vectorized approach.

    This function replaces a slow, iterative process with a scalable pipeline
    based on sparse matrix operations. It first counts the occurrences of all
    BKMX phrases across the entire corpus using CountVectorizer. It then uses
    matrix multiplication to aggregate these phrase counts into topic counts,
    and finally normalizes by document length to produce the topic intensity scores.

    Args:
        text_series: A pandas Series of cleaned, standardized text.
        topic_phrases: A dictionary mapping topic IDs to sets of their
                       associated phrases.

    Returns:
        A pandas DataFrame with the same index as `text_series` and 180
        columns, where each column `bkmx_topic_{id}` contains the normalized
        intensity for that topic.
    """
    # --- Input Validation ---
    if not isinstance(text_series, pd.Series):
        raise TypeError("Input 'text_series' must be a pandas Series.")
    if not isinstance(topic_phrases, dict):
        raise TypeError("Input 'topic_phrases' must be a dictionary.")

    # --- Step 1: Prepare Vocabulary and Mappings for Vectorization ---

    # Create a sorted list of all unique phrases from the taxonomy. This will
    # serve as the fixed vocabulary for the CountVectorizer.
    all_phrases = sorted(list(set.union(*topic_phrases.values())))

    # Create a mapping from each phrase to its integer index in the vocabulary list.
    phrase_to_idx = {phrase: i for i, phrase in enumerate(all_phrases)}

    # Create the transformation matrix that maps phrases to topics.
    # This will be a sparse matrix of shape (n_phrases, n_topics).
    n_phrases = len(all_phrases)
    n_topics = len(topic_phrases)
    topic_ids = sorted(topic_phrases.keys())

    # Prepare the data for constructing the sparse matrix.
    row_indices, col_indices, data = [], [], []
    for topic_idx, topic_id in enumerate(topic_ids):
        for phrase in topic_phrases[topic_id]:
            if phrase in phrase_to_idx:
                row_indices.append(phrase_to_idx[phrase])
                col_indices.append(topic_idx)
                data.append(1)

    # Construct the sparse transformation matrix.
    phrase_to_topic_matrix = csc_matrix((data, (row_indices, col_indices)), shape=(n_phrases, n_topics))

    # --- Step 2: Vectorize Phrase Counts ---

    # Instantiate CountVectorizer with the fixed BKMX vocabulary.
    # Using a pre-defined vocabulary is highly efficient.
    vectorizer = CountVectorizer(vocabulary=phrase_to_idx, lowercase=False)

    # Apply the vectorizer to the text corpus to get phrase counts.
    # The result is a sparse matrix of shape (n_docs, n_phrases).
    phrase_counts_matrix = vectorizer.transform(text_series.fillna(''))

    # --- Step 3: Aggregate to Topic Counts via Matrix Multiplication ---

    # Multiply the document-phrase matrix by the phrase-topic matrix.
    # (n_docs, n_phrases) @ (n_phrases, n_topics) -> (n_docs, n_topics)
    # This efficiently aggregates phrase counts into topic counts.
    topic_counts_matrix = phrase_counts_matrix @ phrase_to_topic_matrix

    # --- Step 4: Normalize by Document Length ---

    # Calculate the total number of tokens in each document.
    # We use a simple whitespace tokenizer for this count.
    # Adding 1e-9 to the denominator prevents division by zero for empty documents.
    doc_lengths = text_series.fillna('').str.split().str.len().values + 1e-9

    # Normalize the topic counts by dividing each row by its document length.
    # We need to reshape doc_lengths to (n_docs, 1) for broadcasting.
    normalized_intensities = topic_counts_matrix.toarray() / doc_lengths[:, np.newaxis]

    # --- Step 5: Create Final DataFrame ---

    # Create a DataFrame from the normalized intensity matrix.
    # Ensure the index is preserved and columns are named correctly.
    intensity_df = pd.DataFrame(
        normalized_intensities,
        index=text_series.index,
        columns=[f"bkmx_topic_{tid}" for tid in topic_ids]
    )

    return intensity_df

# ==============================================================================
# Task 4, Step 3: Aggregate to Metatopics
# ==============================================================================

def _aggregate_to_metatopics(
    intensity_df: pd.DataFrame,
    topic_to_metatopic: TopicToMetatopicMap,
    selected_metatopics: List[str]
) -> pd.DataFrame:
    """
    Aggregates 180 topic intensities into 11 selected metatopic intensities.

    This function groups the 180 topic columns by their parent metatopic and
    sums their intensities to produce a lower-dimensional representation.

    Args:
        intensity_df: DataFrame of 180 topic intensities.
        topic_to_metatopic: A dictionary mapping topic IDs to metatopic names.
        selected_metatopics: A list of metatopic names to include in the output.

    Returns:
        A pandas DataFrame with the same index as `intensity_df` and 11
        columns, one for each selected metatopic.
    """
    # Create a mapping from column name (e.g., 'bkmx_topic_0') to metatopic name.
    col_to_metatopic = {
        f"bkmx_topic_{tid}": meta_name
        for tid, meta_name in topic_to_metatopic.items()
    }

    # Group the DataFrame columns by the metatopic mapping.
    # The .groupby() on axis=1 performs column-wise grouping.
    # The .sum() aggregates the intensities within each group.
    metatopic_df = intensity_df.groupby(by=col_to_metatopic, axis=1).sum()

    # Filter to keep only the metatopics specified in the configuration.
    final_metatopic_df = metatopic_df[selected_metatopics]

    # Rename columns for clarity (e.g., 'Corporate Earnings' -> 'bkmx_Corporate_Earnings').
    final_metatopic_df.columns = [f"bkmx_{c.replace(' ', '_')}" for c in final_metatopic_df.columns]

    return final_metatopic_df

# ==============================================================================
# Task 4: Orchestrator Function
# ==============================================================================

def vectorize_text_with_bkmx(
    events_df: pd.DataFrame,
    config: StudyConfig
) -> pd.DataFrame:
    """
    Orchestrates the BKMX vectorization pipeline for a DataFrame of events.

    This function executes the complete workflow for converting cleaned text
    into a set of 11 metatopic features based on the BKMX taxonomy. It handles
    taxonomy loading, topic intensity computation, and aggregation into
    metatopics, finally merging the new features back into the original DataFrame.

    Args:
        events_df: The DataFrame containing the 'press_release_text_clean' column.
        config: The validated study configuration object.

    Returns:
        The input DataFrame augmented with 11 new columns, each representing
        the intensity of a BKMX metatopic.

    Raises:
        KeyError: If the required 'press_release_text_clean' column is missing.
        TypeError: If inputs are not of the expected type.
    """
    # --- Input Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    if 'press_release_text_clean' not in events_df.columns:
        raise KeyError("DataFrame is missing required column 'press_release_text_clean'.")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # --- Step 1: Load and Process Taxonomy ---
    # This step prepares the taxonomy for efficient use.
    topic_phrases, topic_to_metatopic = _load_and_process_bkmx_taxonomy(config)

    # --- Step 2: Compute 180 Topic Intensities ---
    # This step maps each document to the full 180-topic space.
    intensity_df = _compute_bkmx_topic_intensities(
        events_df['press_release_text_clean'],
        topic_phrases
    )

    # --- Step 3: Aggregate to 11 Metatopics ---
    # This step reduces the dimensionality to the 11 selected metatopics.
    metatopic_features_df = _aggregate_to_metatopics(
        intensity_df,
        topic_to_metatopic,
        config.vectorization_params.bkxmx.metatopics_selected
    )

    # --- Final Integration ---
    # Merge the new metatopic features back into the original DataFrame.
    # The join is performed on the index, which is preserved throughout the process.
    df_augmented = events_df.join(metatopic_features_df)

    return df_augmented
````

````python
# Task 5: Train and Apply Online LDA (oLDA) with Rolling Protocol

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

class OLAParams(BaseModel):
    """
    A Pydantic model for validating the 'olda' section of the study
    configuration.

    This model defines the schema for the hyperparameters of the Online Latent
    Dirichlet Allocation (oLDA) model, ensuring they are correctly typed and
    within valid ranges.

    Attributes:
        num_topics (int): The number of topics to extract. Must be > 1.
        learning_method (str): The learning method for LDA. Must be 'online'.
        learning_offset (float): A parameter that downweights early iterations
                                 in online learning. Must be > 0.
        decay (float): The learning rate decay, controlling how quickly old
                       information is forgotten. Must be in (0.5, 1.0].
        doc_topic_prior (float): The hyperparameter alpha for the document-topic
                                 distribution. Must be > 0.
        topic_word_prior (float): The hyperparameter eta for the topic-word
                                  distribution. Must be > 0.
        max_iter (int): The maximum number of iterations for the learning
                        algorithm. Must be a positive integer.
        random_seed (int): The seed for the random number generator to ensure
                           reproducibility.
    """
    # The number of latent topics to be extracted from the text corpus.
    num_topics: int = Field(..., gt=1)

    # The learning method for the LDA model. For this study, it must be 'online'.
    learning_method: constr(regex=r'^online$')

    # A parameter to control the influence of early batches in online learning.
    learning_offset: float = Field(..., gt=0)

    # The decay rate for the learning process, also known as kappa.
    decay: float = Field(..., gt=0.5, le=1.0)

    # The document-topic prior (alpha). A symmetric prior is used.
    doc_topic_prior: float = Field(..., gt=0)

    # The topic-word prior (eta). A symmetric prior is used.
    topic_word_prior: float = Field(..., gt=0)

    # The maximum number of passes over the data.
    max_iter: int = Field(..., gt=0)

    # The seed for reproducible random number generation.
    random_seed: int


class BoWVectorizerParams(BaseModel):
    """
    A Pydantic model for validating the 'bag_of_words_vectorizer' section of
    the study configuration.

    This model defines the schema for the scikit-learn CountVectorizer, which
    converts text documents into a matrix of token counts.

    Attributes:
        min_df (int): The minimum document frequency for a token to be included
                      in the vocabulary. Must be a positive integer.
        max_df (float): The maximum document frequency for a token. Must be
                        a float between 0.0 and 1.0.
        random_seed (int): The seed for any stochastic processes within the
                           vectorizer, ensuring reproducibility.
    """
    # The minimum number of documents a word must appear in to be included.
    min_df: int = Field(..., gt=0)

    # The maximum proportion of documents a word can appear in to be included.
    max_df: float = Field(..., gt=0.0, lt=1.0)

    # The seed for reproducible random state.
    random_seed: int


class RollingProtocolParams(BaseModel):
    """
    A Pydantic model for validating the 'rolling_protocol' section of the
    study configuration.

    This model defines the schema for the temporal cross-validation or
    rolling window procedure, which is essential for preventing look-ahead
    bias in time-series modeling.

    Attributes:
        initial_train_year (int): The first year of data to be included in the
                                  initial training set.
        score_years (List[int]): A list of the years for which out-of-sample
                                 predictions (scores) will be generated.
    """
    # The first year of data for the initial training window.
    initial_train_year: int

    # The list of years to be scored out-of-sample.
    score_years: List[int] = Field(..., min_items=1)


class VectorizationParams(BaseModel):
    """
    A Pydantic model for validating the 'vectorization_params' section of the
    study configuration.

    This model serves as a container for nested parameter structures related
    to different text vectorization methods.

    Attributes:
        olda (OLAParams): Validated parameters for the oLDA model.
        bag_of_words_vectorizer (BoWVectorizerParams): Validated parameters for
            the CountVectorizer.
        rolling_protocol (RollingProtocolParams): Validated parameters for the
            rolling window procedure.
    """
    # A nested model for the Online LDA hyperparameters.
    olda: OLAParams

    # A nested model for the Bag-of-Words vectorizer settings.
    bag_of_words_vectorizer: BoWVectorizerParams

    # A nested model for the rolling window protocol.
    rolling_protocol: RollingProtocolParams


class StudyConfig(BaseModel):
    """
    A Pydantic model representing a subsection of the main 'study_config'
    dictionary relevant to Task 5.

    This model ensures that the vectorization parameters, specifically those
    for the oLDA pipeline, are present and correctly structured within the
    overall study configuration.

    Attributes:
        vectorization_params (VectorizationParams): The top-level container for
            all vectorization-related settings.
    """
    # The main container for all vectorization parameters.
    vectorization_params: VectorizationParams

# ==============================================================================
# Task 5: Helper and Orchestrator Functions
# ==============================================================================

def _create_bow_and_lda_models(
    config: StudyConfig
) -> Tuple[CountVectorizer, LatentDirichletAllocation]:
    """
    Instantiates CountVectorizer and LatentDirichletAllocation models from config.

    This helper function acts as a factory for creating the scikit-learn models
    used in the oLDA pipeline. It centralizes model instantiation, ensuring they
    are consistently configured according to the study parameters for each
    iteration of the rolling window.

    Crucially, this function assumes that the text provided to the CountVectorizer
    has already been preprocessed (e.g., stop-word removal, lemmatization),
    and therefore does not include any preprocessing steps itself.

    Args:
        config: The validated study configuration object.

    Returns:
        A tuple containing:
        - An un-fitted CountVectorizer instance.
        - An un-fitted LatentDirichletAllocation instance.
    """
    # --- Input Validation ---
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # --- Instantiate CountVectorizer ---
    # Retrieve bag-of-words vectorizer parameters from the configuration.
    bow_params = config.vectorization_params.bag_of_words_vectorizer

    # Instantiate the CountVectorizer with the specified parameters.
    # Note: Preprocessing steps like stop-word removal are explicitly omitted here,
    # as this function assumes it will receive a pre-processed text series.
    vectorizer = CountVectorizer(
        analyzer="word",
        ngram_range=(1, 1),
        min_df=bow_params.min_df,
        max_df=bow_params.max_df,
        binary=False,
        lowercase=True # Lowercasing is a minimal, safe assumption for the vectorizer.
    )

    # --- Instantiate LatentDirichletAllocation ---
    # Retrieve online LDA parameters from the configuration.
    olda_params = config.vectorization_params.olda

    # Instantiate the LatentDirichletAllocation model for online learning.
    # The configuration is mapped directly to the scikit-learn model's arguments.
    lda = LatentDirichletAllocation(
        n_components=olda_params.num_topics,
        learning_method=olda_params.learning_method,
        learning_offset=olda_params.learning_offset,
        learning_decay=olda_params.decay,
        doc_topic_prior=olda_params.doc_topic_prior,
        topic_word_prior=olda_params.topic_word_prior,
        max_iter=olda_params.max_iter,
        random_state=olda_params.random_seed,
        n_jobs=-1  # Use all available CPU cores for fitting to improve performance.
    )

    # Return the two instantiated, but not yet fitted, model objects.
    return vectorizer, lda


def vectorize_text_with_olda_rolling(
    events_df: pd.DataFrame,
    config: StudyConfig
) -> Tuple[pd.DataFrame, LatentDirichletAllocation, CountVectorizer]:
    """
    Orchestrates the rolling oLDA vectorization pipeline and returns the final
    trained model artifacts.

    This function implements the dynamic topic modeling approach. It iterates
    year by year, training a new topic model on all data up to that point and
    then using it to generate topic features for the subsequent year.

    Crucially, this refactored version now also returns the final, most
    comprehensively trained LDA model and its corresponding CountVectorizer.
    These artifacts are essential for the downstream interpretability tasks.

    Args:
        events_df: The DataFrame containing the 'press_release_text_bow'
                   and 'pr_timestamp_et' columns.
        config: The validated study configuration object.

    Returns:
        A tuple containing:
        - The input DataFrame augmented with 50 new oLDA topic columns.
        - The final fitted `LatentDirichletAllocation` model object.
        - The final fitted `CountVectorizer` object.
    """
    # --- Input Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    required_cols = ['press_release_text_bow', 'pr_timestamp_et']
    if not all(col in events_df.columns for col in required_cols):
        raise KeyError(f"DataFrame is missing one or more required columns: {required_cols}.")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # Create a working copy and add a 'calendar_year' column for partitioning.
    df_augmented = events_df.copy()
    df_augmented['calendar_year'] = df_augmented['pr_timestamp_et'].dt.year

    # Retrieve rolling protocol parameters from the configuration.
    protocol_params = config.vectorization_params.rolling_protocol
    initial_train_year = protocol_params.initial_train_year
    score_years = protocol_params.score_years

    # Initialize a list to collect the topic distribution DataFrames.
    all_olda_features: List[pd.DataFrame] = []

    # Initialize variables to hold the final model and vectorizer.
    final_lda_model: LatentDirichletAllocation = None
    final_vectorizer: CountVectorizer = None

    # --- Main Rolling Loop ---
    for score_year in score_years:
        train_end_year = score_year - 1

        # Partition data into training and scoring sets.
        train_mask = df_augmented['calendar_year'].between(initial_train_year, train_end_year, inclusive='both')
        score_mask = df_augmented['calendar_year'] == score_year
        train_df = df_augmented.loc[train_mask]
        score_df = df_augmented.loc[score_mask]

        if train_df.empty or score_df.empty:
            continue

        # Instantiate fresh models for this fold.
        vectorizer, lda = _create_bow_and_lda_models(config)

        # Fit the CountVectorizer *only* on the training text data.
        # This uses the 'press_release_text_bow' column as per the remediation plan.
        train_bow = vectorizer.fit_transform(train_df['press_release_text_bow'])

        # Transform the scoring text data using the fitted vectorizer.
        score_bow = vectorizer.transform(score_df['press_release_text_bow'])

        # Train the LDA model on the training data.
        lda.fit(train_bow)

        # Generate out-of-sample document-topic distributions.
        score_topic_distributions = lda.transform(score_bow)

        # Create a DataFrame from the distributions.
        num_topics = config.vectorization_params.olda.num_topics
        olda_features_df = pd.DataFrame(
            score_topic_distributions,
            index=score_df.index,
            columns=[f'olda_topic_{i}' for i in range(num_topics)]
        )

        # Append the features for the current score year.
        all_olda_features.append(olda_features_df)

        # Capture the model and vectorizer from the final iteration.
        if score_year == max(score_years):
            final_lda_model = lda
            final_vectorizer = vectorizer

    # --- Final Assembly ---
    if all_olda_features:
        # Concatenate the feature DataFrames from all scoring years.
        final_olda_features = pd.concat(all_olda_features)
        # Join the computed oLDA features back to the main DataFrame.
        df_augmented = df_augmented.join(final_olda_features)

    # Clean up the temporary 'calendar_year' column.
    df_augmented = df_augmented.drop(columns=['calendar_year'])

    # Final validation to ensure the models were captured.
    if final_lda_model is None or final_vectorizer is None:
        raise RuntimeError("Failed to capture the final LDA model and vectorizer from the rolling loop.")

    # Return the augmented DataFrame and the final model artifacts.
    return df_augmented, final_lda_model, final_vectorizer
````

````python
# Task 6: Extract Contextual Embeddings via BERT-Family Encoders

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

class TransformerModelParams(BaseModel):
    """
    A Pydantic model for validating the 'transformers' section of the study
    configuration.

    This model defines the schema for parameters related to processing text
    with Transformer-based models like BERT, FinBERT, and MPNET. It ensures
    that model identifiers and batch processing parameters are valid.

    Attributes:
        model_names (Dict[str, str]): A dictionary mapping a short key
            (e.g., 'finbert') to its Hugging Face Hub identifier
            (e.g., 'ProsusAI/finbert'). The dictionary must not be empty.
        batch_size (int): The number of documents to process in a single
                          forward pass. Must be a positive integer.
        max_token_length (int): The maximum number of tokens to consider for
                                each document. Sequences will be truncated or
                                padded to this length. Must be a positive
                                integer.
    """
    # A dictionary mapping internal keys to Hugging Face model identifiers.
    # Field validation ensures the dictionary is not empty.
    model_names: Dict[str, str] = Field(..., min_items=1)

    # The size of each batch sent to the model for inference.
    # Field validation ensures the batch size is greater than zero.
    batch_size: int = Field(..., gt=0)

    # The maximum sequence length for the tokenizer.
    # Field validation ensures the length is greater than zero.
    max_token_length: int = Field(..., gt=0)


class VectorizationParams(BaseModel):
    """
    A Pydantic model for validating the 'vectorization_params' section of the
    study configuration.

    This model serves as a container for nested parameter structures related
    to different text vectorization methods.

    Attributes:
        transformers (TransformerModelParams): A nested model containing the
            validated parameters for the Transformer-based embedding extraction.
    """
    # A nested model for the Transformer-specific parameters.
    transformers: TransformerModelParams


class StudyConfig(BaseModel):
    """
    A Pydantic model representing a subsection of the main 'study_config'
    dictionary relevant to Task 6.

    This model ensures that the vectorization parameters, specifically those
    for the Transformer models, are present and correctly structured within the
    overall study configuration.

    Attributes:
        vectorization_params (VectorizationParams): The top-level container for
            all vectorization-related settings.
    """
    # The main container for all vectorization parameters.
    vectorization_params: VectorizationParams

# Define a type alias for a batch of tokenized data.
TensorBatch = Dict[str, torch.Tensor]

# ==============================================================================
# Task 6: Helper and Orchestrator Functions
# ==============================================================================

def _load_transformer_model(
    model_name: str
) -> Tuple[PreTrainedModel, PreTrainedTokenizer, torch.device]:
    """
    Loads a pre-trained Transformer model and tokenizer, and sets the device.

    This function handles the boilerplate of loading assets from the Hugging
    Face Hub, detecting and selecting the optimal hardware device (GPU/CPU),
    and preparing the model for inference.

    Args:
        model_name: The identifier of the model on the Hugging Face Hub
                    (e.g., 'bert-base-uncased').

    Returns:
        A tuple containing:
        - The loaded pre-trained model, moved to the selected device.
        - The corresponding loaded tokenizer.
        - The torch.device object representing the selected hardware.
    """
    # Automatically select the device: use CUDA (GPU) if available, otherwise CPU.
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    # Load the pre-trained tokenizer associated with the model.
    # `use_fast=True` is specified for performance.
    tokenizer = AutoTokenizer.from_pretrained(model_name, use_fast=True)

    # Load the pre-trained model architecture.
    model = AutoModel.from_pretrained(model_name)

    # Move the model's parameters and buffers to the selected device.
    model.to(device)

    # Set the model to evaluation mode. This is crucial as it disables
    # layers like dropout, ensuring deterministic output for inference.
    model.eval()

    return model, tokenizer, device


def _create_batches(
    text_series: pd.Series,
    tokenizer: PreTrainedTokenizer,
    batch_size: int,
    max_len: int,
    device: torch.device
) -> Generator[Tuple[pd.Index, TensorBatch], None, None]:
    """
    Creates a generator that yields batches of tokenized text data.

    This function efficiently processes a large series of text by tokenizing
    it in chunks (batches) and yielding them one by one. This "lazy" approach
    is highly memory-efficient, as it avoids loading the entire tokenized
    dataset into memory at once.

    Args:
        text_series: A pandas Series of cleaned text to be embedded.
        tokenizer: The pre-trained tokenizer to use.
        batch_size: The number of text documents in each batch.
        max_len: The maximum token length for truncation/padding.
        device: The torch device to which the tensors should be moved.

    Yields:
        A tuple containing:
        - The pandas Index for the documents in the current batch.
        - A dictionary of PyTorch tensors ('input_ids', 'attention_mask')
          for the batch, moved to the specified device.
    """
    # Iterate through the text series in steps of `batch_size`.
    for i in range(0, len(text_series), batch_size):
        # Select the slice of the series for the current batch.
        batch_slice = text_series.iloc[i:i + batch_size]

        # Extract the text content and convert to a list.
        # Handle potential non-string data gracefully.
        texts = [str(text) for text in batch_slice.tolist()]

        # Tokenize the batch of texts.
        # `truncation=True` cuts sequences longer than `max_len`.
        # `padding='max_length'` pads all sequences to `max_len`.
        # `return_tensors='pt'` returns PyTorch tensors.
        inputs = tokenizer(
            texts,
            max_length=max_len,
            padding='max_length',
            truncation=True,
            return_tensors='pt'
        )

        # Move the entire batch of tensors to the target device (GPU/CPU).
        # This minimizes data transfer overhead.
        inputs_on_device = {key: val.to(device) for key, val in inputs.items()}

        # Yield the index of the batch and the tokenized tensors.
        yield batch_slice.index, inputs_on_device


def extract_transformer_embeddings(
    events_df: pd.DataFrame,
    config: StudyConfig,
    artifacts_path: Path
) -> pd.DataFrame:
    """
    Orchestrates the extraction of contextual embeddings and persists token-level
    embeddings to disk for future use.

    This function iterates through a configured list of BERT-family models. For
    each model, it performs two critical operations:
    1. Computes document-level embeddings via mean pooling and augments the
       input DataFrame with them.
    2. Saves the full, computationally expensive token-level embeddings
       (the last hidden state tensor) for each document to disk, enabling
       efficient reuse in downstream interpretability tasks (Task 14).

    Args:
        events_df: The DataFrame containing the 'press_release_text_clean' column
                   and an 'event_id' in its index.
        config: The validated study configuration object.
        artifacts_path: The base directory path for saving the token-level
                        embedding tensors.

    Returns:
        The input DataFrame augmented with new columns for the document-level
        embeddings of each specified Transformer model.

    Raises:
        KeyError: If required columns or index levels are missing.
        TypeError: If inputs are not of the expected type.
    """
    # --- Input Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    if 'press_release_text_clean' not in events_df.columns:
        raise KeyError("DataFrame is missing required column 'press_release_text_clean'.")
    if 'event_id' not in events_df.index.names:
        raise ValueError("DataFrame must have 'event_id' in its index to save embeddings.")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # Retrieve transformer parameters from the configuration.
    transformer_params = config.vectorization_params.transformers
    model_identifiers = transformer_params.model_names
    batch_size = transformer_params.batch_size
    max_len = transformer_params.max_token_length

    # Create a working copy to which we will join the new features.
    df_augmented = events_df.copy()

    # --- Main Loop: Iterate Through Each Model ---
    for model_key, model_name in model_identifiers.items():
        logging.info(f"Processing embeddings for model: {model_key}")

        # --- Step 1: Load Model and Prepare Artifact Directory ---
        model, tokenizer, device = _load_transformer_model(model_name)

        # Create the dedicated subdirectory for this model's token embeddings.
        output_dir = artifacts_path / "token_embeddings" / model_key
        output_dir.mkdir(parents=True, exist_ok=True)

        # Initialize a list to store the document-level (mean-pooled) embeddings.
        all_doc_embeddings: List[np.ndarray] = []

        # --- Step 2: Create Batch Generator ---
        batch_generator = _create_batches(
            df_augmented['press_release_text_clean'],
            tokenizer,
            batch_size,
            max_len,
            device
        )

        # --- Step 3: Process Batches, Compute and Save Embeddings ---
        with torch.no_grad():
            for batch_index, batch_inputs in batch_generator:
                # Perform the forward pass of the model.
                outputs = model(**batch_inputs)

                # Extract the last hidden state tensor. Shape: (batch_size, max_len, hidden_dim).
                last_hidden_state = outputs.last_hidden_state

                # --- Document-Level Embedding Calculation ---
                # Compute the mean-pooled embedding by averaging across the token dimension.
                pooled_embeddings = torch.mean(last_hidden_state, dim=1)
                all_doc_embeddings.append(pooled_embeddings.cpu().numpy())

                # --- Token-Level Embedding Persistence ---
                # Iterate through each item in the batch to save its token tensor.
                for i, doc_multi_index in enumerate(batch_index):
                    # Extract the unique event_id from the multi-index tuple.
                    event_id = doc_multi_index[2]

                    # Define the specific output path for this document's tensor.
                    tensor_path = output_dir / f"{event_id}.pt"

                    # Extract the single (max_len, hidden_dim) tensor for this document.
                    token_tensor = last_hidden_state[i]

                    # Save the tensor to disk. Move to CPU first to ensure portability.
                    torch.save(token_tensor.cpu(), tensor_path)

        # --- Assemble and Merge Document-Level Embeddings ---
        # Concatenate the document embeddings from all batches into a single NumPy array.
        final_doc_embeddings = np.vstack(all_doc_embeddings)

        # Create a new DataFrame for the document-level embeddings.
        embedding_dim = final_doc_embeddings.shape[1]
        embedding_df = pd.DataFrame(
            final_doc_embeddings,
            index=df_augmented.index,
            columns=[f'{model_key}_{i}' for i in range(embedding_dim)]
        )

        # Join the new document-level embedding features with the main DataFrame.
        df_augmented = df_augmented.join(embedding_df)

        # --- Memory Management ---
        del model, tokenizer
        if torch.cuda.is_available():
            torch.cuda.empty_cache()
        gc.collect()

    return df_augmented
````

````python
# Task 7: Estimate Rolling Lasso Soft Scores per Vectorization Method

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

class LassoParams(BaseModel):
    """
    A Pydantic model for validating the 'lasso' section of the study
    configuration.

    This model defines the schema for the hyperparameters of the Lasso (L1
    regularized) linear regression model, which is used for feature selection
    and prediction.

    Attributes:
        lambda_val (float): The regularization strength, corresponding to the
                            lambda (λ) in the Lasso objective function. Must be
                            non-negative. Pydantic's `alias` is used to allow
                            the Python-reserved keyword 'lambda' in the config.
        fit_intercept (bool): Whether to calculate the intercept for this model.
        max_iter (int): The maximum number of iterations for the coordinate
                        descent solver. Must be a positive integer.
        tol (float): The tolerance for optimization convergence. Must be a
                     positive float.
        random_seed (int): The seed for the random number generator to ensure
                           reproducibility of results.
    """
    # The regularization parameter lambda (λ). Must be non-negative.
    # An alias is used because 'lambda' is a reserved keyword in Python.
    lambda_val: float = Field(..., alias='lambda', ge=0)

    # A boolean flag to determine if an intercept should be fitted.
    fit_intercept: bool

    # The maximum number of iterations for the solver to converge.
    max_iter: int = Field(..., gt=0)

    # The tolerance level for the optimization algorithm's stopping criterion.
    tol: float = Field(..., gt=0)

    # The seed for the random number generator for reproducibility.
    random_seed: int


class RollingEstimationParams(BaseModel):
    """
    A Pydantic model for validating the 'rolling_estimation' parameters.

    This model ensures the logical integrity of the time-series rolling window
    setup, which is critical for avoiding look-ahead bias in the generation
    of out-of-sample predictions.

    Attributes:
        initial_train_year (int): The first year of data to be included in the
                                  initial training set.
        score_years (List[int]): A list of the years for which out-of-sample
                                 predictions (scores) will be generated. Must
                                 not be empty.
    """
    # The first year of data for the initial training window.
    initial_train_year: int

    # The list of years to be scored out-of-sample.
    # Field validation ensures the list contains at least one year.
    score_years: List[int] = Field(..., min_items=1)


class ModelingParams(BaseModel):
    """
    A Pydantic model for validating the 'modeling_params' section of the
    study configuration.

    This model serves as a container for nested parameter structures related
    to the predictive modeling stage of the research pipeline.

    Attributes:
        lasso (LassoParams): A nested model containing the validated
                             hyperparameters for the Lasso regression.
        rolling_estimation (RollingEstimationParams): A nested model containing
            the validated parameters for the rolling estimation procedure.
    """
    # A nested model for the Lasso regression hyperparameters.
    lasso: LassoParams

    # A nested model for the rolling window protocol.
    rolling_estimation: RollingEstimationParams


class StudyConfig(BaseModel):
    """
    A Pydantic model representing a subsection of the main 'study_config'
    dictionary relevant to Task 7.

    This model ensures that the modeling parameters, including settings for
    the Lasso model and the rolling estimation protocol, are present and
    correctly structured.

    Attributes:
        modeling_params (ModelingParams): The top-level container for all
            modeling-related settings.
    """
    # The main container for all modeling parameters.
    modeling_params: ModelingParams

# Define a type alias for the nested dictionary of Lasso weights.
YearlyLassoWeights = Dict[str, Dict[int, np.ndarray]]

# ==============================================================================
# Task 7: Helper and Orchestrator Functions
# ==============================================================================

def _get_feature_columns_for_method(
    df: pd.DataFrame,
    method: str
) -> List[str]:
    """
    Identifies the list of feature columns for a given vectorization method.

    This helper function dynamically constructs the list of column names
    based on a prefix convention (e.g., 'bkmx_', 'olda_', 'finbert_').

    Args:
        df: The DataFrame containing all feature columns.
        method: The short name of the vectorization method (e.g., 'bkmx').

    Returns:
        A list of column names corresponding to the specified method.
    """
    # Use a simple prefix-based selection to find the relevant columns.
    return [col for col in df.columns if col.startswith(f'{method}_')]


def estimate_rolling_soft_scores(
    events_df: pd.DataFrame,
    config: StudyConfig,
    artifacts_path: Path
) -> Tuple[pd.DataFrame, YearlyLassoWeights]:
    """
    Orchestrates the estimation of rolling Lasso soft scores and persists the
    fitted models to disk.

    This function implements the core predictive modeling pipeline. For each
    feature set (BKMX, oLDA, etc.), it runs a year-by-year rolling estimation
    to generate out-of-sample "soft scores". Crucially, it now also saves each
    fitted Lasso model to disk and returns the coefficients, making them
    available for downstream interpretability tasks.

    Args:
        events_df: The DataFrame containing the target variable ('ret_announce_cc')
                   and all previously computed feature columns.
        config: The validated study configuration object.
        artifacts_path: The base directory path for saving the fitted Lasso models.

    Returns:
        A tuple containing:
        - The input DataFrame augmented with new columns for the soft scores.
        - A nested dictionary containing the fitted Lasso coefficient vectors
          for each vectorization method and training year.
    """
    # --- Input Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # Create a working copy.
    df_augmented = events_df.copy()
    df_augmented['calendar_year'] = df_augmented['pr_timestamp_et'].dt.year

    # Retrieve parameters from the configuration.
    lasso_params = config.modeling_params.lasso
    protocol_params = config.modeling_params.rolling_estimation
    score_years = protocol_params.score_years

    # Define the vectorization methods to process.
    vectorization_methods = ['bkmx', 'olda', 'bert', 'finbert', 'mpnet']

    # Initialize a dictionary to store the raw coefficient vectors.
    yearly_lasso_weights: YearlyLassoWeights = {method: {} for method in vectorization_methods}

    # --- Outer Loop: Iterate Through Vectorization Methods ---
    for method in vectorization_methods:
        # Create the dedicated subdirectory for this method's Lasso models.
        output_dir = artifacts_path / "lasso_models" / method
        output_dir.mkdir(parents=True, exist_ok=True)

        feature_cols = _get_feature_columns_for_method(df_augmented, method)
        if not feature_cols:
            continue

        yearly_scores: List[pd.Series] = []

        # --- Inner Loop: Rolling Window Estimation ---
        for score_year in score_years:
            train_year = score_year - 1

            train_df = df_augmented[df_augmented['calendar_year'] == train_year]
            score_df = df_augmented[df_augmented['calendar_year'] == score_year]

            if train_df.empty or score_df.empty:
                continue

            X_train = train_df[feature_cols].values
            y_train = train_df['ret_announce_cc'].values
            X_score = score_df[feature_cols].values

            # --- Fit Scaler and Lasso Model ---
            scaler = StandardScaler()
            X_train_std = scaler.fit_transform(X_train)

            lasso = Lasso(
                alpha=lasso_params.lambda_val,
                fit_intercept=lasso_params.fit_intercept,
                max_iter=lasso_params.max_iter,
                tol=lasso_params.tol,
                random_state=lasso_params.random_seed
            )
            lasso.fit(X_train_std, y_train)

            # --- Persist Artifacts ---
            # Save the entire fitted Lasso model object to disk for Task 14.
            model_path = output_dir / f"{train_year}.pkl"
            joblib.dump(lasso, model_path)

            # Store the coefficient vector in the results dictionary for Task 16.
            yearly_lasso_weights[method][train_year] = lasso.coef_

            # --- Predict Out-of-Sample Scores ---
            X_score_std = scaler.transform(X_score)
            predicted_scores = lasso.predict(X_score_std)

            score_series = pd.Series(predicted_scores, index=score_df.index)
            yearly_scores.append(score_series)

        # --- Consolidate and Store Results for the Method ---
        if yearly_scores:
            method_scores = pd.concat(yearly_scores)
            df_augmented[f'soft_{method}'] = method_scores

    # Clean up the temporary column.
    df_augmented = df_augmented.drop(columns=['calendar_year'])

    return df_augmented, yearly_lasso_weights
````

````python
# Task 8: Aggregate Soft Signals and Winsorize Surprise

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

class SoftSignalAggregationParams(BaseModel):
    """
    A Pydantic model for validating the 'soft_signal_aggregation' section of
    the study configuration.

    This model defines the schema for parameters controlling how individual
    soft scores from different models are combined into a single composite signal.

    Attributes:
        models_included (List[str]): A list of short names for the models whose
                                     soft scores should be averaged (e.g.,
                                     ['bkmx', 'olda', 'finbert']). The list
                                     must not be empty.
        output_field (str): The desired column name for the resulting
                            aggregated signal in the DataFrame. The name must
                            be a non-empty string.
    """
    # A list of model keys to be included in the composite signal.
    # Field validation ensures the list contains at least one model.
    models_included: List[str] = Field(..., min_items=1)

    # The name of the new column that will store the aggregated soft score.
    # Field validation ensures the name is a non-empty string.
    output_field: str = Field(..., min_length=1)


class RegressionParams(BaseModel):
    """
    A Pydantic model for validating the 'regression_params' section of the
    study configuration.

    This model defines the schema for parameters related to the econometric
    regression analyses, such as winsorization thresholds.

    Attributes:
        winsorize_surprise_quantiles (Tuple[float, float]): A 2-element tuple
            specifying the lower and upper quantiles for winsorizing the
            earnings surprise variable. Values must be between 0.0 and 1.0,
            and the lower quantile must be strictly less than the upper.
    """
    # A tuple defining the lower and upper quantiles for winsorization.
    winsorize_surprise_quantiles: Tuple[float, float]

    @validator('winsorize_surprise_quantiles')
    def validate_quantiles_logic(cls, v: Tuple[float, float]) -> Tuple[float, float]:
        """
        Validates the format, bounds, and order of the winsorization quantiles.

        Args:
            v: The input value for `winsorize_surprise_quantiles`.

        Returns:
            The validated tuple of quantiles.

        Raises:
            ValueError: If the input is not a 2-element tuple, if values are
                        outside the [0, 1] range, or if lower >= upper.
        """
        # Check if the input is a tuple of length 2.
        if not (isinstance(v, tuple) and len(v) == 2):
            raise ValueError("'winsorize_surprise_quantiles' must be a 2-element tuple.")

        # Unpack the lower and upper quantiles for individual checks.
        lower_q, upper_q = v

        # Check if both elements are valid probabilities (between 0 and 1).
        if not (0.0 <= lower_q <= 1.0 and 0.0 <= upper_q <= 1.0):
            raise ValueError("Quantiles must be between 0.0 and 1.0.")

        # Check if the lower quantile is strictly less than the upper quantile.
        if not lower_q < upper_q:
            raise ValueError("The lower quantile must be strictly less than the upper quantile.")

        # Return the validated value.
        return v


class StudyConfig(BaseModel):
    """
    A Pydantic model representing a subsection of the main 'study_config'
    dictionary relevant to Task 8.

    This model ensures that the parameters for signal aggregation and
    regression preparations (like winsorization) are present and correctly
    structured.

    Attributes:
        soft_signal_aggregation (SoftSignalAggregationParams): The validated
            parameters for creating the composite soft signal.
        regression_params (RegressionParams): The validated parameters for
            econometric analysis preparations.
    """
    # A nested model for the soft signal aggregation parameters.
    soft_signal_aggregation: SoftSignalAggregationParams

    # A nested model for the regression-related parameters.
    regression_params: RegressionParams


# ==============================================================================
# Task 8: Helper and Orchestrator Functions
# ==============================================================================

def _aggregate_soft_signals(
    df: pd.DataFrame,
    config: StudyConfig
) -> pd.DataFrame:
    """
    Computes a composite soft signal by averaging individual model scores.

    This function creates a unified measure of textual sentiment by taking the
    simple arithmetic mean of the soft scores generated by each of the five
    vectorization methods. This composite signal, 'soft_mean', is used in
    subsequent analyses for a holistic view of the text's impact.

    Equation: Soft^Mean_{c,t} = (1/N) * sum(Soft^I_{c,t} for I in Models)

    Args:
        df: The DataFrame containing the individual soft score columns
            (e.g., 'soft_bkmx', 'soft_finbert').
        config: The validated study configuration object.

    Returns:
        The DataFrame with a new column ('soft_mean') containing the
        aggregated signal.
    """
    # Retrieve the list of models to include in the aggregation from the config.
    models_to_include = config.soft_signal_aggregation.models_included

    # Construct the list of column names for the soft scores.
    soft_score_columns = [f'soft_{model}' for model in models_to_include]

    # Verify that all required soft score columns are present in the DataFrame.
    missing_cols = [col for col in soft_score_columns if col not in df.columns]
    if missing_cols:
        raise KeyError(f"DataFrame is missing required soft score columns for aggregation: {missing_cols}")

    # Retrieve the desired name for the output column from the config.
    output_col_name = config.soft_signal_aggregation.output_field

    # Compute the row-wise mean of the soft score columns.
    # `axis=1` specifies the operation is across columns for each row.
    # `skipna=True` (default) ensures robustness to missing scores for a given event.
    df[output_col_name] = df[soft_score_columns].mean(axis=1)

    return df


def _winsorize_earnings_surprise(
    df: pd.DataFrame,
    config: StudyConfig
) -> pd.DataFrame:
    """
    Winsorizes the earnings surprise variable to mitigate outlier influence.

    This function caps the 'surprise' variable at specified lower and upper
    percentiles. This is a standard practice in empirical finance to prevent
    extreme observations from disproportionately affecting regression results.

    Equation: Surprise^wins_{c,t} = max(P_lower, min(Surprise_{c,t}, P_upper))

    Args:
        df: The DataFrame containing the raw 'surprise' column.
        config: The validated study configuration object.

    Returns:
        The DataFrame with a new column ('surprise_winsorized') containing
        the capped values.
    """
    # Verify that the raw 'surprise' column exists.
    if 'surprise' not in df.columns:
        raise KeyError("DataFrame is missing required column 'surprise'.")

    # Retrieve the lower and upper quantile thresholds from the configuration.
    lower_quantile, upper_quantile = config.regression_params.winsorize_surprise_quantiles

    # Calculate the actual values at these quantiles from the data distribution.
    # This is done on the entire series to ensure stable and consistent bounds.
    lower_bound = df['surprise'].quantile(lower_quantile)
    upper_bound = df['surprise'].quantile(upper_quantile)

    # Apply the winsorization by clipping the 'surprise' series at the bounds.
    # The .clip() method efficiently caps the values.
    df['surprise_winsorized'] = df['surprise'].clip(
        lower=lower_bound,
        upper=upper_bound
    )

    return df

def _validate_signal_distributions(
    df: pd.DataFrame,
    config: StudyConfig
) -> None:
    """
    Computes and logs diagnostic statistics for the final analytical signals.

    This function serves as a critical sanity check after signal generation.
    It calculates and logs summary statistics (mean, std, etc.) and a
    correlation matrix for the key variables to be used in the regression
    analysis. This helps to quickly identify potential issues like lack of
    variance, extreme outliers, or unexpected correlations.

    Args:
        df: The DataFrame containing the newly generated signal columns.
        config: The validated study configuration object.
    """
    # --- Input Validation ---
    if not isinstance(df, pd.DataFrame):
        # Log a warning instead of raising an error for a diagnostic function.
        logging.warning("Validation skipped: Input is not a pandas DataFrame.")
        return

    # --- Define Key Variables for Diagnostics ---
    # Construct the list of all relevant signal columns.
    soft_score_cols = [f'soft_{model}' for model in config.soft_signal_aggregation.models_included]
    key_cols = ['ret_announce_cc', 'surprise_winsorized', config.soft_signal_aggregation.output_field] + soft_score_cols

    # Filter out any columns that may not be present in the DataFrame yet.
    cols_to_validate = [col for col in key_cols if col in df.columns]

    # Exit if no valid columns are found to analyze.
    if not cols_to_validate:
        logging.warning("Signal validation skipped: No valid signal columns found in DataFrame.")
        return

    # --- Compute and Log Summary Statistics ---
    # Use .describe() to get a summary of the distributions.
    summary_stats = df[cols_to_validate].describe()

    # Log the summary statistics table, formatted as a string for readability.
    logging.info(f"--- Signal Distribution Summary ---\n{summary_stats.to_string()}")

    # --- Compute and Log Correlation Matrix ---
    # Use .corr() to get the pairwise correlation between signals.
    correlation_matrix = df[cols_to_validate].corr()

    # Log the correlation matrix, formatted as a string.
    logging.info(f"\n--- Signal Correlation Matrix ---\n{correlation_matrix.to_string()}")


def prepare_final_signals(
    events_df: pd.DataFrame,
    config: StudyConfig
) -> pd.DataFrame:
    """
    Orchestrates the final preparation of analytical signals, including
    aggregation, winsorization, and diagnostic validation.

    This function serves as the entry point for Task 8. It sequentially calls
    helper functions to:
    1. Aggregate the five individual soft scores into a single composite signal.
    2. Winsorize the raw earnings surprise variable to handle outliers.
    3. (New) Run a diagnostic validation to log summary statistics and
       correlations of the newly created signals.

    Args:
        events_df: The DataFrame containing all previously computed features.
        config: The validated study configuration object.

    Returns:
        The DataFrame augmented with the 'soft_mean' and 'surprise_winsorized'
        columns, ready for regression analysis.
    """
    # --- Input Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # Create a working copy to avoid modifying the original DataFrame in place.
    df_processed = events_df.copy()

    # --- Step 1: Aggregate Soft Signals ---
    # This function adds the 'soft_mean' column (or as named in config).
    df_processed = _aggregate_soft_signals(df_processed, config)

    # --- Step 2: Winsorize Earnings Surprise ---
    # This function adds the 'surprise_winsorized' column.
    df_processed = _winsorize_earnings_surprise(df_processed, config)

    # --- Step 3: Diagnostic Validation ---
    # This new step provides a crucial sanity check on the generated signals.
    # It computes and logs summary statistics and correlations.
    _validate_signal_distributions(df_processed, config)

    # Return the final DataFrame, now fully prepared for modeling.
    return df_processed
````

````python
# Task 9: Estimate Baseline Cross-Sectional Regressions

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

class RegressionParams(BaseModel):
    """
    A Pydantic model for validating the 'regression_params' section of the
    study configuration.

    This model defines the schema for parameters related to the main econometric
    regression analyses, such as the sample period for the analysis.

    Attributes:
        sample_period (Tuple[str, str]): A 2-element tuple specifying the
            inclusive start and end dates for the regression sample period.
            Dates must be in ISO 8601 format (e.g., '2006-01-01').
    """
    # A tuple defining the start and end dates of the main analysis period.
    sample_period: Tuple[str, str]

    @validator('sample_period')
    def validate_sample_period_logic(cls, v: Tuple[str, str]) -> Tuple[str, str]:
        """
        Validates the format, length, and chronological order of the sample period.

        Args:
            v: The input value for `sample_period`.

        Returns:
            The validated tuple of date strings.

        Raises:
            ValueError: If the input is not a 2-element tuple, if dates are
                        malformed, or if the start date is after the end date.
        """
        # Check if the input is a tuple of length 2.
        if not (isinstance(v, tuple) and len(v) == 2):
            raise ValueError("'sample_period' must be a 2-element tuple of date strings.")

        # Unpack the start and end date strings.
        start_str, end_str = v

        try:
            # Attempt to parse the ISO 8601 date strings into datetime objects.
            start_dt = datetime.datetime.fromisoformat(start_str)
            end_dt = datetime.datetime.fromisoformat(end_str)

            # Check if the start date is chronologically after the end date.
            if start_dt > end_dt:
                # Raise an error if the date range is invalid.
                raise ValueError("The start date of the sample period cannot be after the end date.")

        except ValueError as e:
            # Catch parsing errors or the custom ValueError and re-raise with context.
            raise ValueError(f"Invalid date format or range in 'sample_period': {e}")

        # Return the validated value.
        return v


class StudyConfig(BaseModel):
    """
    A Pydantic model representing a subsection of the main 'study_config'
    dictionary relevant to Task 9.

    This model ensures that the regression parameters, specifically the sample
    period, are present and correctly structured within the overall study
    configuration.

    Attributes:
        regression_params (RegressionParams): The validated parameters for
            econometric analysis preparations.
    """
    # The main container for all regression-related parameters.
    regression_params: RegressionParams

# Define a type alias for the structured regression results.
RegressionResultDict = Dict[str, Any]

# ==============================================================================
# Task 9: Helper and Orchestrator Functions
# ==============================================================================

def _prepare_regression_sample(
    df: pd.DataFrame,
    dependent_var: str,
    independent_vars: List[str],
    config: StudyConfig
) -> pd.DataFrame:
    """
    Prepares the DataFrame for a specific regression analysis.

    This function filters the main DataFrame based on the analysis sample period
    and ensures there are no missing values for the variables involved in the
    regression model.

    Args:
        df: The full DataFrame containing all variables.
        dependent_var: The name of the dependent variable column.
        independent_vars: A list of names for the independent variable columns.
        config: The validated study configuration object.

    Returns:
        A new DataFrame representing the clean, complete-case sample for the
        regression.
    """
    # --- Date Range Filter ---
    # Retrieve the sample period start and end dates from the configuration.
    start_date, end_date = config.regression_params.sample_period

    # Create a boolean mask to select data within the specified date range.
    # The index level 'ann_trade_date' is used for filtering.
    date_mask = df.index.get_level_values('ann_trade_date').to_series().between(
        pd.Timestamp(start_date, tz='America/New_York'),
        pd.Timestamp(end_date, tz='America/New_York'),
        inclusive='both'
    )

    # Apply the date filter.
    sample_df = df.loc[date_mask].copy()

    # --- Completeness Filter ---
    # Define the full set of variables required for this specific regression.
    required_vars = [dependent_var] + independent_vars

    # Remove any rows that have missing values in any of the required variables.
    sample_df.dropna(subset=required_vars, inplace=True)

    return sample_df

RegressionResultDict = Dict[str, Any]
FittedModelsDict = Dict[str, PanelResults]

# ==============================================================================
# Implementation of _run_panel_regression
# ==============================================================================

def _run_panel_regression(
    sample_df: pd.DataFrame,
    dependent_var: str,
    independent_vars: List[str]
) -> PanelResults:
    """
    Estimates a PanelOLS regression with two-way clustered standard errors.

    This function provides a self-contained, robust implementation for running
    panel regressions as required by the study. It takes a DataFrame with the
    standard project index, internally reshapes it to the (entity, time) format
    required by the `linearmodels` library, and estimates an OLS model with
    standard errors clustered by both firm ('permno') and time ('ann_trade_date').

    Args:
        sample_df: The clean, complete-case DataFrame for the regression,
                   expected to have a MultiIndex including 'permno' and
                   'ann_trade_date'.
        dependent_var: The name of the dependent variable column.
        independent_vars: A list of names for the independent variable columns.

    Returns:
        A `PanelResults` object from the `linearmodels` library containing
        the detailed results of the fitted regression model.

    Raises:
        ValueError: If the required index levels ('permno', 'ann_trade_date')
                    are not found in the input DataFrame's index.
    """
    # --- Input Validation ---
    # Verify that the necessary index levels for clustering are present.
    required_index_levels = ['permno', 'ann_trade_date']
    if not all(level in sample_df.index.names for level in required_index_levels):
        raise ValueError(f"Input DataFrame must have {required_index_levels} in its index.")

    # --- Step 1: Prepare Data for linearmodels ---
    # The `linearmodels` library expects a MultiIndex of (entity, time).
    # We transform our standard project index to this format.
    # `reset_index()` moves all index levels to columns.
    # `set_index(['permno', 'ann_trade_date'])` creates the required index.
    panel_data = sample_df.reset_index().set_index(['permno', 'ann_trade_date'])

    # Select the dependent variable (endog) from the re-indexed data.
    y = panel_data[dependent_var]

    # Select the independent variables (exog) and add a constant for the intercept.
    X = sm.add_constant(panel_data[independent_vars])

    # --- Step 2: Instantiate and Fit the PanelOLS Model ---
    # Instantiate the PanelOLS model with the prepared data.
    model = PanelOLS(y, X)

    # Fit the model, specifying the covariance estimation method.
    # `cov_type='clustered'` indicates we want robust standard errors.
    # `cluster_entity=True` automatically clusters by the first index level ('permno').
    # `cluster_time=True` automatically clusters by the second index level ('ann_trade_date').
    results = model.fit(
        cov_type='clustered',
        cluster_entity=True,
        cluster_time=True
    )

    # Return the comprehensive results object.
    return results

def _format_regression_results(results: PanelResults) -> RegressionResultDict:
    """
    Formats the results from a linearmodels regression into a clean dictionary.

    Args:
        results: A fitted `PanelResults` object.

    Returns:
        A dictionary containing key statistics: coefficients, standard errors,
        t-stats, p-values, R-squared, and number of observations.
    """
    # Create a dictionary to store the formatted results.
    formatted_results = {
        "coefficients": results.params.to_dict(),
        "std_errors": results.std_errors.to_dict(),
        "t_stats": results.tstats.to_dict(),
        "p_values": results.pvalues.to_dict(),
        "r_squared_overall": results.rsquared_overall,
        "n_obs": int(results.nobs)
    }
    return formatted_results


def estimate_baseline_regressions(
    events_df: pd.DataFrame,
    config: StudyConfig
) -> Tuple[Dict[str, RegressionResultDict], FittedModelsDict]:
    """
    Orchestrates the estimation of baseline cross-sectional regressions and
    returns both summary results and the fitted model objects.

    This function iterates through each of the five soft signal types (BKMX,
    oLDA, etc.). For each, it runs a regression of announcement returns on the
    earnings surprise and the soft signal, using two-way clustered standard
    errors.

    This refactored version now returns not only the formatted summary of
    results but also the actual fitted `PanelResults` objects, which are
    essential for ensuring consistency in downstream tasks like SHAP analysis.

    Equation: Ret_{c,t} = a + B_0 * Surprise_{c,t}^wins + B_1 * Soft_{c,t}^I + e

    Args:
        events_df: The fully prepared DataFrame containing all features and signals.
        config: The validated study configuration object.

    Returns:
        A tuple containing:
        - A dictionary of formatted regression summary statistics for each model.
        - A dictionary of the actual fitted `PanelResults` objects for each model.
    """
    # --- Input Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # Define the list of soft signal models to iterate through.
    vectorization_methods = ['bkmx', 'olda', 'bert', 'finbert', 'mpnet']

    # Initialize a dictionary to store the formatted summary of all regressions.
    all_summary_results: Dict[str, RegressionResultDict] = {}

    # Initialize a dictionary to store the fitted model objects.
    all_fitted_models: FittedModelsDict = {}

    # --- Main Loop: Iterate Through Each Soft Signal Model ---
    for method in vectorization_methods:
        # Define the column name for the current soft signal.
        soft_score_col = f'soft_{method}'

        # Check if the required soft score column exists before proceeding.
        if soft_score_col not in events_df.columns:
            # Log a warning or simply continue if a feature set is not present.
            continue

        # Define the variables for this specific regression model.
        dependent_var = 'ret_announce_cc'
        independent_vars = ['surprise_winsorized', soft_score_col]

        # --- Step 1: Prepare the Regression Sample ---
        # Filter data for the correct date range and handle missing values.
        sample_df = _prepare_regression_sample(
            events_df, dependent_var, independent_vars, config
        )

        # Skip if the resulting sample is too small to run a regression.
        if len(sample_df) < len(independent_vars) + 1:
            continue

        # --- Step 2: Estimate the PanelOLS Model ---
        # Run the regression with two-way clustered standard errors.
        # The helper function `_run_panel_regression` handles index manipulation.
        regression_results = _run_panel_regression(
            sample_df,
            dependent_var,
            independent_vars
        )

        # --- Step 3: Format and Store the Results ---
        # Parse the results object into a clean summary dictionary.
        formatted_results = _format_regression_results(regression_results)

        # Store the formatted summary for the current method.
        all_summary_results[method] = formatted_results

        # Store the complete fitted model object for the current method.
        all_fitted_models[method] = regression_results

    # Return both the summary statistics and the fitted model objects.
    return all_summary_results, all_fitted_models
````

````python
# Task 10: Estimate Combined Cross-Sectional Regression

# ==============================================================================
# Task 10: Helper and Orchestrator Functions
# ==============================================================================

def _compute_vif(
    df: pd.DataFrame,
    independent_vars: List[str]
) -> Dict[str, float]:
    """
    Computes the Variance Inflation Factor (VIF) for each independent variable.

    VIF is a diagnostic tool used to detect the severity of multicollinearity
    in a regression analysis. A high VIF (often cited as > 5 or 10) suggests
    that a variable is highly correlated with other predictors in the model.

    Args:
        df: The DataFrame used for the regression (complete cases).
        independent_vars: The list of independent variable names.

    Returns:
        A dictionary mapping each variable name to its VIF score.
    """
    # Select the independent variables and add a constant for the intercept.
    X = sm.add_constant(df[independent_vars], has_constant='add')

    # Calculate VIF for each predictor. The list comprehension iterates through
    # each variable, computing its VIF against all other variables.
    # The index `i+1` is used because `sm.add_constant` prepends the constant column.
    vif_data = {
        col: variance_inflation_factor(X.values, i + 1)
        for i, col in enumerate(independent_vars)
    }

    return vif_data


def estimate_combined_regressions(
    events_df: pd.DataFrame,
    config: StudyConfig
) -> Dict[str, RegressionResultDict]:
    """
    Orchestrates the estimation of combined cross-sectional regressions.

    This function estimates two key "horse race" regression models to assess
    the combined and marginal explanatory power of the different soft signals:
    1. Joint Model: Includes the earnings surprise and all five individual
       soft signals simultaneously.
    2. Composite Model: Includes the earnings surprise and the single aggregated
       'soft_mean' signal.

    Both models are estimated using PanelOLS with two-way clustered standard
    errors (by firm and time) for robust inference.

    Args:
        events_df: The fully prepared DataFrame with all features and signals.
        config: The validated study configuration object.

    Returns:
        A dictionary containing the detailed results for both the 'joint_model'
        and the 'composite_model'. Results include coefficients, standard
        errors, p-values, R-squared, and VIF scores for the joint model.
    """
    # --- Input Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # Initialize a dictionary to store the results of both regressions.
    all_results: Dict[str, RegressionResultDict] = {}

    # Define the dependent variable for all models.
    dependent_var = 'ret_announce_cc'

    # ==========================================================================
    # Specification 1: Joint Model with All Five Soft Signals
    # ==========================================================================

    # Define the independent variables for the joint regression.
    joint_independent_vars = [
        'surprise_winsorized',
        'soft_bkmx',
        'soft_olda',
        'soft_bert',
        'soft_mpnet',
        'soft_finbert'
    ]

    # Verify that all required columns exist in the DataFrame.
    if all(col in events_df.columns for col in joint_independent_vars):
        # --- Step 1.1: Prepare the data sample for the joint model ---
        joint_sample_df = _prepare_regression_sample(
            events_df, dependent_var, joint_independent_vars, config
        )

        if not joint_sample_df.empty:
            # --- Step 1.2: Estimate the PanelOLS model ---
            # The DataFrame must be re-indexed for linearmodels: (entity, time).
            joint_regression_results = _run_panel_regression(
                joint_sample_df.swaplevel('permno', 'ann_trade_date'),
                dependent_var,
                joint_independent_vars
            )

            # --- Step 1.3: Format results and add VIF diagnostics ---
            joint_formatted_results = _format_regression_results(joint_regression_results)

            # Compute and add VIF scores as a diagnostic for multicollinearity.
            joint_formatted_results['vif_scores'] = _compute_vif(
                joint_sample_df, joint_independent_vars
            )

            # Store the complete results for the joint model.
            all_results['joint_model'] = joint_formatted_results

    # ==========================================================================
    # Specification 2: Composite Model with 'soft_mean'
    # ==========================================================================

    # Define the independent variables for the composite signal regression.
    composite_independent_vars = ['surprise_winsorized', 'soft_mean']

    # Verify that all required columns exist in the DataFrame.
    if all(col in events_df.columns for col in composite_independent_vars):
        # --- Step 2.1: Prepare the data sample for the composite model ---
        composite_sample_df = _prepare_regression_sample(
            events_df, dependent_var, composite_independent_vars, config
        )

        if not composite_sample_df.empty:
            # --- Step 2.2: Estimate the PanelOLS model ---
            composite_regression_results = _run_panel_regression(
                composite_sample_df.swaplevel('permno', 'ann_trade_date'),
                dependent_var,
                composite_independent_vars
            )

            # --- Step 2.3: Format and store the results ---
            composite_formatted_results = _format_regression_results(composite_regression_results)

            # Store the complete results for the composite model.
            all_results['composite_model'] = composite_formatted_results

    return all_results
````

````python
# Task 11: Compute SHAP Feature Importance for Baseline Regressions

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

# Define a type alias for the SHAP importance results.
ShapImportanceDict = Dict[str, float]

# ==============================================================================
# Task 11: Helper and Orchestrator Functions
# ==============================================================================

def _compute_and_normalize_shap_values(
    model_params: pd.Series,
    X_data: pd.DataFrame
) -> ShapImportanceDict:
    """
    Computes and normalizes SHAP values for a fitted linear model's parameters.

    This function uses the `shap.LinearExplainer` to calculate exact Shapley
    values, which provide a theoretically sound measure of feature contribution
    to a model's prediction. It then aggregates these local, observation-level
    values into a global measure of feature importance by taking the mean of
    the absolute SHAP values and normalizing them to sum to 100%.

    Args:
        model_params: A pandas Series containing the fitted model's coefficients.
                      It is expected to have an entry for each feature in `X_data`
                      and an optional entry for the intercept named 'const'.
        X_data: The DataFrame of independent variables (features) used for the
                explanation. This dataset also serves as the background
                dataset for the SHAP explainer.

    Returns:
        A dictionary mapping each feature name to its normalized global
        importance percentage.
    """
    # --- Input Validation ---
    if not isinstance(model_params, pd.Series):
        raise TypeError("Input 'model_params' must be a pandas Series.")
    if not isinstance(X_data, pd.DataFrame):
        raise TypeError("Input 'X_data' must be a pandas DataFrame.")
    if not set(X_data.columns).issubset(set(model_params.index)):
        raise ValueError("All columns in 'X_data' must have a corresponding coefficient in 'model_params'.")

    # --- Step 1: Instantiate the SHAP Explainer ---

    # Separate the intercept from the feature coefficients. The `shap.LinearExplainer`
    # requires these as distinct inputs.
    # The `.get('const', 0.0)` call safely handles models fit without an intercept.
    coefficients = model_params.drop('const', errors='ignore')
    intercept = model_params.get('const', 0.0)

    # Ensure the coefficients align with the data columns.
    coefficients = coefficients[X_data.columns]

    # Initialize the LinearExplainer. For linear models, this provides exact
    # and efficient computation of Shapley values.
    # The model is defined by its coefficients and intercept.
    # The `data` argument provides the background distribution for features.
    explainer = shap.LinearExplainer(
        model={'coefs': coefficients.values, 'intercept': intercept},
        data=X_data
    )

    # --- Step 2: Compute Mean Absolute SHAP Values ---

    # Calculate the SHAP values for every observation in the provided data.
    # The result is a numpy array of shape (n_samples, n_features).
    shap_values = explainer.shap_values(X_data)

    # To get a global importance measure, compute the mean of the absolute SHAP
    # values for each feature across all samples. This represents the average
    # magnitude of a feature's impact on the model's prediction.
    # Equation: I_j = (1/N) * sum(|phi_{i,j}| for i=1..N)
    mean_abs_shap = np.abs(shap_values).mean(axis=0)

    # --- Step 3: Normalize SHAP Importances to Percentages ---

    # Calculate the sum of all mean absolute SHAP values, which represents the
    # total average prediction magnitude to be distributed among features.
    total_shap_importance = np.sum(mean_abs_shap)

    # Handle the edge case where total importance is zero to avoid division by zero.
    if total_shap_importance == 0:
        # If total importance is zero, all features are equally unimportant.
        # We assign a uniform importance of zero to each.
        num_features = len(X_data.columns)
        normalized_importances = np.zeros(num_features)
    else:
        # Normalize each feature's importance to a percentage of the total.
        # This converts the absolute contributions into relative shares.
        # Equation: I_hat_j = (I_j / sum(I_k)) * 100
        normalized_importances = (mean_abs_shap / total_shap_importance) * 100

    # Create a dictionary mapping the feature names to their normalized importance scores.
    importance_dict = dict(zip(X_data.columns, normalized_importances))

    # Return the final, structured importance report.
    return importance_dict


def compute_shap_importance_for_baselines(
    events_df: pd.DataFrame,
    fitted_models: Dict[str, PanelResults],
    config: StudyConfig
) -> Dict[str, ShapImportanceDict]:
    """
    Orchestrates the computation of SHAP feature importance for pre-fitted
    baseline regression models.

    This function iterates through a dictionary of fitted `linearmodels`
    regression objects (from Task 9). For each model, it computes the
    SHAP-based feature importance for the earnings surprise and the specific
    soft signal, ensuring perfect consistency between the regression analysis
    and the feature attribution.

    Args:
        events_df: The fully prepared DataFrame with all features and signals.
        fitted_models: A dictionary where keys are model names (e.g., 'finbert')
                       and values are the corresponding fitted `PanelResults`
                       objects from Task 9.
        config: The validated study configuration object.

    Returns:
        A dictionary where keys are the model names and values are dictionaries
        containing the normalized SHAP importance percentages for each feature.
    """
    # --- Input Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    if not isinstance(fitted_models, dict):
        raise TypeError("Input 'fitted_models' must be a dictionary of PanelResults.")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # Initialize a dictionary to store the final importance reports.
    all_shap_results: Dict[str, ShapImportanceDict] = {}

    # --- Main Loop: Iterate Through Each Pre-Fitted Model ---
    for method, results_obj in fitted_models.items():
        # The independent variables are the index of the params Series, excluding the intercept.
        independent_vars = results_obj.params.index.drop('const').tolist()
        dependent_var = 'ret_announce_cc' # As defined in the regression task.

        # --- Step 1: Prepare the Exact Same Data Sample ---
        # It is critical to use the exact same data sample that the model was
        # trained on for the SHAP explanation. We regenerate it here.
        sample_df = _prepare_regression_sample(
            events_df, dependent_var, independent_vars, config
        )

        # Skip if the sample is empty.
        if sample_df.empty:
            continue

        # Extract the feature matrix (X data) for the explainer.
        X_data = sample_df[independent_vars]

        # --- Step 2: Compute and Normalize SHAP Values ---
        # Pass the fitted model's parameters and the corresponding feature data
        # to the SHAP helper function.
        shap_importance = _compute_and_normalize_shap_values(
            model_params=results_obj.params,
            X_data=X_data
        )

        # Store the computed importance dictionary for the current method.
        all_shap_results[method] = shap_importance

    # Return the dictionary containing SHAP importance for all baseline models.
    return all_shap_results
````

````python
# Task 12: Label oLDA Topics via LLM

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

class LLMParams(BaseModel):
    """
    A Pydantic model for validating the 'llm' section of the study
    configuration.

    This model defines the schema for parameters controlling interactions with
    a Large Language Model (LLM) API, such as OpenAI's GPT models. It ensures
    that all API call parameters are correctly specified.

    Attributes:
        model_identifier (str): The specific model identifier to be used for
                                the API call (e.g., 'gpt-4o-2024-08-06').
        temperature (float): The sampling temperature, controlling randomness.
                             Must be between 0.0 (deterministic) and 2.0.
        top_p (float): The nucleus sampling parameter. Must be between 0.0
                       and 1.0.
        response_format (Dict[str, str]): Specifies the required format for the
            LLM's response. For structured output, this is typically
            `{'type': 'json_object'}`.
        request_timeout_seconds (int): The maximum time in seconds to wait for
                                       a response from the API. Must be positive.
        max_retries (int): The maximum number of times to retry a failed API
                           call. Must be a non-negative integer.
        rate_limit_rps (int): The maximum number of requests per second to send
                              to the API. Must be a positive integer.
    """
    # The identifier for the specific LLM to be queried.
    model_identifier: str = Field(..., min_length=1)

    # Controls the randomness of the LLM's output. 0.0 is deterministic.
    temperature: float = Field(..., ge=0.0, le=2.0)

    # The nucleus sampling parameter. 1.0 considers all tokens.
    top_p: float = Field(..., ge=0.0, le=1.0)

    # A dictionary specifying the desired output format from the API.
    response_format: Dict[str, str]

    # The timeout in seconds for each API request.
    request_timeout_seconds: int = Field(..., gt=0)

    # The number of retry attempts for a failed API call.
    max_retries: int = Field(..., ge=0)

    # The target rate limit in requests per second.
    rate_limit_rps: int = Field(..., gt=0)


class ChunkingParams(BaseModel):
    """
    A Pydantic model for validating the 'chunking' section of the study
    configuration.

    This model defines parameters for splitting large amounts of data into
    smaller chunks before sending them to an LLM, to respect context window
    limits and manage API costs.

    Attributes:
        topic_batch_size (int): The number of oLDA topics to include in a
                                single API request for labeling. Must be a
                                positive integer.
    """
    # The number of topics to group into a single batch for an LLM query.
    topic_batch_size: int = Field(..., gt=0)


class Prompts(BaseModel):
    """
    A Pydantic model for validating the 'prompts' section of the study
    configuration.

    This model stores the templates for the system and user prompts that will
    be sent to the LLM. Storing them in the configuration makes them easy to
    manage and modify.

    Attributes:
        topic_labeler_system_prompt (str): The system-level instruction that
            sets the context and constraints for the LLM's task.
        topic_labeler_user_prompt (str): The user-level instruction that
            provides the specific data and task for the LLM to perform.
    """
    # The system prompt template, defining the LLM's role and overall rules.
    topic_labeler_system_prompt: str = Field(..., min_length=1)

    # The user prompt template, which will be formatted with specific data.
    topic_labeler_user_prompt: str = Field(..., min_length=1)


class InterpretabilityParams(BaseModel):
    """
    A Pydantic model for validating the 'interpretability_params' section of
    the study configuration.

    This model serves as a container for all parameters related to making the
    NLP models interpretable, primarily through the use of LLMs.

    Attributes:
        llm (LLMParams): Validated parameters for interacting with the LLM API.
        chunking (ChunkingParams): Validated parameters for data chunking.
        prompts (Prompts): Validated prompt templates.
    """
    # A nested model for the LLM API interaction parameters.
    llm: LLMParams

    # A nested model for data chunking parameters.
    chunking: ChunkingParams

    # A nested model for the prompt templates.
    prompts: Prompts


class StudyConfig(BaseModel):
    """
    A Pydantic model representing a subsection of the main 'study_config'
    dictionary relevant to Task 12.

    This model ensures that all parameters required for the LLM-based topic
    labeling task are present and correctly structured.

    Attributes:
        interpretability_params (InterpretabilityParams): The top-level
            container for all interpretability-related settings.
    """
    # The main container for all interpretability parameters.
    interpretability_params: InterpretabilityParams

# Define a type alias for the structured topic data.
TopicData = List[Dict[str, Any]]

# ==============================================================================
# Task 12: Helper and Orchestrator Functions
# ==============================================================================

def _extract_top_tokens_for_lda(
    lda_model: LatentDirichletAllocation,
    vectorizer: CountVectorizer,
    n_top_words: int = 10
) -> TopicData:
    """
    Extracts the top N most probable words for each topic from a fitted LDA model.

    Args:
        lda_model: A fitted scikit-learn LatentDirichletAllocation model.
        vectorizer: The corresponding fitted CountVectorizer used to create the
                    document-term matrix for the LDA model.
        n_top_words: The number of top words to extract for each topic.

    Returns:
        A list of dictionaries, where each dictionary represents a topic and
        contains its ID and a list of its top words.
    """
    # Retrieve the vocabulary (mapping of index to word) from the vectorizer.
    feature_names = vectorizer.get_feature_names_out()

    # Initialize a list to store the data for each topic.
    topics_data: TopicData = []

    # Iterate through each topic's word distribution in the model's components.
    for topic_idx, topic_distribution in enumerate(lda_model.components_):
        # Get the indices of the top N words by sorting the distribution probabilities.
        top_feature_indices = topic_distribution.argsort()[:-n_top_words - 1:-1]

        # Map the indices back to the actual words from the vocabulary.
        top_words = [feature_names[i] for i in top_feature_indices]

        # Append the structured topic data to the results list.
        topics_data.append({"topic_id": topic_idx, "top_tokens": top_words})

    return topics_data


def _query_llm_for_topic_labels(
    topics_chunk: TopicData,
    config: StudyConfig
) -> List[Dict[str, Any]]:
    """
    Queries the OpenAI API to get labels and definitions for a chunk of topics.

    This function handles the complete API interaction lifecycle, including
    prompt formatting, request sending, rate limiting, and robust error
    handling with retries.

    Args:
        topics_chunk: A list of topic dictionaries to be labeled in this request.
        config: The validated study configuration object.

    Returns:
        A list of dictionaries, where each dictionary contains the LLM-generated
        label and definition for a topic.

    Raises:
        RuntimeError: If the API call fails after all retry attempts.
        ValueError: If the API key is not found in environment variables.
    """
    # Retrieve LLM and prompt parameters from the configuration.
    llm_params = config.interpretability_params.llm
    prompts = config.interpretability_params.prompts

    # Securely retrieve the API key from environment variables.
    api_key = os.getenv("OPENAI_API_KEY")
    if not api_key:
        raise ValueError("OPENAI_API_KEY environment variable not set.")

    # Initialize the OpenAI client.
    client = OpenAI(api_key=api_key)

    # Format the user prompt by injecting the topic data as a JSON string.
    user_prompt = prompts.topic_labeler_user_prompt.format(
        topics_json=json.dumps({"topics": topics_chunk}, indent=2)
    )

    # --- API Call with Retry Logic ---
    for attempt in range(llm_params.max_retries):
        try:
            # Send the request to the chat completions endpoint.
            response = client.chat.completions.create(
                model=llm_params.model_identifier,
                messages=[
                    {"role": "system", "content": prompts.topic_labeler_system_prompt},
                    {"role": "user", "content": user_prompt},
                ],
                temperature=llm_params.temperature,
                top_p=llm_params.top_p,
                response_format=llm_params.response_format,
                timeout=llm_params.request_timeout_seconds,
            )

            # Extract the JSON content from the response.
            response_content = response.choices[0].message.content
            if response_content is None:
                raise ValueError("API returned an empty response content.")

            # Parse the JSON string into a Python dictionary.
            parsed_response = json.loads(response_content)

            # Perform basic validation on the parsed response structure.
            if "labeled_topics" not in parsed_response or not isinstance(parsed_response["labeled_topics"], list):
                raise ValueError("LLM response is missing 'labeled_topics' list.")

            # If successful, return the list of labeled topics.
            return parsed_response["labeled_topics"]

        except (APIError, ValueError) as e:
            # If an error occurs, wait before retrying using exponential backoff.
            wait_time = (2 ** attempt)
            time.sleep(wait_time)
            # If this was the last attempt, re-raise the error as a RuntimeError.
            if attempt == llm_params.max_retries - 1:
                raise RuntimeError(f"LLM API call failed after {llm_params.max_retries} attempts: {e}")

    # This line should be unreachable, but is included for type safety.
    raise RuntimeError("Exited retry loop unexpectedly.")


def label_lda_topics_with_llm(
    lda_model: LatentDirichletAllocation,
    vectorizer: CountVectorizer,
    config: StudyConfig
) -> pd.DataFrame:
    """
    Orchestrates the process of labeling oLDA topics using an LLM.

    This function takes a fitted LDA model and its vectorizer, extracts the
    top words for each topic, queries an LLM (like GPT-4o) to generate
    human-readable labels, and returns the result as a structured DataFrame.

    Args:
        lda_model: A fitted scikit-learn LatentDirichletAllocation model.
        vectorizer: The corresponding fitted CountVectorizer.
        config: The validated study configuration object.

    Returns:
        A pandas DataFrame indexed by 'topic_id', containing the top tokens,
        the LLM-generated label, and definition for each topic.
    """
    # --- Input Validation ---
    if not isinstance(lda_model, LatentDirichletAllocation) or not hasattr(lda_model, 'components_'):
        raise TypeError("Input 'lda_model' must be a fitted LatentDirichletAllocation instance.")
    if not isinstance(vectorizer, CountVectorizer) or not hasattr(vectorizer, 'vocabulary_'):
        raise TypeError("Input 'vectorizer' must be a fitted CountVectorizer instance.")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # --- Step 1: Extract Top Tokens for All Topics ---
    all_topics_data = _extract_top_tokens_for_lda(lda_model, vectorizer)

    # --- Step 2: Chunk Data and Query LLM ---
    chunk_size = config.interpretability_params.chunking.topic_batch_size
    rate_limit_delay = 1.0 / config.interpretability_params.llm.rate_limit_rps

    # Initialize a list to store labeled topics from all API calls.
    all_labeled_topics: List[Dict[str, Any]] = []

    # Process the topics in chunks to manage API request size.
    for i in range(0, len(all_topics_data), chunk_size):
        # Get the current chunk of topics.
        topics_chunk = all_topics_data[i:i + chunk_size]

        # Query the LLM for labels for this chunk.
        labeled_chunk = _query_llm_for_topic_labels(topics_chunk, config)

        # Extend the master list with the results from this chunk.
        all_labeled_topics.extend(labeled_chunk)

        # Enforce rate limiting by pausing between API calls.
        time.sleep(rate_limit_delay)

    # --- Step 3: Validate and Structure Final Output ---
    # Convert the original top token data and the LLM labels into DataFrames.
    top_tokens_df = pd.DataFrame(all_topics_data).set_index('topic_id')
    labeled_topics_df = pd.DataFrame(all_labeled_topics).set_index('topic_id')

    # Merge the two DataFrames on their common 'topic_id' index.
    final_df = top_tokens_df.join(labeled_topics_df)

    # Validate that all topics were successfully labeled.
    if final_df['topic_label'].isnull().any():
        raise RuntimeError("Failed to retrieve labels for all topics from the LLM.")

    return final_df
````

````python
# Task 13: Condense Labeled Topics into Metatopics via LLM

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

LabeledTopicsDF = pd.DataFrame
MetatopicTaxonomy = List[Dict[str, Any]]

# ==============================================================================
# Task 13: Helper and Orchestrator Functions
# ==============================================================================

def _validate_metatopic_taxonomy(
    metatopics: MetatopicTaxonomy,
    expected_topic_ids: set,
    expected_metatopic_count: int
) -> None:
    """
    Performs rigorous validation on the LLM-generated metatopic taxonomy.

    This function enforces three critical structural constraints:
    1. Correct Count: The number of metatopics must match the expected count.
    2. Mutual Exclusivity: Each topic ID must be assigned to only one metatopic.
    3. Exhaustiveness: The set of all assigned topic IDs must exactly match the
       set of original topic IDs.

    Args:
        metatopics: The list of metatopic dictionaries parsed from the LLM response.
        expected_topic_ids: A set of all original topic IDs (e.g., {0, 1, ..., 49}).
        expected_metatopic_count: The exact number of metatopics requested.

    Raises:
        ValueError: If any of the structural constraints are violated.
    """
    # 1. Validate the number of metatopics generated.
    if len(metatopics) != expected_metatopic_count:
        raise ValueError(f"LLM returned {len(metatopics)} metatopics, but {expected_metatopic_count} were expected.")

    # 2. Validate mutual exclusivity and exhaustiveness.
    all_assigned_ids: List[int] = []
    for meta in metatopics:
        # Check for required keys in each metatopic dictionary.
        if not all(k in meta for k in ['metatopic_id', 'topic_ids']):
            raise ValueError(f"Metatopic dictionary is malformed: {meta}")
        all_assigned_ids.extend(meta['topic_ids'])

    # Check for duplicate assignments (violates mutual exclusivity).
    if len(all_assigned_ids) != len(set(all_assigned_ids)):
        raise ValueError("Taxonomy validation failed: Some topic IDs were assigned to multiple metatopics.")

    # Check that the set of assigned IDs perfectly matches the original set.
    if set(all_assigned_ids) != expected_topic_ids:
        raise ValueError("Taxonomy validation failed: The set of assigned topic IDs does not match the original set.")


def condense_topics_into_metatopics(
    labeled_topics_df: LabeledTopicsDF,
    config: StudyConfig
) -> MetatopicTaxonomy:
    """
    Orchestrates the condensation of labeled topics into a structured
    metatopic taxonomy using an LLM.

    This function takes the 50 human-labeled topics, queries an LLM with a
    specialized prompt to cluster them into a smaller set of metatopics,
    and rigorously validates the response to ensure it forms a coherent and
    structurally correct taxonomy.

    Args:
        labeled_topics_df: A DataFrame of labeled topics from Task 12,
                           indexed by 'topic_id'.
        config: The validated study configuration object.

    Returns:
        A list of dictionaries representing the final, validated metatopic
        taxonomy. Each dictionary contains the metatopic's ID, label,
        definition, and a list of its constituent topic IDs.

    Raises:
        RuntimeError: If the LLM fails to produce a valid taxonomy after all
                      retry attempts.
        ValueError: If inputs are invalid or the API key is not set.
    """
    # --- Input Validation ---
    if not isinstance(labeled_topics_df, pd.DataFrame):
        raise TypeError("Input 'labeled_topics_df' must be a pandas DataFrame.")
    required_cols = ['topic_label', 'definition']
    if not all(col in labeled_topics_df.columns for col in required_cols):
        raise ValueError(f"Input DataFrame is missing required columns: {required_cols}")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # --- Step 1: Prepare the API Payload ---
    # Retrieve LLM, prompt, and policy parameters from the configuration.
    llm_params = config.interpretability_params.llm
    prompts = config.interpretability_params.prompts
    policy = config.interpretability_params.taxonomy_policy

    # Convert the labeled topics DataFrame to the required JSON format.
    labeled_topics_list = labeled_topics_df.reset_index().to_dict(orient='records')
    payload_json = json.dumps({"labeled_topics": labeled_topics_list}, indent=2)

    # Format the user prompt with the data and the target metatopic count.
    user_prompt = prompts.metatopic_condenser_user_prompt.format(
        n_metatopics=policy.metatopic_count,
        labeled_topics_json=payload_json
    )

    # --- Step 2: Query LLM with Integrated Validation and Retry Logic ---
    # Securely retrieve the API key from environment variables.
    api_key = os.getenv("OPENAI_API_KEY")
    if not api_key:
        raise ValueError("OPENAI_API_KEY environment variable not set.")

    # Initialize the OpenAI client.
    client = OpenAI(api_key=api_key)

    # Define the set of all original topic IDs for validation.
    expected_ids = set(labeled_topics_df.index)

    # Implement the API call within a retry loop.
    for attempt in range(llm_params.max_retries):
        try:
            # Send the request to the chat completions endpoint.
            response = client.chat.completions.create(
                model=llm_params.model_identifier,
                messages=[
                    {"role": "system", "content": prompts.metatopic_condenser_system_prompt.format(n_metatopics=policy.metatopic_count)},
                    {"role": "user", "content": user_prompt},
                ],
                temperature=llm_params.temperature,
                top_p=llm_params.top_p,
                response_format=llm_params.response_format,
                timeout=llm_params.request_timeout_seconds,
            )

            # Extract and parse the JSON response content.
            response_content = response.choices[0].message.content
            if response_content is None:
                raise ValueError("API returned an empty response content.")
            parsed_response = json.loads(response_content)

            # --- Step 3: Rigorously Validate the Parsed Response ---
            if "metatopics" not in parsed_response or not isinstance(parsed_response["metatopics"], list):
                raise ValueError("LLM response is missing 'metatopics' list.")

            # Call the dedicated validation function to check structural integrity.
            _validate_metatopic_taxonomy(
                parsed_response["metatopics"],
                expected_ids,
                policy.metatopic_count
            )

            # If validation passes, the taxonomy is correct. Return it.
            return parsed_response["metatopics"]

        except (APIError, ValueError, json.JSONDecodeError) as e:
            # If any error occurs (API error, parsing error, or validation error),
            # wait before retrying.
            logging.warning(f"Attempt {attempt + 1} failed: {e}. Retrying...")
            if attempt == llm_params.max_retries - 1:
                # If this was the last attempt, raise a final error.
                raise RuntimeError(f"LLM failed to produce a valid metatopic taxonomy after {llm_params.max_retries} attempts: {e}")

            # Apply rate limiting and exponential backoff.
            time.sleep((2 ** attempt))

    # This line should be unreachable but is a safeguard.
    raise RuntimeError("Exited retry loop unexpectedly.")
````

````python
# Task 14: Extract Influential Tokens from BERT-Family Embeddings

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

# Define a type alias for the structured output of influential tokens.
InfluentialTokens = Dict[str, List[Tuple[str, float]]]

# ==============================================================================
# Task 14: Helper and Orchestrator Functions
# ==============================================================================

def _filter_and_select_top_tokens(
    tokens: List[str],
    scores: np.ndarray,
    n_top: int = 5
) -> InfluentialTokens:
    """
    Filters out uninformative tokens and selects the top N positive and negative
    tokens based on their importance scores.

    Args:
        tokens: A list of tokens for a single document.
        scores: A numpy array of corresponding importance scores for each token.
        n_top: The number of top positive and top negative tokens to select.

    Returns:
        A dictionary containing two lists: 'positive' and 'negative', with each
        list containing tuples of (token, score).
    """
    # Combine tokens and scores, and filter out uninformative ones.
    filtered_tokens = []
    for token, score in zip(tokens, scores):
        # Filter criteria:
        # - Exclude special tokens like [CLS], [SEP], [PAD].
        # - Exclude WordPiece fragments (starting with ##).
        # - Exclude tokens that are purely punctuation or numbers.
        if (token not in ['[CLS]', '[SEP]', '[PAD]', '[UNK]', '[MASK]'] and
                not token.startswith('##') and
                any(c.isalpha() for c in token)):
            filtered_tokens.append((token, score))

    # Sort the filtered tokens by score in descending order.
    filtered_tokens.sort(key=lambda x: x[1], reverse=True)

    # Select the top N positive tokens (from the beginning of the sorted list).
    top_positive = filtered_tokens[:n_top]

    # Select the top N negative tokens (from the end of the sorted list).
    top_negative = filtered_tokens[-n_top:]

    return {'positive': top_positive, 'negative': top_negative}


# ==============================================================================
# Implementation of extract_influential_tokens
# ==============================================================================

def extract_influential_tokens(
    events_df: pd.DataFrame,
    model_key: str,
    artifacts_path: Path,
) -> pd.Series:
    """
    Orchestrates the extraction of influential tokens by loading pre-computed
    embeddings and model weights.

    This function implements a highly efficient, production-grade version of the
    token interpretability pipeline. It avoids redundant computation by loading
    yearly Lasso weights and per-document token-level embeddings from disk. It
    then performs the core calculation to map feature weights back to token
    importance scores.

    Args:
        events_df: The DataFrame with event metadata, including 'pr_timestamp_et'
                   and a unique document identifier (e.g., 'event_id').
        model_key: The short name for the model (e.g., 'finbert').
        artifacts_path: The base path to the directory where pre-computed
                        artifacts (Lasso models, token embeddings) are stored.

    Returns:
        A pandas Series, indexed like `events_df`, where each element is a
        dictionary containing the top positive and negative influential tokens.

    Raises:
        FileNotFoundError: If required artifact files (models or embeddings)
                           are not found.
    """
    # --- Step 0: Initialization and Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    if 'event_id' not in events_df.index.names:
        raise ValueError("DataFrame must have 'event_id' in its index to locate embeddings.")

    # Define paths to the pre-computed artifacts.
    lasso_models_path = artifacts_path / "lasso_models" / model_key
    token_embeddings_path = artifacts_path / "token_embeddings" / model_key

    # Load the tokenizer once to convert token IDs back to strings.
    # This assumes the model name can be inferred or is passed in config.
    # For this example, we hardcode the FinBERT tokenizer.
    tokenizer = PreTrainedTokenizer.from_pretrained("ProsusAI/finbert")

    # Add a temporary 'calendar_year' column for partitioning.
    df = events_df.copy()
    df['calendar_year'] = df['pr_timestamp_et'].dt.year

    # List to store results from each scoring year.
    all_results: List[pd.Series] = []

    # --- Main Rolling Loop ---
    score_years = sorted([y for y in df['calendar_year'].unique() if y >= 2006])
    for score_year in score_years:
        train_year = score_year - 1

        # --- Step 1: Load Pre-computed Lasso Weights ---
        # Construct the path to the Lasso model trained on the previous year's data.
        lasso_model_path = lasso_models_path / f"{train_year}.pkl"
        try:
            # Load the fitted Lasso model object.
            lasso_model: Lasso = joblib.load(lasso_model_path)
            # Get the coefficient vector.
            lasso_weights = torch.tensor(lasso_model.coef_, dtype=torch.float32)
        except FileNotFoundError:
            raise FileNotFoundError(f"Required Lasso model not found: {lasso_model_path}")

        # --- Step 2: Iterate Through Documents of the Scoring Year ---
        # Get the subset of documents for the current scoring year.
        score_df = df[df['calendar_year'] == score_year]
        yearly_results = {}

        for doc_id, row in score_df.iterrows():
            # Construct the path to the document's token embedding tensor.
            # doc_id is a tuple from the MultiIndex, we need the event_id.
            event_id = doc_id[2]
            embedding_path = token_embeddings_path / f"{event_id}.pt"

            try:
                # Load the (512, 768) last_hidden_state tensor.
                token_embeddings: torch.Tensor = torch.load(embedding_path)
            except FileNotFoundError:
                # Skip documents for which we don't have a pre-computed embedding.
                continue

            # --- Step 3: Compute Token Importance Scores ---
            # Equation: IS = E @ w
            importance_scores = torch.matmul(token_embeddings, lasso_weights)

            # --- Step 4: Filter and Select Top Tokens ---
            # We need the original tokens, which requires re-tokenizing or loading them.
            # A more robust pipeline would save token IDs alongside the embeddings.
            # Here, we re-tokenize for simplicity.
            inputs = tokenizer(row['press_release_text_clean'], return_tensors='pt', max_length=512, truncation=True)
            tokens = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])

            # Get the top positive/negative tokens using the helper function.
            top_tokens = _filter_and_select_top_tokens(tokens, importance_scores.numpy())
            yearly_results[doc_id] = top_tokens

        # Convert the year's results to a Series and append if not empty.
        if yearly_results:
            all_results.append(pd.Series(yearly_results))

    # --- Final Aggregation ---
    if not all_results:
        return pd.Series(dtype='object', name=f'influential_tokens_{model_key}')

    # Concatenate results from all years into a single Series.
    final_series = pd.concat(all_results)
    final_series.name = f'influential_tokens_{model_key}'

    return final_series
````

````python
# Task 15: Classify Influential Tokens into Metatopics via LLM

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

InfluentialTokensSeries = pd.Series
MetatopicTaxonomy = List[Dict[str, Any]]
TokenPolarityList = List[Dict[str, str]]
TokenToMetatopicMap = Dict[str, str]

# ==============================================================================
# Task 15: Helper and Orchestrator Functions
# ==============================================================================

def _aggregate_unique_tokens(
    tokens_series: InfluentialTokensSeries
) -> TokenPolarityList:
    """
    Aggregates all influential tokens from a series, determines a single
    dominant polarity for each unique token, and returns a unique list.

    Args:
        tokens_series: A pandas Series where each element is a dictionary
                       containing 'positive' and 'negative' token lists.

    Returns:
        A list of unique tokens, where each token is a dictionary with
        'token' and 'polarity' keys.
    """
    # Use a Counter to track the polarity occurrences for each token.
    polarity_counts: Dict[str, Counter] = {}

    # Iterate through the series of influential token dictionaries.
    for _, token_dict in tokens_series.dropna().items():
        # Process positive tokens.
        for token, _ in token_dict.get('positive', []):
            if token not in polarity_counts:
                polarity_counts[token] = Counter()
            polarity_counts[token]['positive'] += 1
        # Process negative tokens.
        for token, _ in token_dict.get('negative', []):
            if token not in polarity_counts:
                polarity_counts[token] = Counter()
            polarity_counts[token]['negative'] += 1

    # Determine the dominant polarity for each unique token.
    unique_token_list: TokenPolarityList = []
    for token, counts in polarity_counts.items():
        # Assign polarity based on the most common occurrence.
        # Defaults to 'positive' in case of a tie.
        dominant_polarity = counts.most_common(1)[0][0]
        unique_token_list.append({'token': token, 'polarity': dominant_polarity})

    return unique_token_list


def classify_influential_tokens(
    influential_tokens: InfluentialTokensSeries,
    metatopic_taxonomy: MetatopicTaxonomy,
    config: StudyConfig
) -> TokenToMetatopicMap:
    """
    Orchestrates the classification of influential tokens into a predefined
    metatopic taxonomy using an LLM.

    This function aggregates unique influential tokens, chunks them, queries an
    LLM with a classification prompt, and validates the structured response to
    create a final token-to-metatopic mapping.

    Args:
        influential_tokens: A pandas Series of influential tokens from Task 14.
        metatopic_taxonomy: The validated metatopic taxonomy from Task 13.
        config: The validated study configuration object.

    Returns:
        A dictionary mapping each classified token to its metatopic ID.

    Raises:
        RuntimeError: If the LLM fails to produce a valid classification.
        ValueError: If inputs are invalid or the API key is not set.
    """
    # --- Input Validation ---
    if not isinstance(influential_tokens, pd.Series):
        raise TypeError("Input 'influential_tokens' must be a pandas Series.")
    if not isinstance(metatopic_taxonomy, list):
        raise TypeError("Input 'metatopic_taxonomy' must be a list.")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # --- Step 1: Aggregate Unique Tokens and Prepare Payloads ---
    unique_tokens = _aggregate_unique_tokens(influential_tokens)

    # Retrieve parameters from the configuration.
    llm_params = config.interpretability_params.llm
    prompts = config.interpretability_params.prompts
    chunk_size = config.interpretability_params.chunking.token_chunk_size_for_categorization

    # Prepare static parts of the prompt.
    metatopics_json = json.dumps(metatopic_taxonomy, indent=2)
    valid_metatopic_ids = {meta['metatopic_id'] for meta in metatopic_taxonomy} | {'Other'}

    # --- Step 2: Query LLM in Chunks ---
    # Securely retrieve the API key and initialize the client.
    api_key = os.getenv("OPENAI_API_KEY")
    if not api_key:
        raise ValueError("OPENAI_API_KEY environment variable not set.")
    client = OpenAI(api_key=api_key)

    # Initialize the final mapping dictionary.
    final_token_map: TokenToMetatopicMap = {}
    rate_limit_delay = 1.0 / llm_params.rate_limit_rps

    # Process the unique tokens in chunks.
    for i in range(0, len(unique_tokens), chunk_size):
        token_chunk = unique_tokens[i:i + chunk_size]
        tokens_chunk_json = json.dumps({"tokens": token_chunk}, indent=2)

        # Format the user prompt for the current chunk.
        user_prompt = prompts.token_classifier_user_prompt.format(
            metatopics_json=metatopics_json,
            tokens_chunk_json=tokens_chunk_json
        )

        # API call with retry logic.
        for attempt in range(llm_params.max_retries):
            try:
                response = client.chat.completions.create(
                    model=llm_params.model_identifier,
                    messages=[
                        {"role": "system", "content": prompts.token_classifier_system_prompt},
                        {"role": "user", "content": user_prompt},
                    ],
                    temperature=llm_params.temperature,
                    response_format=llm_params.response_format,
                    timeout=llm_params.request_timeout_seconds,
                )

                response_content = response.choices[0].message.content
                if response_content is None:
                    raise ValueError("API returned empty content.")

                parsed_response = json.loads(response_content)

                # --- Step 3: Validate and Store Results ---
                if "classified_tokens" not in parsed_response:
                    raise ValueError("Response missing 'classified_tokens' key.")

                classified_chunk = parsed_response["classified_tokens"]

                # Rigorous validation of the chunk's results.
                if len(classified_chunk) != len(token_chunk):
                    raise ValueError(f"LLM returned {len(classified_chunk)} items for a chunk of size {len(token_chunk)}.")

                for item in classified_chunk:
                    if item['metatopic_id'] not in valid_metatopic_ids:
                        raise ValueError(f"LLM returned an invalid metatopic_id: {item['metatopic_id']}")
                    final_token_map[item['token']] = item['metatopic_id']

                # If validation passes, break the retry loop.
                break

            except (APIError, ValueError, json.JSONDecodeError) as e:
                if attempt == llm_params.max_retries - 1:
                    raise RuntimeError(f"LLM token classification failed after {llm_params.max_retries} attempts: {e}")
                time.sleep(2 ** attempt)

        # Enforce rate limit between successful chunk requests.
        time.sleep(rate_limit_delay)

    return final_token_map
````

````python
# Task 16: Compute Metatopic Weights and Explained Variance (oLDA)

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

MetatopicTaxonomy = List[Dict[str, Any]]
YearlyLassoWeights = Dict[int, np.ndarray]
AnalysisResults = pd.DataFrame

# ==============================================================================
# Task 16: Helper and Orchestrator Functions
# ==============================================================================

def _prepare_metatopic_mappings(
    taxonomy: MetatopicTaxonomy,
    num_topics: int
) -> Tuple[Dict[str, List[int]], pd.Series]:
    """
    Prepares efficient lookup mappings from a metatopic taxonomy structure.

    This utility function takes the list-based taxonomy (e.g., from an LLM
    output) and processes it into two key data structures needed for efficient
    downstream analysis:
    1. A dictionary mapping each metatopic name to the list of its topic IDs.
    2. A pandas Series mapping each topic ID to its parent metatopic name,
       which is ideal for use in `groupby` operations.

    Args:
        taxonomy: The metatopic taxonomy, structured as a list of dictionaries.
                  Each dictionary must contain 'metatopic_label' and 'topic_ids'.
        num_topics: The total number of topics in the underlying topic model
                    (e.g., 50 for oLDA).

    Returns:
        A tuple containing:
        - metatopic_to_ids: A dictionary mapping metatopic names (str) to
                            lists of their constituent topic IDs (int).
        - topic_to_metatopic_series: A pandas Series where the index is the
                                     topic ID (int) and the value is the
                                     corresponding metatopic name (str).

    Raises:
        TypeError: If the input `taxonomy` is not a list.
        KeyError: If dictionaries within the taxonomy are missing required keys.
    """
    # --- Input Validation ---
    if not isinstance(taxonomy, list):
        raise TypeError("Input 'taxonomy' must be a list of dictionaries.")

    # --- Step 1: Create Metatopic-to-TopicIDs Mapping ---
    try:
        # Use a dictionary comprehension to create a mapping from each metatopic's
        # label to its list of associated topic IDs.
        metatopic_to_ids: Dict[str, List[int]] = {
            meta['metatopic_label']: meta['topic_ids'] for meta in taxonomy
        }
    except KeyError as e:
        # Raise an error if the taxonomy structure is missing expected keys.
        raise KeyError(f"Taxonomy dictionary is malformed. Missing key: {e}")

    # --- Step 2: Create TopicID-to-Metatopic Reverse Mapping ---
    # Use a nested dictionary comprehension to create a "reverse" mapping from
    # each individual topic ID back to its parent metatopic's name.
    id_to_metatopic: Dict[int, str] = {
        topic_id: meta_name
        for meta_name, topic_ids in metatopic_to_ids.items()
        for topic_id in topic_ids
    }

    # --- Step 3: Convert Reverse Mapping to a Pandas Series ---
    # Create a pandas Series from the reverse mapping dictionary.
    # The index of the Series will be the topic IDs.
    # This format is highly efficient for performing groupby operations on other
    # Series or DataFrame columns that are indexed by topic ID.
    topic_to_metatopic_series = pd.Series(id_to_metatopic, index=range(num_topics)).sort_index()

    # Return both the forward and reverse mapping structures.
    return metatopic_to_ids, topic_to_metatopic_series

def compute_olda_metatopic_analytics(
    events_df: pd.DataFrame,
    olda_lasso_weights: YearlyLassoWeights,
    metatopic_taxonomy: MetatopicTaxonomy,
    num_topics: int = 50
) -> Tuple[AnalysisResults, AnalysisResults]:
    """
    Computes the time-series of metatopic weights (polarity) and explained
    variance shares for the oLDA model.

    This function performs a deep dive into the oLDA model's dynamics by:
    1. Aggregating the topic-level Lasso weights to the metatopic level for each year.
    2. Decomposing the total variance explained by the oLDA soft signal into
       the share contributed by each metatopic for each year.

    Args:
        events_df: The DataFrame containing the oLDA topic attention columns.
        olda_lasso_weights: A dictionary mapping each training year (e.g., 2005)
                            to the numpy array of fitted Lasso coefficients for
                            the 50 oLDA topics.
        metatopic_taxonomy: The validated metatopic taxonomy from Task 13.
        num_topics: The number of topics in the oLDA model.

    Returns:
        A tuple of two DataFrames:
        - weights_df: A DataFrame with (year, metatopic) MultiIndex showing the
                      aggregate Lasso weight for each metatopic over time.
        - variance_df: A DataFrame with (year, metatopic) MultiIndex showing the
                       explained variance share for each metatopic over time.
    """
    # --- Input Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    if not isinstance(olda_lasso_weights, dict):
        raise TypeError("Input 'olda_lasso_weights' must be a dictionary.")
    if not isinstance(metatopic_taxonomy, list):
        raise TypeError("Input 'metatopic_taxonomy' must be a list.")

    # --- Step 0: Initialization ---
    # Prepare mappings from the taxonomy for efficient lookup.
    metatopic_to_ids, topic_to_metatopic_series = _prepare_metatopic_mappings(metatopic_taxonomy, num_topics)

    # Define the years to analyze (scoring years).
    score_years = sorted([y for y in events_df['pr_timestamp_et'].dt.year.unique() if y >= 2006])

    # Get the names of the oLDA feature columns.
    olda_cols = [f'olda_topic_{i}' for i in range(num_topics)]

    # Initialize lists to store the yearly results.
    yearly_weights_data = []
    yearly_variance_data = []

    # --- Main Loop: Iterate Through Each Scoring Year ---
    for year in score_years:
        train_year = year - 1

        # Ensure we have the necessary weights for this year's analysis.
        if train_year not in olda_lasso_weights:
            continue

        # Retrieve the Lasso weights trained on `t-1` data.
        weights_t = olda_lasso_weights[train_year]

        # --- Step 1: Compute Metatopic Total Weights ---
        # Group the weights by metatopic and sum them up.
        # Equation: w_M = sum(w_k for k in M)
        weights_series = pd.Series(weights_t, index=range(num_topics))
        metatopic_weights = weights_series.groupby(topic_to_metatopic_series).sum()

        # Store the results for this year.
        for metatopic, weight in metatopic_weights.items():
            yearly_weights_data.append({'year': year, 'metatopic': metatopic, 'weight': weight})

        # --- Step 2: Compute Topic Attention Covariance Matrix ---
        # Select the data for the current scoring year `t`.
        year_df = events_df[events_df['pr_timestamp_et'].dt.year == year]
        if year_df.empty:
            continue

        # Compute the 50x50 covariance matrix of topic attentions for year `t`.
        cov_matrix_t = year_df[olda_cols].cov().values

        # --- Step 3: Compute Metatopic Explained Variance ---
        # Calculate the total variance explained by the model in year `t`.
        # Equation: TotalVar_t = w_t^T * Sigma_t * w_t
        total_variance = weights_t.T @ cov_matrix_t @ weights_t

        if total_variance == 0:
            continue

        # For each metatopic, calculate its share of the total variance.
        for metatopic, topic_indices in metatopic_to_ids.items():
            # Slice the weights and covariance matrix for the current metatopic.
            w_slice = weights_t[topic_indices]
            cov_slice = cov_matrix_t[np.ix_(topic_indices, topic_indices)]

            # Calculate the variance attributable to this metatopic block.
            # Equation: Var(M)_t = sum_{i in M} sum_{j in M} w_i * w_j * Cov(f_i, f_j)
            metatopic_variance = w_slice.T @ cov_slice @ w_slice

            # Calculate the percentage share.
            variance_share = (metatopic_variance / total_variance) * 100

            # Store the result for this year.
            yearly_variance_data.append({'year': year, 'metatopic': metatopic, 'variance_share': variance_share})

    # --- Final Assembly ---
    # Create DataFrames from the collected yearly data.
    weights_df = pd.DataFrame(yearly_weights_data).set_index(['year', 'metatopic'])
    variance_df = pd.DataFrame(yearly_variance_data).set_index(['year', 'metatopic'])

    return weights_df, variance_df
````

````python
# Task 17: Aggregate Token-Level Polarity Counts by Metatopic (BERT-Family)

# ==============================================================================
# Custom Type Aliases
# ==============================================================================

TokenToMetatopicMap = Dict[str, str]
MetatopicTaxonomy = List[Dict[str, Any]]
AnalysisSummary = pd.DataFrame

# ==============================================================================
# Implementation of analyze_token_polarity_by_metatopic
# ==============================================================================

def analyze_token_polarity_by_metatopic(
    events_df: pd.DataFrame,
    token_maps: Dict[str, TokenToMetatopicMap],
    metatopic_taxonomy: MetatopicTaxonomy
) -> AnalysisSummary:
    """
    Aggregates and analyzes the polarity of influential tokens by metatopic
    using a performant, vectorized approach.

    This function quantifies the thematic content of positive and negative
    narratives as identified by the Transformer-based models. It uses an
    efficient `explode` operation to transform the nested token data into a
    long format, then maps and aggregates the tokens to produce a final
    summary table of counts and percentage shares.

    Args:
        events_df: The main DataFrame, which must contain the influential
                   token columns (e.g., 'influential_tokens_finbert').
        token_maps: A dictionary where keys are model names ('bert', 'finbert')
                    and values are the token-to-metatopic mapping dictionaries.
        metatopic_taxonomy: The validated metatopic taxonomy from Task 13.

    Returns:
        A multi-indexed DataFrame summarizing the token counts and percentage
        shares by model, metatopic, and polarity.
    """
    # --- Input Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    if not isinstance(token_maps, dict):
        raise TypeError("Input 'token_maps' must be a dictionary.")
    if not isinstance(metatopic_taxonomy, list):
        raise TypeError("Input 'metatopic_taxonomy' must be a list.")

    # Initialize a list to store the summary DataFrame for each model.
    all_model_summaries: List[pd.DataFrame] = []

    # Create a mapping from metatopic ID to human-readable label for later use.
    id_to_label_map = {meta['metatopic_id']: meta['metatopic_label'] for meta in metatopic_taxonomy}
    id_to_label_map['Other'] = 'Other'

    # --- Main Loop: Iterate Through Each Model ---
    for model_key in ['bert', 'finbert', 'mpnet']:
        token_col = f'influential_tokens_{model_key}'
        token_map = token_maps.get(model_key)

        if token_col not in events_df.columns or not token_map:
            continue

        # --- Step 1: Un-nest the Influential Tokens Data via Explode ---
        # Select the relevant series and drop documents with no influential tokens.
        token_series = events_df[token_col].dropna()

        # Helper function to transform the dict into a list of (token, polarity) tuples.
        def flatten_tokens(d: Dict[str, List[Tuple[str, float]]]) -> List[Tuple[str, str]]:
            flat_list = []
            flat_list.extend([(token, 'Positive') for token, _ in d.get('positive', [])])
            flat_list.extend([(token, 'Negative') for token, _ in d.get('negative', [])])
            return flat_list

        # Apply the flattening function and then explode the lists into separate rows.
        # This is a highly efficient, vectorized way to create the long-format data.
        long_series = token_series.apply(flatten_tokens).explode().dropna()

        if long_series.empty:
            continue

        # Create a DataFrame from the exploded series of tuples.
        long_df = pd.DataFrame(long_series.tolist(), index=long_series.index, columns=['token', 'polarity'])

        # --- Step 2: Map Tokens to Metatopics ---
        # Use the pre-computed map to assign a metatopic ID to each token.
        long_df['metatopic_id'] = long_df['token'].map(token_map).fillna('Other')

        # Map the metatopic IDs to their human-readable labels.
        long_df['metatopic'] = long_df['metatopic_id'].map(id_to_label_map)

        # --- Step 3: Count Occurrences by Metatopic and Polarity ---
        # Use groupby().size() and unstack() to create a pivot table of counts.
        summary_counts = long_df.groupby(['metatopic', 'polarity']).size().unstack(fill_value=0)

        # Ensure both 'Positive' and 'Negative' columns exist, even if one has no counts.
        if 'Positive' not in summary_counts: summary_counts['Positive'] = 0
        if 'Negative' not in summary_counts: summary_counts['Negative'] = 0

        # --- Step 4: Compute Percentage Shares ---
        # Calculate the total number of classified positive and negative tokens (excluding 'Other').
        total_positive = summary_counts.loc[summary_counts.index != 'Other', 'Positive'].sum()
        total_negative = summary_counts.loc[summary_counts.index != 'Other', 'Negative'].sum()

        # Calculate the percentage share for each metatopic for each polarity.
        summary_counts['Positive Share (%)'] = (summary_counts['Positive'] / total_positive * 100) if total_positive > 0 else 0.0
        summary_counts['Negative Share (%)'] = (summary_counts['Negative'] / total_negative * 100) if total_negative > 0 else 0.0

        # Add a 'model' column to identify the source model for later aggregation.
        summary_counts['model'] = model_key

        # Append the summary for this model to the master list.
        all_model_summaries.append(summary_counts.reset_index())

    # --- Final Assembly ---
    if not all_model_summaries:
        return pd.DataFrame()

    # Concatenate the summaries from all models into a single DataFrame.
    final_summary = pd.concat(all_model_summaries, ignore_index=True)

    # Set a multi-index for clear, hierarchical presentation.
    final_summary = final_summary.set_index(['model', 'metatopic'])

    # Reorder columns for the final report.
    final_summary = final_summary[['Positive', 'Negative', 'Positive Share (%)', 'Negative Share (%)']]

    return final_summary

````

````python
# Task 18: Create Orchestrator Function for Market-Efficiency Trading Simulation

# ==============================================================================
# Custom Type Aliases
# ==============================================================================

Portfolio = pd.DataFrame
BooleanMask = pd.Series

# ==============================================================================
# Task 18, Step 1: Signal-Agreement Filtering Function
# ==============================================================================

def filter_by_signal_agreement(
    df: pd.DataFrame,
    surprise_col: str,
    soft_col: str
) -> BooleanMask:
    """
    Creates a boolean mask to filter for events where hard and soft signals agree.

    This function identifies events where the signs of the earnings surprise
    (hard signal) and the textual soft score (soft signal) are identical and
    non-zero. This is a core condition for the high-conviction trading strategy.

    Equation: mask_{c,t} = I(sign(Surprise_{c,t}) == sign(Soft_{c,t}))
              where Surprise and Soft are non-zero.

    Args:
        df: The input DataFrame of announcement events.
        surprise_col: The name of the column containing the earnings surprise.
        soft_col: The name of the column containing the soft signal score.

    Returns:
        A pandas Series of booleans (mask) that is True for rows where
        the signals agree and False otherwise.
    """
    # --- Input Validation ---
    if surprise_col not in df.columns or soft_col not in df.columns:
        raise KeyError("One or both specified signal columns are not in the DataFrame.")

    # --- Compute Signs ---
    # Use np.sign to get the sign of each signal (-1, 0, or 1).
    # This function correctly handles NaNs, which will result in False.
    sign_surprise = np.sign(df[surprise_col])
    sign_soft = np.sign(df[soft_col])

    # --- Create Agreement Mask ---
    # The signals agree if their signs are equal.
    agreement_mask = (sign_surprise == sign_soft)

    # Exclude neutral cases where the sign is 0 for either signal.
    non_zero_mask = (sign_surprise != 0) & (sign_soft != 0)

    # The final mask requires both agreement and non-zero signals.
    return agreement_mask & non_zero_mask

# ==============================================================================
# Task 18, Step 2: Liquidity Filtering Function
# ==============================================================================

def filter_by_liquidity(
    df: pd.DataFrame,
    max_rel_spread: float
) -> BooleanMask:
    """
    Creates a boolean mask to filter for events with sufficient liquidity.

    This function identifies events that meet two criteria at the 9:45 AM
    market snapshot: the quote must be valid (not locked/crossed), and the
    relative bid-ask spread must be below a specified threshold. This avoids
    trading illiquid stocks with prohibitive transaction costs.

    Equation: mask_{c,t} = I(rel_spread_{c,t} < max) AND I(is_valid_{c,t})

    Args:
        df: The input DataFrame of announcement events.
        max_rel_spread: The maximum allowable relative spread (e.g., 0.20).

    Returns:
        A pandas Series of booleans (mask) that is True for rows that meet
        the liquidity criteria.
    """
    # --- Input Validation ---
    required_cols = ['nbbo_945_rel_spread', 'nbbo_945_is_valid']
    if not all(col in df.columns for col in required_cols):
        raise KeyError(f"DataFrame is missing one or more required liquidity columns: {required_cols}.")

    # --- Create Spread and Validity Masks ---
    # Mask for events where the relative spread is below the maximum threshold.
    spread_mask = df['nbbo_945_rel_spread'] < max_rel_spread

    # Mask for events where the 9:45 AM quote is marked as valid.
    valid_quote_mask = df['nbbo_945_is_valid'] == True

    # The final mask requires both conditions to be met.
    # The .fillna(False) ensures that any rows with missing liquidity data are excluded.
    return (spread_mask & valid_quote_mask).fillna(False)

# ==============================================================================
# Task 18, Step 3: Portfolio Formation and Weighting Function
# ==============================================================================

def form_long_short_portfolio(
    daily_df: pd.DataFrame,
    surprise_col: str,
    soft_col: str,
    rank_col: str,
    weighting_col: str
) -> Tuple[Portfolio, Portfolio]:
    """
    Forms a balanced, market-cap-weighted long-short portfolio for a single day.

    This function implements the core portfolio construction logic:
    1. Partitions events into long (positive signals) and short (negative signals) candidates.
    2. Balances the portfolio by selecting min(|L|, |S|) positions for each side.
    3. Ranks candidates by the specified signal and selects the top N.
    4. Computes market-capitalization weights for the selected positions.

    Args:
        daily_df: A DataFrame containing all tradable events for a single day.
        surprise_col: The name of the earnings surprise column.
        soft_col: The name of the soft signal column.
        rank_col: The name of the column to use for ranking candidates.
        weighting_col: The name of the market cap column for weighting.

    Returns:
        A tuple of two DataFrames: (long_portfolio, short_portfolio).
        Each DataFrame includes the original data plus a 'weight' column.
    """
    # --- Step 1: Partition into Long/Short Candidates ---
    # Long candidates have positive agreement.
    long_candidates = daily_df[(daily_df[surprise_col] > 0) & (daily_df[soft_col] > 0)]

    # Short candidates have negative agreement.
    short_candidates = daily_df[(daily_df[surprise_col] < 0) & (daily_df[soft_col] < 0)]

    # --- Step 2: Balance the Portfolio ---
    # Determine the number of positions based on the smaller side.
    n_positions = min(len(long_candidates), len(short_candidates))

    # If no balanced portfolio can be formed, return empty DataFrames.
    if n_positions == 0:
        return pd.DataFrame(), pd.DataFrame()

    # --- Step 3: Rank and Select ---
    # For the long side, rank by the signal in descending order and take the top N.
    long_portfolio = long_candidates.nlargest(n_positions, rank_col).copy()

    # For the short side, rank by the signal in ascending order (most negative) and take the top N.
    short_portfolio = short_candidates.nsmallest(n_positions, rank_col).copy()

    # --- Step 4: Compute Weights ---
    # Equation: w_{c,t} = MCap_{c,t} / sum(MCap_{j,t} for j in Portfolio)
    # Calculate the total market cap for each portfolio.
    long_total_mcap = long_portfolio[weighting_col].sum()
    short_total_mcap = short_portfolio[weighting_col].sum()

    # Assign weights, handling division by zero if total market cap is zero.
    if long_total_mcap > 0:
        long_portfolio['weight'] = long_portfolio[weighting_col] / long_total_mcap
    else:
        long_portfolio['weight'] = 1 / n_positions # Fallback to equal weight

    if short_total_mcap > 0:
        short_portfolio['weight'] = short_portfolio[weighting_col] / short_total_mcap
    else:
        short_portfolio['weight'] = 1 / n_positions # Fallback to equal weight

    return long_portfolio, short_portfolio

# ==============================================================================
# Task 18: Orchestrator / Factory Function
# ==============================================================================

def create_trading_simulation_orchestrator() -> Dict[str, Callable]:
    """
    Creates and returns a dictionary of the core portfolio construction functions.

    This function acts as a factory, providing a structured collection of the
    validated and documented helper functions required to run the market
    efficiency trading simulation. This promotes a clean separation of concerns
    between the definition of the logic (Task 18) and its execution in a
    time-series loop (Task 19).

    Returns:
        A dictionary of callable functions for filtering and portfolio formation.
    """
    # This orchestrator assembles the building blocks for the next task.
    # It returns the functions themselves, ready to be used in a simulation loop.
    return {
        "filter_by_signal_agreement": filter_by_signal_agreement,
        "filter_by_liquidity": filter_by_liquidity,
        "form_long_short_portfolio": form_long_short_portfolio,
    }
````

````python
# Task 19: Execute Market-Efficiency Trading Simulation

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

Portfolio = pd.DataFrame
BooleanMask = pd.Series
AlphaResults = Dict[str, Dict[str, float]]

# ==============================================================================
# Task 19: Orchestrator Function
# ==============================================================================

def execute_market_efficiency_simulation(
    events_df: pd.DataFrame,
    simulation_helpers: Dict[str, Callable],
    config: Dict[str, Any]
) -> AlphaResults:
    """
    Executes the market efficiency trading simulation and estimates alphas.

    This function simulates a daily long-short trading strategy based on the
    agreement between hard (surprise) and soft (text) signals. It tests two
    return calculation methods and two ranking criteria, and then runs alpha
    regressions on the resulting portfolio return time series to test for
    abnormal profits, thereby testing the semi-strong form of the Efficient
    Market Hypothesis.

    Args:
        events_df: The fully prepared DataFrame with all features and signals.
        simulation_helpers: A dictionary of helper functions from Task 18.
        config: The validated study configuration dictionary.

    Returns:
        A dictionary containing the alpha regression results (alpha, t-stat,
        p-value) for each simulated strategy.
    """
    # --- Step 0: Initialization ---
    # Extract helper functions and configuration parameters.
    filter_agree = simulation_helpers['filter_by_signal_agreement']
    filter_liquid = simulation_helpers['filter_by_liquidity']
    form_portfolio = simulation_helpers['form_long_short_portfolio']

    sim_params = config['study_config']['trading_simulation_params']
    max_spread = sim_params['max_relative_spread_entry']

    # Initialize a list to store daily return records.
    daily_returns_records = []

    # --- Step 1: Daily Portfolio Formation and Return Calculation Loop ---
    # Group the DataFrame by the announcement trading date to process each day.
    for date, daily_events in events_df.groupby('ann_trade_date'):
        # Apply signal agreement and liquidity filters.
        agree_mask = filter_agree(daily_events, 'surprise_winsorized', 'soft_mean')
        liquid_mask = filter_liquid(daily_events, max_spread)
        tradable_events = daily_events[agree_mask & liquid_mask]

        # Process for each ranking method ('surprise' and 'soft').
        for rank_method in ['surprise', 'soft']:
            rank_col = 'surprise_winsorized' if rank_method == 'surprise' else 'soft_mean'

            # Form the long and short portfolios for the day.
            long_pf, short_pf = form_portfolio(
                tradable_events, 'surprise_winsorized', 'soft_mean', rank_col, 'market_cap_t_minus_1'
            )

            if long_pf.empty or short_pf.empty:
                continue

            # --- Step 2: Compute Returns for Both Strategies ---
            # Strategy 1 (OC1): Entry at 9:45 AM quote, exit at 4:00 PM close price.
            long_pf['ret_oc1'] = (long_pf['price_ann_close_adj'] / long_pf['nbbo_945_ask']) - 1
            short_pf['ret_oc1'] = (short_pf['price_ann_close_adj'] / short_pf['nbbo_945_bid']) - 1

            # Strategy 2 (OC2): Entry at 9:45 AM quote, exit at 4:00 PM quote.
            long_pf['ret_oc2'] = (long_pf['nbbo_close_bid'] / long_pf['nbbo_945_ask']) - 1
            short_pf['ret_oc2'] = (short_pf['nbbo_close_ask'] / short_pf['nbbo_945_bid']) - 1

            # Calculate weighted portfolio returns for each strategy.
            ret_long_oc1 = (long_pf['ret_oc1'] * long_pf['weight']).sum()
            ret_short_oc1 = (short_pf['ret_oc1'] * short_pf['weight']).sum()
            ret_long_oc2 = (long_pf['ret_oc2'] * long_pf['weight']).sum()
            ret_short_oc2 = (short_pf['ret_oc2'] * short_pf['weight']).sum()

            # Compute the daily long-short return. LS = R_long - R_short.
            ls_return_oc1 = ret_long_oc1 - ret_short_oc1
            ls_return_oc2 = ret_long_oc2 - ret_short_oc2

            # Store the daily record.
            daily_returns_records.append({
                'date': date,
                'rank_method': rank_method,
                'ls_return_oc1': ls_return_oc1,
                'ls_return_oc2': ls_return_oc2
            })

    # --- Step 3: Aggregate Daily Returns and Estimate Alphas ---
    # Convert the list of records into a DataFrame.
    returns_df = pd.DataFrame(daily_returns_records).set_index(['date', 'rank_method'])

    # Prepare factor data by taking the daily values from the original DataFrame.
    factor_cols = ['ff_mkt_rf', 'ff_smb', 'ff_hml']
    factor_df = events_df.groupby('ann_trade_date')[factor_cols].first()

    # Initialize dictionary to store final alpha results.
    alpha_results: AlphaResults = {}

    # Iterate through each unique strategy configuration to run regressions.
    for rank_method in returns_df.index.get_level_values('rank_method').unique():
        for strat_key in ['ls_return_oc1', 'ls_return_oc2']:
            # Select the return series for the current strategy.
            ls_returns = returns_df.xs(rank_method, level='rank_method')[strat_key].dropna()

            # Merge with factor data.
            regression_data = pd.merge(ls_returns.rename('ls_return'), factor_df, left_index=True, right_index=True)

            # Define dependent and independent variables.
            y = regression_data['ls_return']
            X_ff3 = sm.add_constant(regression_data[factor_cols])
            X_intercept = sm.add_constant(pd.Series(1.0, index=y.index, name='const'))

            # Estimate Alpha-only and Fama-French 3-Factor models.
            for model_type, X in [('alpha_only', X_intercept), ('ff3_alpha', X_ff3)]:
                # Fit the OLS model.
                model = sm.OLS(y, X).fit(cov_type='HAC', cov_kwds={'maxlags': 5}) # Newey-West standard errors

                # Store the key results (alpha, t-stat, p-value).
                result_key = f"strategy_{strat_key.split('_')[-1]}_rank_{rank_method}_{model_type}"
                alpha_results[result_key] = {
                    'alpha_bps': model.params['const'] * 10000, # Alpha in basis points
                    't_stat': model.tvalues['const'],
                    'p_value': model.pvalues['const']
                }

    return alpha_results
````

````python
# Task 20: Create Orchestrator Function for Hacking Scenario (Precision@k)

# ==============================================================================
# Custom Type Aliases
# ==============================================================================

BooleanMask = pd.Series
PrecisionDict = Dict[int, float]

# ==============================================================================
# Task 20, Step 1: High-Volume Announcement Day Filter
# ==============================================================================

def filter_high_announcement_days(
    df: pd.DataFrame,
    min_count: int
) -> BooleanMask:
    """
    Creates a boolean mask to identify trading days with a sufficient number of
    announcements.

    This function is used to focus the Precision@k analysis on days where there
    is a meaningful cross-section of events to rank. It uses a highly efficient
    `groupby().transform('size')` operation to avoid slow iteration.

    Args:
        df: The input DataFrame of announcement events, with 'ann_trade_date'
            in its index.
        min_count: The minimum number of announcements a day must have to be
                   included (e.g., 20).

    Returns:
        A pandas Series of booleans (mask) that is True for rows corresponding
        to high-volume announcement days.
    """
    # --- Input Validation ---
    if 'ann_trade_date' not in df.index.names:
        raise ValueError("Input DataFrame must have 'ann_trade_date' in its index.")
    if min_count <= 0:
        raise ValueError("'min_count' must be a positive integer.")

    # --- Compute Daily Counts ---
    # Group by the announcement date.
    # `.transform('size')` broadcasts the count of each group back to the
    # original DataFrame's shape, creating a Series aligned with the index.
    daily_counts = df.groupby('ann_trade_date')['ann_trade_date'].transform('size')

    # --- Create Mask ---
    # The final mask is True where the daily count meets the minimum threshold.
    return daily_counts >= min_count

# ==============================================================================
# Task 20, Step 2: Precision@k Computation Function
# ==============================================================================
# ==============================================================================
# Implementation of compute_precision_at_k
# ==============================================================================

def compute_precision_at_k(
    predicted_scores: pd.Series,
    realized_scores: pd.Series,
    k_values: List[int],
    ascending: bool = False
) -> PrecisionDict:
    """
    Computes Precision@k for a given set of k values, based on raw scores.

    This function measures the effectiveness of a predictive ranking by
    calculating the fraction of its top-k predictions that are also in the
    top-k of the realized outcomes. It encapsulates the entire logic of
    ranking, selecting the top k, and computing the overlap.

    Equation: P@k = |(Top-k Predicted) INTERSECT (Top-k Realized)| / k

    Args:
        predicted_scores: A pandas Series of scores from a predictive model.
        realized_scores: A pandas Series of the true, realized outcomes (e.g.,
                         returns). Must share the same index as predicted_scores.
        k_values: A list of integers representing the 'k's for which to
                  calculate precision (e.g., [1, 5, 10]).
        ascending: A boolean flag to control the sort direction.
                   - False (default): Ranks in descending order (for 'winners').
                   - True: Ranks in ascending order (for 'losers').

    Returns:
        A dictionary mapping each k to its calculated Precision@k score.

    Raises:
        ValueError: If the input Series do not have matching indices.
    """
    # --- Input Validation ---
    if not predicted_scores.index.equals(realized_scores.index):
        raise ValueError("Indices of 'predicted_scores' and 'realized_scores' must match.")
    if not all(isinstance(k, int) and k > 0 for k in k_values):
        raise ValueError("'k_values' must be a list of positive integers.")

    # Initialize a dictionary to store the precision results.
    precision_results: PrecisionDict = {}

    # Determine the correct ranking method based on the 'ascending' flag.
    # For winners (highest scores are best), we use nlargest.
    # For losers (lowest scores are best), we use nsmallest.
    rank_method_pred = predicted_scores.nsmallest if ascending else predicted_scores.nlargest
    rank_method_real = realized_scores.nsmallest if ascending else realized_scores.nlargest

    # Iterate through each value of k to compute precision.
    for k in k_values:
        # Ensure k is not larger than the number of items to rank.
        if k > len(predicted_scores):
            precision_results[k] = float('nan') # Or 0.0, depending on desired behavior for invalid k.
            continue

        # --- Step 1: Identify Top-k Sets ---
        # Get the set of indices for the top-k predicted events.
        predicted_top_k_indices = set(rank_method_pred(k).index)

        # Get the set of indices for the top-k realized outcomes.
        realized_top_k_indices = set(rank_method_real(k).index)

        # --- Step 2: Compute Overlap ---
        # Find the intersection of the two sets. The length of this
        # intersection is the number of correct predictions in the top-k.
        overlap_count = len(predicted_top_k_indices.intersection(realized_top_k_indices))

        # --- Step 3: Calculate Precision@k ---
        # The precision is the overlap count divided by k.
        precision = overlap_count / k

        # Store the result for the current k.
        precision_results[k] = precision

    # Return the dictionary containing all computed precision scores.
    return precision_results

# ==============================================================================
# Task 20, Step 3: Winner/Loser Partitioning Function
# ==============================================================================

def partition_winners_losers(
    df: pd.DataFrame,
    return_col: str
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Partitions a DataFrame into 'winners' (top half) and 'losers' (bottom half)
    based on a return column.

    This function is used to evaluate the Precision@k metric separately for
    predicting the best-performing stocks (winners) and the worst-performing
    stocks (losers).

    Args:
        df: The input DataFrame of announcement events.
        return_col: The name of the column containing the returns to be used
                    for partitioning.

    Returns:
        A tuple of two DataFrames: (winners_df, losers_df).
    """
    # --- Input Validation ---
    if return_col not in df.columns:
        raise KeyError(f"DataFrame is missing the specified return column: '{return_col}'.")

    # --- Calculate Median ---
    # Compute the median return, which will serve as the partition boundary.
    median_return = df[return_col].median()

    # --- Partition DataFrames ---
    # Winners are defined as all events with returns strictly greater than the median.
    winners_df = df[df[return_col] > median_return].copy()

    # Losers are defined as all events with returns less than or equal to the median.
    # This ensures a complete and non-overlapping partition of the data.
    losers_df = df[df[return_col] <= median_return].copy()

    return winners_df, losers_df

# ==============================================================================
# Task 20: Orchestrator / Factory Function
# ==============================================================================

def create_hacking_scenario_orchestrator() -> Dict[str, Callable]:
    """
    Creates and returns a dictionary of the core helper functions for the
    hacking scenario analysis.

    This function acts as a factory, providing a structured collection of the
    validated and documented helper functions required to run the Precision@k
    evaluation in Task 21. This promotes a clean separation of concerns between
    the definition of the logic (Task 20) and its execution (Task 21).

    Returns:
        A dictionary of callable functions for filtering, partitioning, and
        evaluating precision.
    """
    # This orchestrator assembles the building blocks for the next task.
    # It returns the functions themselves, ready to be used in the analysis.
    return {
        "filter_high_announcement_days": filter_high_announcement_days,
        "compute_precision_at_k": compute_precision_at_k,
        "partition_winners_losers": partition_winners_losers,
    }
````

````python
# Task 21: Execute Hacking Scenario (Precision@k Evaluation)

# ==============================================================================
# Custom Type Aliases and Imports
# ==============================================================================

PrecisionDict = Dict[int, float]
AnalysisResults = pd.DataFrame

# ==============================================================================
# Task 21: Helper and Orchestrator Functions
# ==============================================================================

def _evaluate_precision_for_group(
    df_group: pd.DataFrame,
    k_values: List[int],
    is_winners: bool,
    compute_precision_func: Callable[..., PrecisionDict]
) -> Dict[str, PrecisionDict]:
    """
    Computes Precision@k for a specific group (winners or losers) across
    multiple predefined predictors using a simplified, direct approach.

    This function serves as the core evaluation engine for the hacking scenario.
    It iterates through a set of predictive signals ('Surprise', 'Soft') and
    calls a dedicated precision calculation function for each, passing the raw
    scores directly.

    Args:
        df_group: The DataFrame for the specific group to be evaluated (e.g.,
                  all winners or all losers).
        k_values: A list of integers representing the 'k's for which to
                  calculate precision (e.g., [1, 5, 10]).
        is_winners: A boolean flag that is True if the group is 'winners' and
                    False if it is 'losers'. This controls the sort direction.
        compute_precision_func: The remediated `compute_precision_at_k`
                                function that accepts raw score Series and
                                handles ranking internally.

    Returns:
        A dictionary where keys are the predictor names ('Surprise', 'Soft')
        and values are the corresponding Precision@k result dictionaries.
    """
    # --- Input Validation ---
    if not isinstance(df_group, pd.DataFrame):
        raise TypeError("Input 'df_group' must be a pandas DataFrame.")
    required_cols = ['ret_announce_cc', 'surprise_winsorized', 'soft_mean']
    if not all(col in df_group.columns for col in required_cols):
        raise KeyError(f"DataFrame is missing one or more required columns: {required_cols}.")

    # Initialize a dictionary to store the precision results for this group.
    group_results: Dict[str, PrecisionDict] = {}

    # Define the ranking direction based on the group type.
    # For winners (is_winners=True), we want the highest scores, so we sort descending (ascending=False).
    # For losers (is_winners=False), we want the lowest scores, so we sort ascending (ascending=True).
    sort_ascending = not is_winners

    # Define the predictive signals to be evaluated.
    predictors = {'Surprise': 'surprise_winsorized', 'Soft': 'soft_mean'}

    # Define the series of realized outcomes (ground truth).
    realized_scores = df_group['ret_announce_cc']

    # --- Evaluate Each Predictive Signal ---
    # Iterate through each predictor to calculate its Precision@k.
    for predictor_name, col_name in predictors.items():
        # Select the series of predicted scores for the current predictor.
        predicted_scores = df_group[col_name]

        # Call the precision function directly with the raw score series.
        # The function now encapsulates all ranking and comparison logic.
        group_results[predictor_name] = compute_precision_func(
            predicted_scores=predicted_scores,
            realized_scores=realized_scores,
            k_values=k_values,
            ascending=sort_ascending
        )

    # Return the dictionary containing the results for all predictors for this group.
    return group_results

# ==============================================================================
# Task 21 Helper Function
# ==============================================================================

def _evaluate_precision_for_group(
    df_group: pd.DataFrame,
    k_values: List[int],
    is_winners: bool,
    compute_precision_func: Callable[[Dict[int, Set], Dict[int, Set]], PrecisionDict]
) -> Dict[str, PrecisionDict]:
    """
    Computes Precision@k for a specific group (winners or losers) across
    multiple predefined predictors.

    This function serves as the core evaluation engine for the hacking scenario.
    It takes a DataFrame subset (e.g., all "winners"), determines the true
    top-k stocks based on realized returns, and then measures how well different
    predictive signals ('Surprise' and 'Soft') were able to identify those
    top-k stocks.

    Args:
        df_group: The DataFrame for the specific group to be evaluated (e.g.,
                  all winners or all losers).
        k_values: A list of integers representing the 'k's for which to
                  calculate precision (e.g., [1, 5, 10]).
        is_winners: A boolean flag that is True if the group is 'winners' and
                    False if it is 'losers'. This controls the sort direction.
        compute_precision_func: A callable (e.g., the `compute_precision_at_k`
                                function from Task 20) that takes two
                                dictionaries of top-k index sets and returns
                                a dictionary of precision scores.

    Returns:
        A dictionary where keys are the predictor names ('Surprise', 'Soft')
        and values are the corresponding Precision@k result dictionaries.
    """
    # --- Input Validation ---
    if not isinstance(df_group, pd.DataFrame):
        raise TypeError("Input 'df_group' must be a pandas DataFrame.")
    required_cols = ['ret_announce_cc', 'surprise_winsorized', 'soft_mean']
    if not all(col in df_group.columns for col in required_cols):
        raise KeyError(f"DataFrame is missing one or more required columns: {required_cols}.")

    # Initialize a dictionary to store the precision results for this group.
    group_results: Dict[str, PrecisionDict] = {}

    # Define the ranking direction based on the group type.
    # For winners (is_winners=True), we want the highest returns, so we sort descending (ascending=False).
    # For losers (is_winners=False), we want the lowest returns, so we sort ascending (ascending=True).
    sort_ascending = not is_winners

    # --- Step 1: Determine the "Ground Truth" Top-k Events ---
    # For each k, identify the set of indices of the actual top-k performing stocks
    # based on the realized announcement day return.
    realized_top_k_indices: Dict[int, Set] = {
        k: set(df_group.sort_values('ret_announce_cc', ascending=sort_ascending).head(k).index)
        for k in k_values
    }

    # --- Step 2: Evaluate Each Predictive Signal ---
    # Define the predictive signals to be evaluated.
    predictors = {'Surprise': 'surprise_winsorized', 'Soft': 'soft_mean'}

    # Iterate through each predictor to calculate its Precision@k.
    for predictor_name, col_name in predictors.items():
        # For each k, identify the set of indices of the top-k stocks as ranked
        # by the current predictive signal.
        predicted_top_k_indices: Dict[int, Set] = {
            k: set(df_group.sort_values(col_name, ascending=sort_ascending).head(k).index)
            for k in k_values
        }

        # Call the provided precision function to compute the overlap between
        # the predicted and realized top-k sets for all values of k.
        group_results[predictor_name] = compute_precision_func(
            predicted_top_k_indices, realized_top_k_indices
        )

    # Return the dictionary containing the results for all predictors for this group.
    return group_results

def execute_hacking_scenario_evaluation(
    events_df: pd.DataFrame,
    simulation_helpers: Dict[str, Callable],
    config: StudyConfig
) -> AnalysisResults:
    """
    Executes the full "hacking scenario" analysis to evaluate the Precision@k
    of hard, soft, and combined signals.

    This function simulates a scenario of having pre-release information and
    measures how well different signals can identify the top and bottom
    performing stocks on high-volume announcement days. It evaluates three
    distinct prediction strategies:
    1. Ranking by earnings surprise alone.
    2. Ranking by the composite soft signal alone.
    3. Ranking by surprise, but only on the subset of stocks where the hard
       and soft signals agree.

    Args:
        events_df: The fully prepared DataFrame with all features and signals.
        simulation_helpers: A dictionary of remediated helper functions from
                            Tasks 18 & 20.
        config: The validated study configuration object.

    Returns:
        A multi-indexed DataFrame summarizing the Precision@k results for
        winners and losers across the three different prediction strategies.
    """
    # --- Step 0: Initialization and Input Validation ---
    if not isinstance(events_df, pd.DataFrame):
        raise TypeError("Input 'events_df' must be a pandas DataFrame.")
    if not isinstance(config, StudyConfig):
        raise TypeError("Input 'config' must be a validated StudyConfig object.")

    # Extract helper functions and parameters from inputs.
    filter_days = simulation_helpers['filter_high_announcement_days']
    filter_agree = simulation_helpers['filter_by_signal_agreement']
    partitioner = simulation_helpers['partition_winners_losers']
    evaluate_group = simulation_helpers['_evaluate_precision_for_group']
    compute_precision = simulation_helpers['compute_precision_at_k']

    hacking_params = config.hacking_simulation_params
    k_values = hacking_params.precision_at_k_values

    # --- Step 1: Filter to the Analysis Samples ---
    # Base Sample: Filter for days with a minimum number of announcements.
    high_volume_mask = filter_days(events_df, hacking_params.min_daily_announcements)
    base_sample = events_df[high_volume_mask].copy()

    # Agreement Sample: Further filter the base sample for signal agreement.
    agreement_mask = filter_agree(base_sample, 'surprise_winsorized', 'soft_mean')
    agreement_sample = base_sample[agreement_mask].copy()

    # --- Step 2: Compute Precision@k for All Scenarios ---
    # Initialize a list to store all result records before creating the final DataFrame.
    results_records: List[Dict[str, Any]] = []

    # --- Scenarios 1 & 2: Individual Signals on Base Sample ---
    # Partition the base sample into winners and losers.
    base_winners, base_losers = partitioner(base_sample, 'ret_announce_cc')

    # Evaluate predictors on the winners group.
    winner_results = evaluate_group(base_winners, k_values, True, compute_precision)
    for predictor, p_at_k in winner_results.items():
        results_records.append({'Group': 'Winners', 'Predictor': predictor, **p_at_k})

    # Evaluate predictors on the losers group.
    loser_results = evaluate_group(base_losers, k_values, False, compute_precision)
    for predictor, p_at_k in loser_results.items():
        results_records.append({'Group': 'Losers', 'Predictor': predictor, **p_at_k})

    # --- Scenario 3: Agreement Signal ---
    # Partition the agreement sample into its own winners and losers.
    agree_winners, agree_losers = partitioner(agreement_sample, 'ret_announce_cc')

    # Evaluate the 'Agreement' strategy on the winners group.
    # The strategy is to rank the agreement subset by the surprise signal.
    if not agree_winners.empty:
        p_at_k_winners = compute_precision(
            predicted_scores=agree_winners['surprise_winsorized'],
            realized_scores=agree_winners['ret_announce_cc'],
            k_values=k_values,
            ascending=False # Winners = descending rank
        )
        results_records.append({'Group': 'Winners', 'Predictor': 'Agreement', **p_at_k_winners})

    # Evaluate the 'Agreement' strategy on the losers group.
    if not agree_losers.empty:
        p_at_k_losers = compute_precision(
            predicted_scores=agree_losers['surprise_winsorized'],
            realized_scores=agree_losers['ret_announce_cc'],
            k_values=k_values,
            ascending=True # Losers = ascending rank
        )
        results_records.append({'Group': 'Losers', 'Predictor': 'Agreement', **p_at_k_losers})

    # --- Step 3: Tabulate and Report Results ---
    if not results_records:
        return pd.DataFrame() # Return empty DataFrame if no results were generated.

    # Convert the list of records into a DataFrame.
    results_df = pd.DataFrame(results_records)

    # Set a multi-index for clear, hierarchical presentation.
    results_df = results_df.set_index(['Group', 'Predictor'])

    # Rename columns for final presentation (e.g., 1 -> 'P@1').
    results_df.columns = [f'P@{k}' for k in results_df.columns]

    # Sort the index for a clean, organized final table.
    results_df = results_df.sort_index()

    return results_df
````

````python
# Task 22: Compute Diagnostic Statistics and Embedding Cosine Similarities

# ==============================================================================
# Custom Type Aliases
# ==============================================================================

StatsSummary = pd.DataFrame
CosineResult = Dict[str, float]
HeatmapData = pd.DataFrame
DiagnosticResults = Dict[str, Any]

# ==============================================================================
# Task 22, Step 1: Dataset Evolution Statistics
# ==============================================================================

def _report_dataset_evolution(
    df: pd.DataFrame,
    raw_df: pd.DataFrame
) -> StatsSummary:
    """
    Generates a yearly summary of dataset characteristics.

    This function replicates the statistics shown in Figure 1 of the paper,
    providing a year-by-year overview of the number of articles, unique firms,
    and average document length.

    Args:
        df: The main, cleaned DataFrame for analysis.
        raw_df: The raw DataFrame before filtering, used to get the initial count.

    Returns:
        A pandas DataFrame indexed by year, with columns for each summary statistic.
    """
    # Create a 'year' column for grouping.
    df['year'] = df.index.get_level_values('ann_trade_date').year
    raw_df['year'] = raw_df.index.get_level_values('ann_trade_date').year

    # Group by year and compute aggregates.
    yearly_stats = df.groupby('year').agg(
        num_articles_cleaned=('doc_length_clean', 'size'),
        num_unique_firms=('permno', pd.Series.nunique),
        avg_doc_length_chars=('doc_length_clean', 'mean')
    )

    # Get the raw article count.
    raw_counts = raw_df.groupby('year').size().rename('num_articles_raw')

    # Combine the statistics into a single DataFrame.
    summary_df = pd.concat([raw_counts, yearly_stats], axis=1).fillna(0)

    # Clean up temporary columns.
    df.drop(columns='year', inplace=True)
    raw_df.drop(columns='year', inplace=True)

    return summary_df.astype(int)

# ==============================================================================
# Task 22, Step 2: Embedding Cosine Similarities
# ==============================================================================

def _compute_embedding_cosine_similarity(
    df: pd.DataFrame,
    model1_key: str,
    model2_key: str
) -> CosineResult:
    """
    Computes the average pairwise cosine similarity between the embeddings of
    two Transformer models.

    Equation: cosine(a, b) = (a . b) / (||a|| * ||b||)

    Args:
        df: The DataFrame containing the embedding columns for both models.
        model1_key: The prefix for the first model's columns (e.g., 'bert').
        model2_key: The prefix for the second model's columns (e.g., 'finbert').

    Returns:
        A dictionary containing the mean and standard deviation of the
        cosine similarities.
    """
    # Get the embedding columns for each model.
    cols1 = [c for c in df.columns if c.startswith(f'{model1_key}_')]
    cols2 = [c for c in df.columns if c.startswith(f'{model2_key}_')]

    if not cols1 or not cols2:
        raise ValueError("Could not find embedding columns for one or both models.")

    # Extract the embeddings as NumPy arrays.
    e1 = df[cols1].values
    e2 = df[cols2].values

    # Compute the dot product for the numerator (vectorized).
    dot_product = np.sum(e1 * e2, axis=1)

    # Compute the L2 norms for the denominator (vectorized).
    norm1 = np.linalg.norm(e1, axis=1)
    norm2 = np.linalg.norm(e2, axis=1)

    # Calculate cosine similarity, handling potential division by zero.
    denominator = norm1 * norm2
    cosine_similarities = np.divide(dot_product, denominator, out=np.zeros_like(dot_product), where=(denominator!=0))

    # Return the mean and standard deviation.
    return {
        'mean': np.mean(cosine_similarities),
        'std': np.std(cosine_similarities)
    }

# ==============================================================================
# Task 22, Step 3: Double-Sort Heatmap Data
# ==============================================================================

def _construct_double_sort_heatmap_data(
    df: pd.DataFrame
) -> HeatmapData:
    """
    Creates a 5x5 grid of mean returns by sorting events into quintiles of
    earnings surprise and soft signal score.

    Args:
        df: The DataFrame containing 'surprise_winsorized', 'soft_mean', and
            'ret_announce_cc'.

    Returns:
        A 5x5 pandas DataFrame where each cell is the mean announcement return
        for that surprise/soft signal quintile.
    """
    # Create categorical columns for the quintiles of each signal.
    df['surprise_quintile'] = pd.qcut(df['surprise_winsorized'], 5, labels=False, duplicates='drop')
    df['soft_quintile'] = pd.qcut(df['soft_mean'], 5, labels=False, duplicates='drop')

    # Use pivot_table to compute the mean return for each of the 25 cells.
    heatmap_df = df.pivot_table(
        values='ret_announce_cc',
        index='surprise_quintile',
        columns='soft_quintile',
        aggfunc='mean'
    )

    # Clean up temporary columns.
    df.drop(columns=['surprise_quintile', 'soft_quintile'], inplace=True)

    return heatmap_df

# ==============================================================================
# Task 22: Orchestrator Function
# ==============================================================================

def compute_diagnostic_statistics(
    events_df: pd.DataFrame,
    raw_events_df: pd.DataFrame
) -> DiagnosticResults:
    """
    Orchestrates the computation of all diagnostic statistics for the study.

    This function runs a suite of validation and exploratory analyses:
    1. Reports the year-by-year evolution of the dataset size and characteristics.
    2. Computes the cosine similarity between different Transformer embeddings
       to understand their relationships.
    3. Generates the data for a double-sort heatmap to visualize the joint
       predictive power of hard and soft signals.

    Args:
        events_df: The main, fully processed DataFrame for analysis.
        raw_events_df: The DataFrame before filtering, used for raw counts.

    Returns:
        A dictionary containing the results of all diagnostic analyses,
        including DataFrames for yearly stats and heatmap data, and a
        dictionary for cosine similarity results.
    """
    # --- Input Validation ---
    if not isinstance(events_df, pd.DataFrame) or not isinstance(raw_events_df, pd.DataFrame):
        raise TypeError("Inputs must be pandas DataFrames.")

    # --- Step 1: Report Dataset Evolution ---
    yearly_stats = _report_dataset_evolution(events_df.copy(), raw_events_df.copy())

    # --- Step 2: Compute Embedding Cosine Similarities ---
    cosine_sims = {
        'bert_vs_finbert': _compute_embedding_cosine_similarity(events_df, 'bert', 'finbert'),
        'bert_vs_mpnet': _compute_embedding_cosine_similarity(events_df, 'bert', 'mpnet')
    }

    # --- Step 3: Construct Double-Sort Heatmap Data ---
    heatmap_data = _construct_double_sort_heatmap_data(events_df.copy())

    # --- Assemble Final Results ---
    diagnostic_results: DiagnosticResults = {
        'yearly_stats': yearly_stats,
        'cosine_similarities': cosine_sims,
        'heatmap_data': heatmap_data
    }

    return diagnostic_results
````

````python
# Task 23: Validate Study Findings Against Reported Results

# ==============================================================================
# Custom Type Aliases
# ==============================================================================

RegressionResults = Dict[str, Dict[str, Any]]
ShapResults = Dict[str, Dict[str, float]]
AlphaResults = Dict[str, Dict[str, float]]
PrecisionResults = pd.DataFrame
ValidationReport = List[str]

# ==============================================================================
# Task 23: Helper and Orchestrator Functions
# ==============================================================================

def _validate_regression_results(
    computed_results: RegressionResults
) -> ValidationReport:
    """
    Validates computed regression results against benchmark values from the paper.

    This function compares key statistics (R-squared, coefficients) from the
    computed baseline regressions against the values reported in Table 1 of the
    source paper. It flags any deviations that exceed predefined tolerance levels.

    Args:
        computed_results: A dictionary containing the detailed results from the
                          baseline regression analysis (Task 9).

    Returns:
        A list of formatted strings constituting a validation report section.
    """
    # Define benchmark values derived from Table 1 of the source paper.
    # A full implementation would include benchmarks for all models.
    benchmarks = {
        'finbert': {
            'r_squared_overall': 0.044,
            'coef_surprise': 0.7579,
            'coef_soft': 0.6006,
        }
    }

    # Initialize the report with a section header.
    report: ValidationReport = ["--- 1. Regression Results (Table 1) Validation ---"]

    # Iterate through each model's benchmark values.
    for model, bench in benchmarks.items():
        # Check if the computed results for the current model exist.
        if model not in computed_results:
            report.append(f"[FAIL] {model}: Results not found.")
            continue

        # Retrieve the computed results for the current model.
        res = computed_results[model]

        # --- Validate R-squared ---
        # Calculate the absolute difference from the benchmark.
        r2_diff = abs(res['r_squared_overall'] - bench['r_squared_overall'])
        # Check if the difference is within the tolerance (0.5 percentage points).
        status = "PASS" if r2_diff <= 0.005 else "FAIL"
        report.append(f"[{status}] {model} R^2: Computed={res['r_squared_overall']:.4f}, Bench={bench['r_squared_overall']:.4f}")

        # --- Validate Surprise Coefficient ---
        # Retrieve the computed coefficient.
        coef_s = res['coefficients']['surprise_winsorized']
        # Calculate the percentage difference from the benchmark.
        coef_diff_s = abs(coef_s - bench['coef_surprise']) / abs(bench['coef_surprise'])
        # Check if the difference is within the tolerance (10%).
        status = "PASS" if coef_diff_s <= 0.10 else "FAIL"
        report.append(f"[{status}] {model} Coef(Surprise): Computed={coef_s:.4f}, Bench={bench['coef_surprise']:.4f}")

        # --- Validate Soft Signal Coefficient ---
        # Retrieve the computed coefficient for the specific soft signal.
        coef_f = res['coefficients'][f'soft_{model}']
        # Calculate the percentage difference from the benchmark.
        coef_diff_f = abs(coef_f - bench['coef_soft']) / abs(bench['coef_soft'])
        # Check if the difference is within the tolerance (10%).
        status = "PASS" if coef_diff_f <= 0.10 else "FAIL"
        report.append(f"[{status}] {model} Coef(Soft): Computed={coef_f:.4f}, Bench={bench['coef_soft']:.4f}")

    return report


def _validate_shap_results(
    computed_results: ShapResults
) -> ValidationReport:
    """
    Validates computed SHAP importance shares against benchmark values.

    This function compares the computed relative importance of the 'surprise'
    and 'soft' signals against the key findings reported in the paper's SHAP
    analysis, particularly for the best-performing FinBERT model.

    Args:
        computed_results: A dictionary containing the SHAP importance results
                          from the analysis in Task 11.

    Returns:
        A list of formatted strings constituting a validation report section.
    """
    # Define benchmark values from the paper's SHAP analysis section.
    benchmarks = {
        'finbert': {'surprise_winsorized': 48.0, 'soft_finbert': 52.0}
    }

    # Initialize the report with a section header.
    report: ValidationReport = ["\n--- 2. SHAP Importance Results Validation ---"]

    # Iterate through each model's benchmark values.
    for model, bench in benchmarks.items():
        # Check if the computed results for the current model exist.
        if model not in computed_results:
            report.append(f"[FAIL] {model}: SHAP results not found.")
            continue

        # Retrieve the computed results.
        res = computed_results[model]

        # --- Validate Surprise SHAP Share ---
        # Retrieve the computed SHAP percentage.
        shap_s = res['surprise_winsorized']
        # Calculate the absolute difference in percentage points.
        diff_s = abs(shap_s - bench['surprise_winsorized'])
        # Check if the difference is within the tolerance (3.0 percentage points).
        status = "PASS" if diff_s <= 3.0 else "FAIL"
        report.append(f"[{status}] {model} SHAP(Surprise): Computed={shap_s:.1f}%, Bench={bench['surprise_winsorized']:.1f}%")

        # --- Validate Soft Signal SHAP Share ---
        # Retrieve the computed SHAP percentage for the specific soft signal.
        shap_f = res[f'soft_{model}']
        # Calculate the absolute difference in percentage points.
        diff_f = abs(shap_f - bench[f'soft_{model}'])
        # Check if the difference is within the tolerance (3.0 percentage points).
        status = "PASS" if diff_f <= 3.0 else "FAIL"
        report.append(f"[{status}] {model} SHAP(Soft): Computed={shap_f:.1f}%, Bench={bench[f'soft_{model}']:.1f}%")

    return report


def _validate_simulation_results(
    alpha_results: AlphaResults,
    precision_results: PrecisionResults
) -> ValidationReport:
    """
    Validates market efficiency (alpha) and hacking scenario (Precision@k) results.

    This function compares the outputs of the two main simulations against the
    key numerical findings reported in the paper to verify the fidelity of the
    entire pipeline.

    Args:
        alpha_results: The dictionary of alpha regression results from Task 19.
        precision_results: The DataFrame of Precision@k results from Task 21.

    Returns:
        A list of formatted strings constituting a validation report section.
    """
    # Initialize the report with a section header.
    report: ValidationReport = ["\n--- 3. Simulation Results Validation ---"]

    # --- Market Efficiency Alpha Validation ---
    # Check a key result: Strategy 2 (quote-to-quote), ranked by soft, alpha-only model.
    key = 'strategy_oc2_rank_soft_alpha_only'
    if key in alpha_results:
        # Retrieve the computed results for this strategy.
        res = alpha_results[key]
        alpha_bps = res['alpha_bps']
        p_val = res['p_value']

        # Check if the computed alpha is within the expected range (~-51 bps).
        status_val = "PASS" if -60.0 <= alpha_bps <= -40.0 else "FAIL"
        # Check if the result is statistically significant at the 1% level.
        status_sig = "PASS" if p_val < 0.01 else "FAIL"

        # Append formatted results to the report.
        report.append(f"[{status_val}] Alpha (Strat 2, Soft): Computed={alpha_bps:.2f} bps, Bench=~-51 bps")
        report.append(f"[{status_sig}] Alpha Sig. (Strat 2, Soft): Computed p={p_val:.4f}, Bench p<0.01")
    else:
        report.append(f"[FAIL] Alpha (Strat 2, Soft): Results not found.")

    # --- Hacking Scenario Precision@k Validation ---
    if not precision_results.empty:
        try:
            # Check P@10 for individual signals (Surprise on Winners).
            p10_surprise = precision_results.loc[('Winners', 'Surprise'), 'P@10']
            # Check if the value is within the reported range.
            status_ind = "PASS" if 0.25 <= p10_surprise <= 0.36 else "FAIL"
            report.append(f"[{status_ind}] P@10 (Surprise): Computed={p10_surprise:.3f}, Bench=[0.25, 0.36]")

            # Check P@10 for the agreement signal (ranked by Surprise).
            p10_agree = precision_results.loc[('Winners', 'Agreement'), 'P@10']
            # Check if the value is within the reported range.
            status_agree = "PASS" if 0.44 <= p10_agree <= 0.52 else "FAIL"
            report.append(f"[{status_agree}] P@10 (Agreement): Computed={p10_agree:.3f}, Bench=[0.44, 0.52]")
        except KeyError:
            # Handle cases where the expected index is not in the results DataFrame.
            report.append("[FAIL] Precision@k: Could not find expected ('Winners', 'Surprise'/'Agreement') index.")
    else:
        report.append("[FAIL] Precision@k: Results DataFrame is empty.")

    return report


def validate_all_study_findings(
    regression_results: RegressionResults,
    shap_results: ShapResults,
    alpha_results: AlphaResults,
    precision_results: PrecisionResults
) -> None:
    """
    Orchestrates the validation of all key study findings against reported
    benchmark values from the source paper.

    This function serves as the final end-to-end check of the replication
    pipeline. It calls specialized validation helpers to compare computed
    results for:
    1. Baseline regression models (coefficients and R-squared).
    2. SHAP feature importances.
    3. Market efficiency simulation (alphas).
    4. Hacking scenario simulation (Precision@k).

    The results are printed to the console as a comprehensive report.

    Args:
        regression_results: The output from `estimate_baseline_regressions`.
        shap_results: The output from `compute_shap_importance_for_baselines`.
        alpha_results: The output from `execute_market_efficiency_simulation`.
        precision_results: The output from `execute_hacking_scenario_evaluation`.
    """
    # --- Input Validation ---
    if not all(isinstance(res, dict) for res in [regression_results, shap_results, alpha_results]):
        raise TypeError("Regression, SHAP, and Alpha results must be dictionaries.")
    if not isinstance(precision_results, pd.DataFrame):
        raise TypeError("Precision results must be a pandas DataFrame.")

    # --- Run All Validation Steps ---
    # Step 1: Validate regression results.
    report1 = _validate_regression_results(regression_results)

    # Step 2: Validate SHAP importance results.
    report2 = _validate_shap_results(shap_results)

    # Step 3: Validate simulation results.
    report3 = _validate_simulation_results(alpha_results, precision_results)

    # --- Assemble and Print Final Report ---
    full_report = "\n".join(report1 + report2 + report3)

    print("="*80)
    print("COMPREHENSIVE VALIDATION REPORT")
    print("="*80)
    print(full_report)
    print("="*80)

    if "FAIL" in full_report:
        print("\nValidation completed with one or more significant deviations.")
    else:
        print("\nValidation completed successfully. All key metrics align with benchmarks.")
````

````python
# Top-Level Orchestrator

# ==============================================================================
# Master Orchestrator Function
# ==============================================================================

def run_full_analysis_pipeline(
    events_df: pd.DataFrame,
    raw_events_df: pd.DataFrame,
    study_params: Dict[str, Any],
    artifacts_path: Path
) -> Dict[str, Any]:
    """
    Executes the complete end-to-end research pipeline for the study "Extracting
    the Structure of Press Releases for Predicting Earnings Announcement Returns".

    This master orchestrator serves as the single entry point to run the entire
    analysis. It manages the full workflow, from initial data validation and
    cleaning, through feature engineering, modeling, interpretation, simulation,
    and final validation. It handles the persistence and loading of intermediate
    artifacts to ensure efficiency, consistency, and reproducibility.

    Args:
        events_df: The primary DataFrame containing the full, unprocessed dataset.
        raw_events_df: A pristine copy of the DataFrame before any filtering.
        study_params: The dictionary containing the complete study configuration.
        artifacts_path: The base directory for saving/loading artifacts.

    Returns:
        A dictionary containing all major results artifacts generated during
        the pipeline execution.

    Raises:
        Exception: Any exception from a sub-task will be caught, logged, and
                   re-raised to halt the pipeline.
    """
    # ==========================================================================
    # SETUP
    # ==========================================================================

    # Configure logging for detailed progress tracking.
    logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s', force=True)

    # Initialize a dictionary to store all final results.
    results: Dict[str, Any] = {}

    # Implement top-level error handling to gracefully manage pipeline failures.
    try:
        # ======================================================================
        # TASKS 1-3: DATA PREPARATION
        # ======================================================================

        # Task 1: Validate all inputs and the configuration schema.
        logging.info("--- Starting Task 1: Validating Inputs and Schema ---")
        df, config = validate_inputs_and_schema(events_df, study_params)
        logging.info("Task 1 complete. Inputs are valid.")

        # Task 2: Apply the study's sample selection filters.
        logging.info("--- Starting Task 2: Applying Sample Filters ---")
        df, filter_report = apply_sample_filters(df, config)
        results['filter_report'] = filter_report
        logging.info(f"Task 2 complete. Final sample size: {len(df)}")

        # Task 3: Cleanse HTML and produce two text versions for different models.
        logging.info("--- Starting Task 3: Cleaning and Standardizing Text Data ---")
        df = cleanse_and_standardize_text_data(df, config)
        logging.info("Task 3 complete.")

        # ======================================================================
        # TASKS 4-6: FEATURE ENGINEERING (VECTORIZATION)
        # ======================================================================

        logging.info("--- Starting Task 4: BKMX Vectorization ---")
        df = vectorize_text_with_bkmx(df, config)
        logging.info("Task 4 complete.")

        logging.info("--- Starting Task 5: Online LDA Vectorization ---")
        df, final_lda_model, final_lda_vectorizer = vectorize_text_with_olda_rolling(df, config)
        logging.info("Task 5 complete.")

        logging.info("--- Starting Task 6: Transformer Embeddings ---")
        df = extract_transformer_embeddings(df, config, artifacts_path)
        logging.info("Task 6 complete.")

        # ======================================================================
        # TASKS 7-11: PREDICTIVE MODELING AND EVALUATION
        # ======================================================================

        logging.info("--- Starting Task 7: Estimating Rolling Soft Scores ---")
        df, yearly_lasso_weights = estimate_rolling_soft_scores(df, config, artifacts_path)
        results['yearly_lasso_weights'] = yearly_lasso_weights
        logging.info("Task 7 complete.")

        logging.info("--- Starting Task 8: Preparing Final Analytical Signals ---")
        df = prepare_final_signals(df, config)
        logging.info("Task 8 complete.")

        logging.info("--- Starting Task 9: Estimating Baseline Regressions ---")
        baseline_reg_results, fitted_baseline_models = estimate_baseline_regressions(df, config)
        results['baseline_regressions'] = baseline_reg_results
        logging.info("Task 9 complete.")

        logging.info("--- Starting Task 10: Estimating Combined Regressions ---")
        combined_reg_results = estimate_combined_regressions(df, config)
        results['combined_regressions'] = combined_reg_results
        logging.info("Task 10 complete.")

        logging.info("--- Starting Task 11: Computing SHAP Importance ---")
        shap_results = compute_shap_importance_for_baselines(df, fitted_baseline_models, config)
        results['shap_importance'] = shap_results
        logging.info("Task 11 complete.")

        # ======================================================================
        # TASKS 12-17: INTERPRETABILITY PIPELINE
        # ======================================================================

        logging.info("--- Starting Task 12: Labeling oLDA Topics via LLM ---")
        labeled_topics = label_lda_topics_with_llm(final_lda_model, final_lda_vectorizer, config)
        results['labeled_topics'] = labeled_topics
        logging.info("Task 12 complete.")

        logging.info("--- Starting Task 13: Condensing Topics into Metatopics via LLM ---")
        metatopic_taxonomy = condense_topics_into_metatopics(labeled_topics, config)
        results['metatopic_taxonomy'] = metatopic_taxonomy
        logging.info("Task 13 complete.")

        logging.info("--- Starting Task 14: Extracting Influential Tokens ---")
        for model_key in config.vectorization_params.transformers.model_names:
            df[f'influential_tokens_{model_key}'] = extract_influential_tokens(df, model_key, artifacts_path)
        logging.info("Task 14 complete.")

        logging.info("--- Starting Task 15: Classifying Influential Tokens via LLM ---")
        token_maps = {}
        for model_key in config.vectorization_params.transformers.model_names:
            token_maps[model_key] = classify_influential_tokens(df[f'influential_tokens_{model_key}'], metatopic_taxonomy, config)
        results['token_to_metatopic_maps'] = token_maps
        logging.info("Task 15 complete.")

        logging.info("--- Starting Task 16: Computing oLDA Metatopic Analytics ---")
        weights_df, variance_df = compute_olda_metatopic_analytics(df, yearly_lasso_weights['olda'], metatopic_taxonomy)
        results['olda_metatopic_weights'] = weights_df
        results['olda_metatopic_variance'] = variance_df
        logging.info("Task 16 complete.")

        logging.info("--- Starting Task 17: Analyzing Token Polarity by Metatopic ---")
        token_polarity_summary = analyze_token_polarity_by_metatopic(df, token_maps, metatopic_taxonomy)
        results['token_polarity_summary'] = token_polarity_summary
        logging.info("Task 17 complete.")

        # ======================================================================
        # TASKS 18-21: SIMULATION PIPELINE
        # ======================================================================

        logging.info("--- Starting Simulation Pipeline (Tasks 18-21) ---")
        trading_helpers = create_trading_simulation_orchestrator()
        hacking_helpers = create_hacking_scenario_orchestrator()

        logging.info("--- Starting Task 19: Executing Market Efficiency Simulation ---")
        alpha_results = execute_market_efficiency_simulation(df, trading_helpers, study_params)
        results['alpha_results'] = alpha_results
        logging.info("Task 19 complete.")

        logging.info("--- Starting Task 21: Executing Hacking Scenario Evaluation ---")
        precision_results = execute_hacking_scenario_evaluation(df, {**trading_helpers, **hacking_helpers}, config)
        results['precision_results'] = precision_results
        logging.info("Task 21 complete.")

        # ======================================================================
        # TASKS 22-23: FINAL DIAGNOSTICS AND VALIDATION
        # ======================================================================

        logging.info("--- Starting Task 22: Computing Diagnostic Statistics ---")
        diagnostic_results = compute_diagnostic_statistics(df, raw_events_df)
        results['diagnostic_stats'] = diagnostic_results
        logging.info("Task 22 complete.")

        logging.info("--- Starting Task 23: Validating All Study Findings ---")
        validation_report = validate_all_study_findings(
            regression_results=baseline_reg_results,
            shap_results=shap_results,
            alpha_results=alpha_results,
            precision_results=precision_results
        )
        results['validation_report'] = validation_report
        logging.info("Task 23 complete.")

        # --- PIPELINE FINISHED ---
        logging.info("====== FULL ANALYSIS PIPELINE FINISHED SUCCESSFULLY ======")

        # Return the comprehensive dictionary of all results.
        return results

    except Exception as e:
        # If any unhandled exception occurs at any point in the pipeline,
        # log it with a full traceback and re-raise to terminate execution.
        logging.critical(f"PIPELINE FAILED: A critical error occurred during execution. Error: {e}", exc_info=True)
        raise e

````
